Growth models, realistic or not, are often represented discretely in time, for example, time ticking by from generation to generation. In practice, however, you experience time as continuous: between today and tomorrow there are an infinite number of individual moments.
An important tool for working with continuous functions and their derivatives is the finite difference method. In it, you approximate the derivative of a function at a point using a difference quotient over a certain time interval.

### Import necessary modules

```python
import numpy as np
import matplotlib.pyplot as plt
```

<div>
    <font color=#690027 markdown="1">  
        <h2>1. Introduction</h2>    </font>
</div>

When modeling, you use functions that convert a given input into an output.
Consider, for example, the function $f$ defined by $f(t) = t sin (t)$. This function maps a real value $t$ to $t sin(t)$, where $t$ is an angle (in radians). In Python you can define this function as follows:

```python
def f(t):
    '''
    Geeft het resultaat van de functie f met als functievoorschrift
    f(t) = t sin(t) terug.

    Parameters
    ----------
    t: float
        de variabele t
    '''

    return t * np.sin(t)
```

Using the following code, you can generate the graph of this function.

```python
# itereer over alle waarden van 0 tot 30 met tussenstappen van 0,01
data_t = np.arange(0, 30.005, 0.01)

# bereken resultaat van functie voor alle beschouwde waarden
data_y = f(data_t)

# maak nieuw grafiekvenster aan
plt.figure()

# plot alle punten
plt.plot(data_t, data_y)

# voeg x-as en y-as toe
plt.hlines(0, -1, 30, color="black")
plt.vlines(0, -30, 30, color="black")

# breng verduidelijkende labels aan
plt.xlabel("t")
plt.ylabel("f(t)")

# voeg titel toe
plt.title("De grafiek van $f$ met $f(t) = t sin(t)$")

# toon grafiek
plt.show()
```

<h2>2. Estimating the derivative at a point</h2>    </font>
</div>

For a continuous function $f$ the difference quotient over the interval $[a, b]$ is equal to:  $$ \frac{f(b)-f(a)}{b-a}.$$Here, $[a, b] \subset dom f$.

For a function $f$ that is differentiable at $a$, it holds:$$ f'(a) = \lim_{x \to a} \frac{f(x)-f(a)}{x-a}.$$Or in other words, for a function $f$ that is differentiable at $x$, it holds:$$ f'(x) = \lim_{\Delta x \to 0} \frac{\Delta y}{\Delta x}.$$

You can use the difference quotient over a given interval to estimate the derivative of a function at a point.$$f'(t) = \frac{\text{d}f(t)}{\text{d}t} \approx \frac{f(t + \Delta t) - f(t)}{\Delta t}.$$
In Python, this finite difference method is relatively easy to implement:

```python
def differentiequotient(f, t, delta_t):
    '''
    Geeft het differentiequotiënt terug van de functie f voor de waarde t,
    waarbij het tijdsinterval delta_t gegeven is.

    Parameters
    ----------
    f: function
        de functie f
    t: float
        de variabele t
    delta_t: float
        het tijdsinterval

    Returns
    -------
    float
        het differentiequotiënt
    '''

    return (f(t + delta_t) - f(t)) / delta_t
```

To estimate the derivative of the function $f$ at $t = 2$ and with $\Delta t = 0$, is it sufficient to make the following function call:

```python
print(differentiequotient(f, 2, 0.1))
```

What happens if you consider a smaller value for $\Delta t$?

```python
print(differentiequotient(f, 2, 0.01))
```

And even smaller?

```python
print(differentiequotient(f, 2, 0.001))
```

<div>
    <font color=#690027 markdown="1">  
        <h2>3. Derivative</h2>    </font>
</div>

When you take smaller steps, the absolute value of the difference $f(t + \Delta t) - f(t)$ typically decreases, but you're also dividing by an ever smaller number. If $f$ is differentiable at $t$, then, when you consider the limit as $\Delta t$ goes to 0, the approximation converges to the derivative $f'(t)$:
$$f'(t) = \lim\limits_{\Delta t \to 0} \frac{f(t + \Delta t) - f(t)}{\Delta t}.$$

        <h3>4.1 Graph of the derivative function</h3>    </font>

Suppose that $f$ is differentiable, i.e., that at every point of $f$ the derivative exists. If you consider the derivative at every point of $f$, then you have determined a new function, namely the derivative function of $f$.
The derivative of $f$ with $f(t) = t sin(t)$ is equal to $f'$ with $f'(t) = t cos(t) + sin(t)$, whose graph is shown below.

```python
def plot_afgeleide():
    '''
    Genereert een grafiek van f'.
    Parameters
    ----------
    delta_t: float
        het tijdsinterval
    '''

    # itereer over alle waarden van 0 tot 30 met tussenstappen van 0,01
    data_x = np.arange(0, 30.005, 0.01)

    # bereken de afgeleide voor alle beschouwde waarden
    data_y_afgeleide = data_x * np.cos(data_x) + np.sin(data_x)

    # maak nieuw grafiekvenster aan
    plt.figure(figsize=(15,10))

    # bereken resultaat van functie voor alle beschouwde waarden 
    data_y = f(data_t)      
    # plot f
    plt.plot(data_t, data_y, color="blue", label="Functie")
    
    # plot de afgeleide
    plt.plot(data_x, data_y_afgeleide, color="orange", label="Afgeleide functie")

    # voeg x-as en y-as toe
    plt.hlines(0, -1, 30, color="black")
    plt.vlines(0, -30, 30, color="black")

    # voeg een legende toe
    plt.legend(loc="upper right", bbox_to_anchor=(1.54, 1.027))

    # breng verduidelijkende labels aan
    plt.xlabel("t")
    plt.ylabel("f\'(t)")

    # voeg een titel toe
    plt.title(f"De grafiek van $f\'$")

    # toon de grafiek
    plt.show()
```

```python
plot_afgeleide()
```

<h3>4.2 Graph of the approximation of the derivative function</h3>    </font>
</div>

If, for example, you now compute the difference quotient for $\Delta t = 1$ at each point of $f$, then you have determined a new function that approximates the derivative function of $f$.
Below, the graph of the derivative of $f$, namely $f'$ with $f'(t) = t cos(t) + sin(t)$, is shown, together with an approximation of it using difference quotients with $\Delta t = 1$.

```python
def plot_afgeleide_en_benadering(delta_t):
    '''
    Genereert een grafiek van f'(t) en een schatting van deze functie met de
    eindige differentiemethode met gegeven delta_t.

    Parameters
    ----------
    delta_t: float
        het tijdsinterval
    '''

    # itereer over alle waarden van 0 tot 30 met tussenstappen van 0,01
    data_x = np.arange(0, 30.005, 0.01)

    # bereken de afgeleide voor alle beschouwde waarden
    data_y = data_x * np.cos(data_x) + np.sin(data_x)

    # maak nieuw grafiekvenster aan
    plt.figure(figsize=(15,10))

    # plot de afgeleide
    plt.plot(data_x, data_y, label="Afgeleide functie")

    # bereken het differentiequotiënt voor delta_t
    data_z = differentiequotient(f, data_x, delta_t)

    # plot de schattingen
    plt.plot(data_x, data_z, label="Benadering")

    # voeg x-as en y-as toe
    plt.hlines(0, -1, 30, color="black")
    plt.vlines(0, -30, 30, color="black")

    # voeg een legende toe
    plt.legend(loc="upper right", bbox_to_anchor=(1.54, 1.027))

    # breng verduidelijkende labels aan
    plt.xlabel("t")
    plt.ylabel("f\'(t)")

    # voeg een titel toe
    plt.title(f"De grafiek van $f\'$, met een schatting voor \
              $\Delta t = {delta_t}$")

    # toon de grafiek
    plt.show()
```

```python
plot_afgeleide_en_benadering(1)
```

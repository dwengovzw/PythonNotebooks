## Correspondence
Heloisa had close contact with the four beneficiaries in her will. In her tower room, several letters were found from each of these people.<br>You are now viewing the contents of some of those letters.

In the next cell, functions are defined that will help you read the texts from the various letters into Python.<br>Execute the following code cells.

```python
def lees_bestanden_in_map(mapnaam):
    """Leest tekst in alle bestanden in map en geeft die teksten terug als een lijst van strings.""" 
    bestanden = os.listdir(mapnaam)
    inhoud = []
    for bestand in bestanden:
        if bestand.endswith(".txt"):
            inhoud.append(lees_bestand(mapnaam + "/" + bestand))
    return inhoud

def overloop_mappen_in_map(mapnaam):
    """Overloopt mappen van auteurs in een map en leest alle teksten in die mappen in."""
    auteurs = {}
    mappen = os.listdir(mapnaam)
    for map in mappen:
        if os.path.isdir(mapnaam + "/" + map):
            auteurs[map] = lees_bestanden_in_map(mapnaam + "/" + map)
    return auteurs
```

## Writing style
Every person has their own writing style. By analyzing this style, you can determine which texts were written by the same author. You apply this technique here.<br>By comparing the writing style of the will with that of the beneficiaries in the will, you can identify a suspect for the murder.To do that, you use a simple technique to determine an author's writing style: you count how often the writer uses certain simple words such as de, het, u, jij, ... in a text. The number of times each of these words occurs in a text by an author is, in a way, a kind of fingerprint of that author.

To determine an author's fingerprint, perform the following steps.
* **Lowercasing:** Replace all uppercase letters in the text with lowercase letters. Lowercasing is necessary because you will count how many times each word occurs and you want to count the word with and without a capital letter in the same way.* **Tokenization:** Split all sentences into meaningful units or 'tokens', such as words and punctuation marks. Because words are separated by a space, this splitting can be done based on the spaces present in the sentences. Remove the punctuation marks first.* **Filtering:** To create a fingerprint of the text, you look only at the use of common words (e.g., de, het, u, uw, je, jij, ...). So you filter out from the text the words that do not occur often.* **Frequency analysis:** You count how often these specific words are used. This gives you a fingerprint of the text that you can compare with those of other texts.* **Normalizing:** To ensure that you can compare texts of different lengths, you normalize the frequency of each word. This means that the number of times a word occurs is divided by the total number of counted words.

## The fingerprint of the will
First determine the fingerprint of the will. Afterwards, compare this fingerprint with the fingerprints of the texts of the various authors.

### Lowercasing
Replace the uppercase letters in the will with lowercase letters.

```python
# laat het testament zien op het scherm
print(testament)

# vervang hoofdletters door kleine letters
testament = testament.lower()

# laat het testament nogmaals zien op het scherm
print(testament)
```

### Tokenization
Split the text into individual words. To do so, first remove the punctuation. Then split the text wherever there is a space.<br>This will give you a list of individual words.

```python
def verwijder_leestekens(tekst):
    """Verwijderen leestekens uit tekst."""
    leestekens = string.punctuation
    tekst_zonder_leestekens = ""
    for karakter in tekst:
        if karakter not in leestekens:
            tekst_zonder_leestekens = tekst_zonder_leestekens + karakter           
    return tekst_zonder_leestekens
```

```python
# verwijder leestekens en splits tekst van testament in woorden
testament = verwijder_leestekens(testament).split()
```

```python
# toon resultaat
print(testament)
```

### Filtering
To create a fingerprint of a text, you use only a fixed set of common words. The code in the next code cell filters each text and keeps only the words that are useful for the fingerprint.

```python
# lijst met woorden die zulllen geteld worden
vingerafdruk_woorden = ["de", "het", "een", "en", "in", "op", "van", "voor", "achter", "onder", "boven", "tussen", "tegen", "met", "zonder", "bij", "door", "naar", "uit", "over", "langs", "rond", "om", "naast", "binnen", "buiten", "ik", "jij", "u", "wij", "jullie", "zij", "mijn", "jouw", "uw", "zijn", "haar", "onze", "hun", "die", "dat", "wie", "wat", "welke", "men"]

# maak nieuwe lijst woorden met woorden van testament die van belang zijn voor vingerafdruk
testament = [woord for woord in testament if woord in vingerafdruk_woorden]
```

```python
# toon resultaat
print(testament)
```

### Frequency analysis
Now that you only have the words left that are relevant to the fingerprint, count how many times each of these words appears in the text.

```python
# bereken frequentie van elk woord in testament
frequentie_testament = {}
for woord in vingerafdruk_woorden:
    frequentie_testament[woord] = testament.count(woord)
```

```python
# toon resultaat
print(frequentie_testament)
```

### Normalizing
Because the number of words you count depends on the length of the text, it is difficult to compare texts of different lengths. In order to compare texts of different lengths, you will normalize the frequency of each word. Normalization is a step that is used very often in AI systems and is necessary to bring inputs to a similar 'level'.<br>Here you normalize the number of times a word occurs by dividing that count by the total number of words that were counted. So you're essentially converting the frequency into a relative frequency.

```python
def normaliseer(frequentie):
    """Normaliseert frequentie van elk woord in tekst."""
    som = sum(frequentie.values())
    for woord in frequentie:
        if som > 0:
            frequentie[woord] = frequentie[woord] / som
        else:
            frequentie[woord] = 0
    return frequentie
```

```python
# normaliseer frequentie van vingerafdrukwoorden in testament
frequentie_testament = normaliseer(frequentie_testament)
```

```python
# toon resultaat
print(frequentie_testament)
```

By plotting these frequencies in a graph, you get a **visual representation of the fingerprint of the will**.

```python
# maak voor het testament een staafdiagram van de frequenties van de woorden
x = np.arange(len(vingerafdruk_woorden))
breedte = 1

fig, ax = plt.subplots(figsize=(20, 10))
rects_testament = ax.bar(x - breedte/2, list(frequentie_testament.values()), breedte, label='Testament')
ax.set_xticks(x)
ax.set_xticklabels(vingerafdruk_woorden, rotation=45, ha="right")
ax.legend()
plt.show()
```

## Apply to the letters
You have now completed all the steps to create a fingerprint of the text in the will. Now apply the same steps to the letters from Heloisa's children and the canon.<br>To do that in an organized way, define a function that determines the fingerprint of a piece of text.

```python
def zet_tekst_om_naar_vingerafdruk(tekst):
    """Bepaalt vingerafdruk van tekst."""
    tekst = tekst.lower() # Lowercasing
    tekst = verwijder_leestekens(tekst).split()   # verwijder leestekens en splits in woorden
    tekst = [woord for woord in tekst if woord in vingerafdruk_woorden] # enkel woorden van belang voor vingerafdruk
    # bereken frequentie van elk woord in tekst
    frequentie = {}
    for woord in vingerafdruk_woorden:
        frequentie[woord] = tekst.count(woord)
    frequentie = normaliseer(frequentie)          # normaliseer frequentie
    return frequentie
```

## Determining distance
To determine which author has a style that matches the style of the will, you compare the fingerprint of each letter with that of the will. You do this by calculating the Euclidean distance between the fingerprint of each letter and that of the will.

To determine the distance, you must first convert **each** letter into a fingerprint. You can do that using the following code.

```python
# overloop alle teksten per auteur en bepaal vingerafdrukken van die teksten 
vingerafdrukken = {}
for auteur in teksten_per_auteur:
    vingerafdrukken[auteur] = []
    for i, tekst in enumerate(teksten_per_auteur[auteur]):
        vingerafdrukken[auteur].append(zet_tekst_om_naar_vingerafdruk(tekst))
```

Now the distance can be calculated between the fingerprint of each letter and that of the will.

```python
# bereken afstand tussen vingerafdruk van elke brief en deze van testament
afstanden = {}
for auteur in vingerafdrukken:
    afstanden[auteur] = []
    for vingerafdruk in vingerafdrukken[auteur]:
        afstand = 0
        for woord in vingerafdruk:
            afstand += abs(vingerafdruk[woord] - frequentie_testament[woord])
        afstanden[auteur].append(afstand)
```

```python
# laat afstanden zien
for auteur in afstanden:
    print(auteur, afstanden[auteur])
```

## Clustering
Based on the distances, it can be difficult to see how far the texts of each author are from each other. To see which texts belong together and which do not, you can use *clustering* techniques. These techniques search for structures in data. Specifically, this technique groups the elements in the dataset that lie close to each other. Here you use the *K-means* clustering technique. With this technique, you must specify in advance how many groups you expect. Because there are texts from five authors here, you choose five groups.

In order to cluster the data, you need to convert it into a proper format. The cell below combines the fingerprint of the testament and of the various authors into a matrix. In addition, the authors' names are stored in a list. With this list, you can check how good the clustering is.

```python
frequenties = []       # matrix van vingerafdrukken
labels = []            # lijst avn auteurs  
frequenties.append(list(frequentie_testament.values()))
labels.append("Testament")
for auteur in vingerafdrukken:
    for frequentie in vingerafdrukken[auteur]:
        frequenties.append(list(frequentie.values()))
        labels.append(auteur)
```

By running the following code cell, the data will be clustered into five groups.

```python
# cluster verschillende teksten van auteurs en testament
kmeans = KMeans(n_clusters=5)
kmeans.fit(frequenties)
```

Now you can iterate over the data points and show which cluster they belong to.

```python
# laat clusters zien
for i, label in enumerate(labels):
    print(f'Tekst: {label}, Cluster: {kmeans.labels_[i]}')
```

You see authors' names: each name corresponds to one of that author's texts.<br>Next to it you can see the grade of the group to which this text belongs.
Note that the result of K-means is not always the same. This is because the algorithm starts with random cluster centroids. The choice of these points therefore affects the final result. It’s a good idea to run the previous two cells a few times to get an idea of how the clustering can vary.

You can also visualize the result of the clustering. To do that on a 2D plane, you need to convert the fingerprints, which currently consist of a sequence of multiple numbers, into 2D coordinates. The technique you use for that is *principal component analysis* (PCA). This technique will therefore convert the fingerprints (vectors of length 45) into 2D coordinates.

```python
# via PCA de vingerafdrukken omzetten naar 2D-coördinaten  
pca = PCA(n_components=2)
pca.fit(frequenties)
frequenties_pca = pca.transform(frequenties)
```

Next, you can display the fingerprints in a scatter plot. At the same time, you can use the cluster numbers to give the points on the plot a color that corresponds to that cluster. This way, you can see which texts belong together according to the 'K-means' algorithm.

```python
# geef de vingerafdrukken weer in een spreidingsdiagram
plt.scatter(frequenties_pca[:, 0], frequenties_pca[:, 1], c=kmeans.labels_)
for i, label in enumerate(labels):
    plt.text(frequenties_pca[i, 0], frequenties_pca[i, 1], label)
plt.show()
```

From these results you can conclude that the writing style of the will most closely resembles that of Elisabeth. Therefore, Elisabeth is our prime suspect.<br>Nevertheless, the writing style of the will is still relatively close to Heloisa's own.
So there is an indication that the will was written by Elisabeth, but we are not yet certain of that. It is also not clear what Elisabeth’s motive would have been. To gain more clarity about Heloisa’s death, you will have to analyze other pieces of evidence.

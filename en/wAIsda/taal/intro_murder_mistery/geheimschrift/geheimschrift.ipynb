#### Replace letters
Now that the letters of *ryvfnorgu* may match the letters of Elisabeth, you can replace the letters of *ryvfnorgu* throughout the letters and see if that yields anything.<br>You do that via the following code cells.

```python
def vervang_letters_in_brieven(brieven, mapping):
    """Vervang alle letters in de brieven volgens de meegegeven mapping."""
    nieuwe_brieven = []
    for brief in brieven:
        nieuwe_brief = ""
        for letter in brief:
            if letter.lower() in mapping:
                nieuwe_brief += mapping[letter.lower()].upper()
            else:
                nieuwe_brief += letter
        nieuwe_brieven.append(nieuwe_brief)
    return nieuwe_brieven
```

To estimate how often a letter occurs on average in the Dutch language, you use a text file containing the contents of a number of Wikipedia articles (*wikipedia.txt*). The code in the next code cell reads the file and calculates how often each letter occurs in the text.

```python
# Lees enkele teksten van op Wikipedia
wikipedia = lees_bestand("wikipedia.txt")

# bepaal frequentie van tekens in de tekst
frequentie_tekens_wikipedia = bepaal_frequentie_tekens_in_tekst(wikipedia)

# sorteer frequenties
frequentie_tekens_wikipedia = dict(sorted(frequentie_tekens_wikipedia.items(), key=lambda item: item[1], reverse=True))

# zet om naar relatieve frequentie
aantal_tekens = sum(frequentie_tekens_wikipedia.values())
for teken in frequentie_tekens_wikipedia:
    frequentie_tekens_wikipedia[teken] /= aantal_tekens/100
    
# laat frequenties zien
print(frequentie_tekens_wikipedia)
```

Now display the letter frequencies in a chart. You can do this by executing the following code cell.

```python
from matplotlib import pyplot as plt
import numpy as np
# maak een staafdiagram van de frequenties van de woorden
x = np.arange(len(frequentie_tekens_wikipedia))
breedte = 1

fig, ax = plt.subplots(figsize=(20, 10))
rects_testament = ax.bar(x - breedte/2, list(frequentie_tekens_wikipedia.values()), breedte, label='Testament')
ax.set_xticks(x)
ax.set_xticklabels(frequentie_tekens_wikipedia, rotation=45, ha="right")
ax.legend()
plt.show()
```

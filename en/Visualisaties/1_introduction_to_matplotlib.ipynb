# Introduction to Data visualisation with Matplotlib

## Objective
This notebook aims to introduce you to Matplotlib, a powerful library for creating static, animated, and interactive visualisations in Python. We will cover basic plots and customization techniques to help you create insightful and aesthetically pleasing charts.


```python
import matplotlib.pyplot as plt
import numpy as np
```

## Basic Plotting

### Line Plot
Let's start with a simple line plot. We will create an array of values for the x-axis and another array for the y-axis using a sine function. Then we will plot these values.


```python
# Sample data
x = np.linspace(0, 10, 100)  # 100 evenly spaced values from 0 to 10
y = np.sin(x)  # Sine of each value in x

# Basic line plot
plt.figure(figsize=(10, 6))  # Set the figure size
plt.plot(x, y)  # Plot x and y
plt.title('Simple Line Plot')  # Add a title
plt.xlabel('X-axis')  # Label for x-axis
plt.ylabel('Y-axis')  # Label for y-axis
plt.show()  # Display the plot
```

### Scatter Plot
Next, we'll create a scatter plot using random data points. Scatter plots are useful for showing relationships between two variables.


```python
# Sample data
x = np.random.rand(50)  # 50 random values for x
y = np.random.rand(50)  # 50 random values for y

# Basic scatter plot
plt.figure(figsize=(10, 6))  # Set the figure size
plt.scatter(x, y)  # Create a scatter plot
plt.title('Simple Scatter Plot')  # Add a title
plt.xlabel('X-axis')  # Label for x-axis
plt.ylabel('Y-axis')  # Label for y-axis
plt.show()  # Display the plot
```

## Customizing Plots

### Titles and Labels
We can add and customize titles, labels, and legends to make our plots more informative and visually appealing.


```python
# Customizing the line plot
plt.figure(figsize=(10, 6))  # Set the figure size
plt.plot(x, y, label='Sine Wave', color='r', linestyle='--')  # Plot with label, color, and linestyle
plt.title('Customized Line Plot')  # Add a title
plt.xlabel('X-axis')  # Label for x-axis
plt.ylabel('Y-axis')  # Label for y-axis
plt.legend()  # Add a legend
plt.grid(True)  # Add a grid
plt.show()  # Display the plot
```

### Colors, Markers, and Line Styles
We can change the colors, markers, and line styles to make our plots stand out and convey more information.


```python
# Using different markers and colors
plt.figure(figsize=(10, 6))  # Set the figure size
plt.plot(x, y, 'o-g', label='Sine Wave')  # Green line with circle markers
plt.title('Line Plot with Markers')  # Add a title
plt.xlabel('X-axis')  # Label for x-axis
plt.ylabel('Y-axis')  # Label for y-axis
plt.legend()  # Add a legend
plt.show()  # Display the plot
```

## Plot Types

### Bar Plots
Bar plots are useful for comparing quantities across different categories. Let's create a simple bar plot.


```python
# Sample data
categories = ['A', 'B', 'C', 'D']  # Categories
values = [3, 7, 5, 4]  # Values for each category

# Bar plot
plt.figure(figsize=(10, 6))  # Set the figure size
plt.bar(categories, values, color='skyblue')  # Create a bar plot
plt.title('Simple Bar Plot')  # Add a title
plt.xlabel('Categories')  # Label for x-axis
plt.ylabel('Values')  # Label for y-axis
plt.show()  # Display the plot
```

### Histograms
Histograms are used to show the distribution of a dataset. Here, we create a histogram with random data.


```python
# Sample data
data = np.random.randn(1000)  # 1000 random values from a normal distribution

# Histogram
plt.figure(figsize=(10, 6))  # Set the figure size
plt.hist(data, bins=30, color='purple', edgecolor='black')  # Create a histogram
plt.title('Histogram')  # Add a title
plt.xlabel('Value')  # Label for x-axis
plt.ylabel('Frequency')  # Label for y-axis
plt.show()  # Display the plot
```

### Pie Charts
Pie charts are used to show proportions of a whole. Let's create a simple pie chart.


```python
# Sample data
sizes = [15, 30, 45, 10]  # Sizes of each slice
labels = ['A', 'B', 'C', 'D']  # Labels for each slice
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']  # Colors for each slice

# Pie chart
plt.figure(figsize=(10, 6))  # Set the figure size
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=140)  # Create a pie chart
plt.title('Simple Pie Chart')  # Add a title
plt.axis('equal')  # Equal aspect ratio ensures the pie is drawn as a circle.
plt.show()  # Display the plot
```

## Subplots
Subplots allow us to display multiple plots in a single figure. This is useful for comparing different plots side by side.


```python
# Sample data
x = np.linspace(0, 10, 100)  # 100 evenly spaced values from 0 to 10
y1 = np.sin(x)  # Sine of each value in x
y2 = np.cos(x)  # Cosine of each value in x

# Subplots
fig, axs = plt.subplots(2, 1, figsize=(10, 10))  # Create 2 subplots, one on top of the other

# First subplot
axs[0].plot(x, y1, 'b')  # Plot sine wave in blue
axs[0].set_title('Sine Wave')  # Add a title
axs[0].set_xlabel('X-axis')  # Label for x-axis
axs[0].set_ylabel('Y-axis')  # Label for y-axis

# Second subplot
axs[1].plot(x, y2, 'r')  # Plot cosine wave in red
axs[1].set_title('Cosine Wave')  # Add a title
axs[1].set_xlabel('X-axis')  # Label for x-axis
axs[1].set_ylabel('Y-axis')  # Label for y-axis

plt.tight_layout()  # Adjust layout to prevent overlap
plt.show()  # Display the plots
```

## Advanced Customization

### Annotations
Annotations can be added to highlight specific points in the plot. This can be useful to draw attention to particular data points.


```python
# Annotating a plot
plt.figure(figsize=(10, 6))  # Set the figure size
plt.plot(x, y, 'b')  # Plot sine wave in blue
plt.title('Annotated Line Plot')  # Add a title
plt.xlabel('X-axis')  # Label for x-axis
plt.ylabel('Y-axis')  # Label for y-axis
plt.annotate('Max', xy=(np.pi/2, 1), xytext=(np.pi/2+1, 1.5),  # Add annotation
             arrowprops=dict(facecolor='black', shrink=0.05))  # Arrow pointing to the annotation
plt.show()  # Display the plot
```

## Interactive Plots (Optional)
Using IPython widgets, we can create interactive plots that allow users to adjust parameters dynamically.


```python
from ipywidgets import interactive  # Import interactive from ipywidgets

def plot_func(frequency=1.0):
    y = np.sin(frequency * x)  # Sine wave with adjustable frequency
    plt.figure(figsize=(10, 6))  # Set the figure size
    plt.plot(x, y)  # Plot the sine wave
    plt.title(f'Sine Wave with Frequency {frequency}')  # Add a title with the frequency
    plt.xlabel('X-axis')  # Label for x-axis
    plt.ylabel('Y-axis')  # Label for y-axis
    plt.show()  # Display the plot

interactive_plot = interactive(plot_func, frequency=(0.1, 2.0))  # Create interactive plot with frequency slider
interactive_plot  # Display the interactive plot
```

## Saving Plots
We can save our plots to files for later use or sharing. This is particularly useful for including plots in reports or presentations.


```python
# Save the plot
plt.figure(figsize=(10, 6))  # Set the figure size
plt.plot(x, y, 'b')  # Plot sine wave in blue
plt.title('Line Plot to Save')  # Add a title
plt.xlabel('X-axis')  # Label for x-axis
plt.ylabel('Y-axis')  # Label for y-axis
plt.savefig('line_plot.png')  # Save the plot as a PNG file
plt.show()  # Display the plot
```

# Matplotlib visualisation exercises

Now that you have a basic idea of how to use the matplotlib library, and how to load, inspect and manipulate data using the pandas library, we will bring these two together to create some interesting visualisations.

In the `visualisations/data` directory, you can find three datasets:
- `most_populated_cities_per_continent.csv`: This dataset contains information about the most populated cities across different continents. It includes the continent, country, city name, and population of each city. The dataset can be used to analyze urban population distribution across continents.
- `revenues_per_music_format.csv`: This dataset records the number of medals won by various countries in different macro areas during the Olympic Games. It includes the year of the Olympics, the macroarea, the country, and the total number of medals won. This dataset is useful for analyzing the performance of countries or regions in the Olympics over time.
- `olympic_medals.csv`: This dataset provides data on the revenues generated by different music formats over the years. It includes the category and format of the music, the year, and the corresponding revenue in millions, both in actual figures and adjusted for inflation. This dataset can be used to study trends in the music industry, particularly how revenue from various formats has changed over time.

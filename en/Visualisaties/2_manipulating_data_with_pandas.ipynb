
# Introduction to data manipulation with Pandas

## Objective
This notebook will introduce you to the basics of the Pandas library in Python. Pandas is a powerful tool for data manipulation and analysis. By the end of this notebook, you will learn how to:
- Load datasets stored as CSV files.
- Inspect and understand the structure of a DataFrame.
- Extract and manipulate data.


```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
```


## Loading the Data

Pandas makes it easy to load datasets from various formats, including CSV files. Let's start by loading a dataset stored in a CSV file.


```python
# Load the data from a CSV file
# For this example, we'll use a hypothetical dataset called 'sample_data.csv'.
df = pd.read_csv('./data/sample_data.csv')
```

```python
# Display the first few rows of the DataFrame to get an overview of the data
df.head()
```


## Inspecting the Data

Once you've loaded a dataset, it's important to inspect it to understand its structure and content. Here are some common tasks you can perform to inspect a DataFrame:
- Check the number of rows and columns.
- View the column names.
- Check the data types of each column.
- Look at the unique values in a specific column.


```python
# Check the number of rows and columns
print("Number of rows and columns:", df.shape)
```

```python
# View the column names
print("Column names:", df.columns)
```

```python
# Check the data types of each column
print("Data types:", df.dtypes)
```

```python
# View the unique values in a specific column (e.g., 'Category')
print("Unique values in 'Category' column:", df['Category'].unique())
```


## Extracting Data

You can extract specific data from a DataFrame by selecting columns, filtering rows, or using conditions. Here are some examples:
- Extracting a single column.
- Extracting multiple columns.
- Filtering rows based on a condition.


```python
# Extract a single column (e.g., 'Category')
category_column = df['Category']
print(category_column.head())
```

```python
# Storing a single column as an array (e.g., 'Category')
category_column = df['Category'].to_numpy()
print(category_column.shape)
```

```python
# Extract multiple columns (e.g., 'Category' and 'Sales')
subset_df = df[['Category', 'Sales']]
print(subset_df.head())
```

```python
# Filter rows based on a condition (e.g., rows where 'Sales' > 100)
filtered_df = df[df['Sales'] > 100]
print(filtered_df.head())
```


## Manipulating Data

Pandas allows you to manipulate data easily. Here are some common operations:
- Calculating the sum or mean of a column.
- Grouping data by a specific column and performing aggregate functions.
- Filtering and selecting data based on multiple conditions.


```python
# Calculate the sum of a column (e.g., 'Sales')
total_sales = df['Sales'].sum()
print("Total Sales:", total_sales)
```

```python
# Calculate the mean of a column (e.g., 'Sales')
average_sales = df['Sales'].mean()
print("Average Sales:", average_sales)
```

```python
# Group by a specific column (e.g., 'Category') and calculate the sum of 'Sales' for each group
sales_by_category = df.groupby('Category')['Sales'].sum()
print(sales_by_category)
```

```python
# Filter rows based on multiple conditions (e.g., 'Sales' > 100 and 'Category' == 'Technology')
filtered_df_2 = df[(df['Sales'] > 100) & (df['Category'] == 'Technology')]
print(filtered_df_2.head())
```

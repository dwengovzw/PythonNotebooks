Sometimes you iterate through a list or a string. Sometimes you want to repeat the same instruction a certain number of times. To avoid having to enter the same instruction or algorithm multiple times, you use a loop.<br>With both a <b>for loop</b> and a <b>while loop</b>, you can add repetition to your script.</div>

<div>
    <font color=#690027 markdown="1">
        <h2>1. For loop</h2>    </font>
</div>

The for loop is used for a *bounded repetition*: an instruction is executed a certain number of times. The number of times the instruction is executed is explicitly stated or is bounded by, e.g., the number of elements in a list.

### Example 1.1

In the following script there are two lists. A list `wiskundigen`, in which the women were forgotten, and a list `wiskundige_vrouwen` with female mathematicians.<br> The script will automatically add all the female mathematicians to the list wiskundigen. <br>The loop is bounded by the number of elements in the list `wiskundige_vrouwen`.<br>Run the Python script. Do you understand it?

```python
# begrensde herhaling

wiskundigen = ["Fermat", "Gauss", "Euler", "Fibonacci", "Galois"]
wiskundige_vrouwen = ["Noether", "Nightingale", "Lovelace"]

for vrouw in wiskundige_vrouwen:
    wiskundigen.append(vrouw)

print(wiskundigen)
```

You have probably already noticed that Python doesn't use `;` or `{}`.Python does use `:`. <br>The use of *indentation* is very specific. In certain places, four spaces are used to indent. This way, the code in the for loop is 'grouped'. <br>The variable `vrouw` refers to an element in the list `wiskundigen`; this variable has not been declared in advance.

<div class="alert alert-block alert-info"> 
The instructions of a loop that must be executed follow after <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">:</span>  and are grouped by indenting. This is called <b> indentation</b>. <br>Where the indentation stops, the loop has been executed once.</div>

### Example 1.2

You can also use a *bounded repetition* where you explicitly state the number of times the instruction is executed. In the following code cell, a for loop is used that runs 20 times.<br> Run the following code.

```python
# begrensde herhaling

getal = 10
print("getal bij de start:", getal)
for i in range(20):                  # lus wordt 20 keer doorlopen; i neemt waarden van 0 t.e.m. 19 aan
    print("i:", i)
    getal = getal + i                # verkorte instructie: getal += i 
    print("\tgetal:", getal)        # \t doet inspringen
```

`range(20)` generates the consecutive natural numbers from 0 up to and including 19.

        <h2>2. While loop</h2>    </font>
</div>

In addition to the for loop, you can also use a while loop as a repetition structure. In a while loop, the number of times the loop is executed is determined by a condition. In that case, you are working with a *conditional repetition*.

```python
# voorwaardelijke herhaling

lijst = ["A", "B", "C", "D", "E", "F", "G", "H", "I"]
getal = 0

# de eerste 6 letters van de lijst afdrukken
while getal < 6:
    print("getal =", getal, "\nletter =" , lijst[getal])      # \n doet overgaan naar nieuwe lijn
    getal +=1
```

Note that the while loop is executed repeatedly as long as the condition is satisfied. However, in order to execute the while loop for the first time, the variable `getal` must already be initialized before the while loop: the variable must be defined and assigned a value; here, that value is 0.

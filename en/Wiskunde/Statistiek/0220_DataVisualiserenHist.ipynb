However, to get a clear picture of how the different **age groups** are represented in Belgium, it is better to present this data with a histogram.<br>You will present this data in an organized manner in 10-year age groups. (By choosing 10 years as the class width, the artificial partitioning is eliminated.)

<div class="alert alert-block alert-success">
In this histogram, the age classes will be on the x-axis and the relative frequency on the y-axis. Moreover, the <b>area of each bar</b> must be equal to the <b>(relative) frequency</b>.    </div>

### Example 2.1Run the following code cell. Do you understand the different instructions?

```python
# histogram van de leeftijdsgroepen met absolute frequenties (m.b.v. matplotlib)

# klassenbreedte = 10

# intervallen
# bins = [min(bevolking["leeftijd"]) + i * klassenbreedte for i in range((max(bevolking["leeftijd"]) - min(bevolking["leeftijd"])) // klassebreedte + 2)]
klassengrenzen = np.arange(0, 111, 10)

# plot histogram
plt.figure(figsize=(15,8))

plt.hist(bevolking["leeftijd"], bins=klassengrenzen, weights=bevolking["aantal"], color="yellowgreen", edgecolor="black")

plt.xlabel("leeftijd", fontsize=20)
plt.ylabel("aantal", fontsize=20)
plt.title("Belgische bevolking volgens leeftijd", fontsize=30)

# plt.xticks(range(min(bevolking["leeftijd"]), max(bevolking["leeftijd"]) + klassebreedte, klassebreedte))
plt.xticks(klassengrenzen)

plt.show()
```

```python
# histogram van de leeftijdsgroepen met absolute frequenties (m.b.v. pandas)

klassengrenzen = np.arange(0, 111, 10)

# plot histogram
plt.figure()

bevolking.hist("leeftijd", bins=klassengrenzen, weights=bevolking["aantal"], color="yellowgreen", edgecolor="black", figsize=(15,8))

plt.xlabel("leeftijd", fontsize=20)
plt.ylabel("aantal", fontsize=20)
plt.title("Belgische bevolking volgens leeftijd", fontsize=30)

plt.xticks(klassengrenzen)

plt.show()
```

You can already see roughly what such a histogram looks like. **But you still need to switch to relative frequencies instead of absolute frequencies.**

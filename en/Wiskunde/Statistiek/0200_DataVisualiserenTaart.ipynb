<h2>2. Pie chart</h2>    </font>
</div>

This was monotonous and perhaps not very pleasant work. Moreover, looking at a series of numbers is not that appealing, and it takes some effort to correctly gauge the proportions. By visualizing the information from the dataset in a chart, a lot of information can quickly become clear.

There are many ways to draw charts with Python. An easy approach in this context is to apply the `plot()` function from the pandas module to an absolute frequency table. <br>You can then use the `kind` parameter to indicate what kind of chart you want to draw. By setting `kind="pie"` you draw, for example, a **pie chart** (**pie chart**).

### Example 2.1

```python
plt.figure()

absolute_frequentie_vastgoed.plot(kind="pie")  # teken cirkeldiagram van de tevredenheid over vastgoedprijzen

plt.show()
```

At a glance, you can now extract quite a lot of information from the data. In this diagram, you can see, for example, that the number of people who answered 'happy' is in the minority.

### Example 2.2

You can further improve the presentation by removing the label on the left and by adding a title at the top. You can also make the figure larger.<br>To do so, assign values to the `title` and `label` parameters of the `plot()` function.

```python
plt.figure(figsize=(8,8))

absolute_frequentie_vastgoed.plot(kind="pie", title="Tevredenheid over vastgoedprijzen", label="")  # Teken hetzelfde cirkeldiagram met titel

plt.show()
```

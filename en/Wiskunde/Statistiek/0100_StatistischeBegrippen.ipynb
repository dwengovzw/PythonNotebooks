Statistics is the science that examines a collection of data in a mathematical way. A **dataset** is such a collection of data. The 'pandas' module provides many useful features for working with datasets using Python.

```python
import pandas as pd  # importeer pandas-module
```

First, 'pandas' allows you to read a dataset into Python. The instruction below loads the dataset `datasetvoorbeeld.csv`, which is stored in the `data` folder - so at the location `data/datasetvoorbeeld.csv` - and refers to it with a variable named `voorbeeld`.

<div class="alert alert-block alert-warning"> 
In the notebook 'Scatter plot - Data from file', it is explained how to work with a CSV file in Python.</div>

```python
voorbeeld = pd.read_csv("data/datasetvoorbeeld.csv")
voorbeeld
```

The dataset is represented in a table, more specifically in the 'pandas' **DataFrame** data structure.<br>A DataFrame consists of rows and columns. Each **element** of the DataFrame is a piece of data.<br>These elements can be integers, real numbers, text, symbols, timestamps ...
The columns of a DataFrame can contain the values of a **sample variable**, such as mass, speed, temperature, or a score. Each data point in a column corresponds to a particular value of the sample variable. The name of a column is called the **key** of the column.<br>A row of a DataFrame contains the values of several sample variables for a given element of the population under study.

<div class="alert alert-box alert-danger">
In statistics, the term 'sample variable' is used. <br>For example: if you measure the speed of a car at 5 points in time, then you have 5 values of the sample variable 'car speed'.    
In a programming language, however, you also talk about a 'variable'. That is, when you want to store something in the computer's memory and refer to it.    
To avoid confusion, in what follows, the term <b>sample variable</b> is always used for the data in a column of a DataFrame and <b>variable</b> in the other case.</div>

<div>
    <font color=#690027 markdown="1">
        <h3>1.2. Size of a dataset</h3>    </font>
</div>

The number of rows in a dataset is called the **size** of a dataset. You can retrieve the size with the `len()` function. That is one of the many possible operations on a dataset.

```python
voorbeeld_grootte = len(voorbeeld)  # sla grootte van dataset voorbeeld op in variabele voorbeeld_grootte
voorbeeld_grootte                   # toon grootte op scherm
```

'pandas' lets you isolate a column from a dataset and perform calculations on it. You use the column's *key* to specify the desired column.

```python
kolom2 = voorbeeld["Kolom 2"]       # zonder Kolom 2 af en verwijs ernaar met variabele kolom2
kolom2                              # toon Kolom 2 op scherm
```

You can also perform various operations on a column. The method `sum()` returns the sum of all values in the column.

```python
kolom2_som = kolom2.sum()  # verwijs met variabele kolom2_som naar som van gegevens uit Kolom 2 
kolom2_som                 # toon som van gegevens uit Kolom 2 op het scherm
```

You can also perform arithmetic operations (+, -, . , ÷) on all data in a column at once.

```python
kolom2_dubbel = 2 * kolom2  # verdubbel alle gegevens uit Kolom 2 en verwijs naar nieuwe kolom met kolom2_dubbel
kolom2_dubbel               # toon kolom2_dubbel op scherm
```

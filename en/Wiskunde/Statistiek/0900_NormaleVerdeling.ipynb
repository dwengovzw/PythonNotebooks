<div>
    <font color=#690027 markdown="1">
        <h2>1. The graph of the normal distribution in Python</h2>    </font>
</div>

The probability density function corresponding to the normal distribution has the following functional form: 
$$f(x) = \frac{1}{\sigma\sqrt{2\pi}}e^{-\frac{1}{2}\frac{{x-\mu}^2}{\sigma^2}}.$$
Here, $\mu$ is the **mean** and $\sigma $ is the **standard deviation**.
The graph is a **bell curve** (the so-called **Gaussian curve**). <br>Depending on the values of the two parameters, mean and standard deviation, you will see a different bell curve.

<div class="alert alert-block alert-success"> 
In a standard normal distribution, the mean is $\mu = 0$ and the standard deviation is $\sigma = 1$.    </div>

The code below draws a graph of a standard normal distribution. Run the code.

### Example 1.1: the graph of the standard normal distribution

```python
# grafiek standaardnormale verdeling
# functie en haar parameters, originelen en beelden
gemiddelde = 0
standaardafwijking = 1

x = np.linspace(-5, 5, 100)                    # x-waarden genereren om grafiek te tekenen
y = norm.pdf(x, gemiddelde, standaardafwijking) # bepaal functiewaarden van normale verdeling met gegeven gemiddelde en standaardafwijking

# grafiek
plt.figure()

plt.plot(x, y)                                   # teken grafiek van  normale verdeling

plt.show()
```

In Python, you can compute the function values of the normal distribution via the `pdf()` function of the 'norm' submodule of the 'scipy.stats' module. This function takes the mean and the standard deviation as parameters.

<div class="alert alert-block alert-info"> 
<b>pdf</b> stands for <b><em>probability density function</em></b>; that is the English term for a <b>probability density function</b>.</div>

    <font color=#690027 markdown="1">
        <h2>2. The cumulative density function of the normal distribution</h2>    </font>
</div>


You can access the **cumulative distribution function** of the normal distribution via the `cdf()` function from the same 'norm' module.

### Example 2.1: the graph of the standard normal distribution

Using the following code cell, plot not only the standard normal distribution but also its cumulative distribution function. Run the code.

```python
# grafiek standaardnormale verdeling en cumulatieve
# functies met parameters, originelen en beelden
gemiddelde = 0
standaardafwijking = 1

x = np.linspace(-5, 5 ,100)                         # x-waarden genereren om grafiek te tekenen
y_pdf = norm.pdf(x, gemiddelde, standaardafwijking)  # bepaalfunctiewaarden van normale verdeling
y_cdf = norm.cdf(x, gemiddelde, standaardafwijking)  # bepaal functiewaarden van cumulatieve dichtheidsfunctie 

# grafiek
plt.figure(figsize=(12,8))

plt.plot(x, y_pdf)   # teken grafiek van normale verdeling
plt.plot(x, y_cdf)   # teken grafiek van cumulatieve dichtheidsfunctie

plt.show()
```

<div class="alert alert-block alert-info"> 
    Similarly, <b>cdf</b> stands for <b><em>cumulative density function</em></b>. That is the English term for a <b>cumulative density function</b>.</div>

<div class="alert alert-block alert-success"> 
The cumulative distribution function is another representation of the probability distribution. The probability that a random variable is less than a given value can be calculated as follows using the cumulative distribution function:    $$\textit{cdf}(a) = P(X < a),$$  In other words, the area between the probability distribution function $f$ and the x-axis over the interval $]-∞,a[$.    
$\textit{cdf}(10) = P(X < 10)$ for example corresponds to the area between the graph of $f$ and the x-axis over the interval $]-∞, 10[$. 
</div>

### Example 2.2

- With the following code cell, you calculate the area between the bell curve and the x-axis.

```python
totale_kans = 2 * norm.cdf(6, gemiddelde, standaardafwijking)
print(totale_kans)
```

- You can also approximate this area numerically via the limit of a sequence that you obtain by shifting the bound further and further toward positive infinity.

```python
for grens in range(6, 50, 2):
    oppervlakte = norm.cdf(grens, gemiddelde, standaardafwijking)
    print(oppervlakte)
```

<b>Slicing</b> is a technique in pandas to divide ordered data, such as a column in a DataFrame here, into pieces. This can be done by following the column’s <b>key</b> with <b>[m:n]</b>. This way, only part of the column is considered: the part from the element with index m up to and including the element with index n-1. You can also omit m or n to slice from the beginning or to the end of the column. <br>A few examples:<ul>
<li><code>kolom[2:7]</code> contains the third through seventh elements of <code>kolom</code>;</li><li><code>kolom[:10]</code> contains the first through tenth elements of <code>kolom</code>;</li><li><code>kolom[10:]</code> contains all elements from the eleventh element up to and including the end of <code>kolom</code>.</li></ul>
<br>You can also use a <b>condition</b>, <b> a logical expression</b>, to filter certain elements from a column using slicing. <br>For example:<ul>
<li><code>kolom[kolom < 7]</code> contains all values from <code>kolom</code> that are less than 7;</li><li><code>kolom[90 < kolom]</code> contains all values from <code>kolom</code> that are greater than 90;</li><li><code>kolom[kolom == "hallo"]</code> contains all values from <code>kolom</code> that are equal to the text "hello".</li>

        <h2>3. Quartiles and interquartile range</h2>    </font>
</div>

<div class="alert alert-box alert-success">
There are other <b>measures of central tendency</b>: the first and third quartiles (the second quartile is equal to the median).    Suppose the numbers are ordered from smallest to largest. Then the median divides the numbers into two groups: the part less than or equal to the median and the part greater than or equal to the median. The <b>first quartile</b> is the median of the first part; the <b>third quartile</b> is the median of the second part. <br><br>The <b>interquartile range</b> is the difference between the first and second quartile and is a <b>measure of dispersion</b>.<br>The <b>range</b> is the difference between the smallest and the largest number and is also a <b>measure of dispersion</b>.    </div>

### Example 3.1You can compute the first and third quartiles using the `quantile()` method. In the example below, you calculate the first quartile of a video's length in minutes.

```python
mrbeast_videoduur_minuten.quantile(0.25)  # eerste kwartiel
```

The method `quantile()` takes a real number as an argument. Here that's `0.25` to indicate that you want the first quartile, that is, the value for which 25% of the other data are smaller.

### Example 4.1

With the pandas module you can also draw a **boxplot**. Run the following code cell.

```python
mrbeast_videoduur_gefilterd_minuten.plot(kind='box')
```

You interpret a box plot as follows:* The green line represents the median.* The top and bottom lines of the rectangle represent the first and third quartiles.* The top and bottom line on the vertical bar represent the minimum and maximum.* The dots are outliers. Pandas automatically detects them and places them at an appropriate distance from the 'box' of the boxplot.

<div class="alert alert-box alert-success">
There are other measures of central tendency and dispersion: the <b>measure of central tendency</b> <b>mode</b> and the <b>measure of dispersion</b> <b>standard deviation</b>.<br>    The <b>mode</b> is the value with the highest (absolute or relative) frequency. If you are working with classes, then the <b>modal class</b> is the class with the highest frequency.<br>The <b>standard deviation</b> is the mean squared deviation with respect to the mean. <br>    
</div>

<div class="alert alert-box alert-info">
Using the pandas module, you calculate the <b>mode</b> with the method <code>mode()</code> and the <b>standard deviation</b> with the method <code>std()</code> (std from the English <em>standard deviation</em>).<br>
    When using the <code>mode()</code> method, pandas automatically chooses the most precise possible division into classes.</div>

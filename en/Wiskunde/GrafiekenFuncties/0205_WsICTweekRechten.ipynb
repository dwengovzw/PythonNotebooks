You can draw an x-axis and y-axis using the `vlines()` and `hlines()` functions from the Matplotlib module.<br>You can also set the calibration and adjust the size of the screen.<br>Run the code cell below.

```python
# ASSENSTELSEL TEKENEN

plt.figure(figsize=(15,10))                      # creëer tekenvenster

plt.hlines(0, -10, 20)                           # plot horizontale rechte van x = -10 tot x = 20 waarvoor y = 0 (x-as)
plt.vlines(0, -10, 10)                           # plot verticale rechte van y = -10 tot y = 10 waarvoor x = 0 (y-as)
 
plt.axis(xmin=-10, xmax=20, ymin=-10, ymax=10)   # vensterinstellingen
plt.xticks(np.arange(-10, 21, step=1))           # labels op de x-as
plt.yticks(np.arange(-10, 11, step=1))           # labels op de y-as

plt.grid(True)                                   # plot rooster

plt.show()                                       # toon figuur
```

A line visually represents the linear relationship between two variables. Therefore, a line is nothing more than a set of points. The equation of a line (that is not parallel to the y-axis) will always be given here in the form $y = ax + b$.<br><br>The diagram above shows the problem you need to tackle, but divided into subproblems (**decomposition**). As shown in the diagram, to be able to draw a straight line, you need points that you will visualize within a drawing window.

### Example 3.1

Consider the line with equation $y = 0.8 \; x - 5$.- To plot the graph of this line, Matplotlib needs x-coordinates provided by a NumPy array.<br>- Matplotlib also needs the corresponding y-coordinates. For this you give the instruction "Calculate the y-coordinate for each x-coordinate, using the equation of the line." In Python code that's `y = 0.8 * x - 5`, provided that the list of x-coordinates has already been entered.
Run the following code cells.

```python
# GRAFIEK VAN RECHTE MET GEGEVEN VERGELIJKING: gegevens invoeren 

x = np.arange(-4, 12, 1)                     # kies x-coördinaten van punten die geplot worden        

# x-coördinaaten laten zien
print ("x = ", x)                           
```

```python
# GRAFIEK VAN RECHTE MET GEGEVEN VERGELIJKING: gegevens invoeren 

x = np.arange(-4, 12, 1)                     # kies x-coördinaten van punten die geplot worden        

y = 0.8 * x - 5                              # overeenkomstige y-coördinaten voor rechte

# coördinaten laten zien
print ("x = ", x)                           
print ("y = ", y)
```

The necessary lists are available! <br> Now all that remains is to draw the graph by combining these lists with what you learned in point 2.<br>You can display the points with the `scatter()` function from the Matplotlib module.

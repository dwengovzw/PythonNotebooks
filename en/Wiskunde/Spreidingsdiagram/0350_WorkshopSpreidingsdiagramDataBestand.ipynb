<div>
    <font color=#690027 markdown="1">
        <h2>1. Import data</h2>    </font>
</div>

You import the necessary modules by running the following code cell.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

The CSV file has already been placed **on the server**. To use the CSV file, however, you must first **load** it into the notebook.<br>So read the file by executing the following code cell:

```python
csv = pd.read_csv("data/punten.csv")                     # inlezen bestand
```

And retrieve the result:

```python
csv
```

If you want to see only the beginning of the table, or only its last rows, you can do so using the methods `head()` and `tail()`.

```python
csv.head()
```

```python
csv.tail()
```

The table has 15 rows and 2 columns. <br>

To plot this data in a graph, you need two NumPy lists: one of x-coordinates and one of y-coordinates. With 'pandas' that's quite straightforward because you can use the table header as a **key** *(key)*.

```python
# data opsplitsen

x = csv["x"]     # hoofding van kolom is sleutel, die speelt rol van index, x-coördinaat komt op de x-as
y = csv["y"]     # y-coördinaat komt op y-as
```

Take a look at what such a list looks like:

```python
print(x)
```

This is still a kind of table. Actually, for x you only need the second column of this table.You can achieve that by using typecasting: by converting this pandas table to a NumPy list, you will get the desired result. **This is a very important step.**

```python
# data omzetten naar NumPy-lijsten

x = np.array(x)
y = np.array(y)
```

```python
# data bekijken

print(x)
print(y)
```

<div>
    <font color=#690027 markdown="1">
        <h2>2. Visualizing data</h2>    </font>
</div>

<div class="alert alert-block alert-warning"> 
    In the notebook <em>'Scatter plot - Enter data manually'</em> you already learned how to create a scatter plot. Feel free to look back if needed.</div>

The range of the x-axis is easy to see because the x's are ordered in the list. <br>For the y's, that's not the case. To determine the range on the y-axis, it's useful to know what the smallest and largest y-value are. You use the methods `min()` and `max()` for that. Because they are methods, they come after the object y in the code, separated by a dot, for example `y.max()`.<br><br>It is not always straightforward to determine the appropriate window settings for a dataset. For a dataset like this, it is still possible to estimate by looking at the data and choosing bounds.<br>For a larger and/or unordered dataset, however, it is best to use a more automated method: Python code that does this for you.<br><br>Go ahead and try this method now!

```python
# minima en maxima bepalen

x_min = x.min()
x_max = x.max()

y_min = y.min()
y_max = y.max()

print(x_min, x_max, y_min, y_max)
```

Do your estimates match what the Python script has found?

Now that you know the bounds of the dataset, you can enter them to create an appropriate plotting window.

<div class="alert alert-block alert-danger">
It is always best not to make misleading charts and therefore be sure to also show 0 on the y-axis. Otherwise, any increases and decreases will be greatly exaggerated.</div>

To make it clear which data are shown on the graph, place a label on the x- and y-axes using the pyplot functions `xlabel()` and `ylabel`.<br>Also give the graph the title 'Scatter plot', and if you find it a bit small, you can of course always use the argument `figsize=(a,b)` to make the graph larger.<br><br>Complete the code and run it.

```python
# oefening

plt.figure()                                       # creëer tekenvenster

plt.axis(xmin=25, xmax=120, ymin=..., ymax=...)    # vensterinstellingen
plt.xticks(np.arange(0, 120, step=10))
plt.yticks(np.arange(..., ..., step=10))

plt.scatter(x, y)                                  # plot punten

plt.title(...)                                     # titel
plt.xlabel("x")                                    # omschrijving bij horizontale as
plt.ylabel("y")                                    # omschrijving bij verticale as    

plt.show()                                         # toon grafiek
```

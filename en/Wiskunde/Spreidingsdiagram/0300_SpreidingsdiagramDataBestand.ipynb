### Example 2.4You can remove elements from a NumPy list by the *index* using the `delete()` function.

The range of the x-axis is easy to see because the x's are ordered in the list. <br>For the y's, this is not the case. To determine the range on the y-axis, it is useful to know what the smallest and the largest y-value are. You use the methods `min()` and `max()` for that. Because they are methods, they appear in the statement after the object y, separated by a dot, for example `y.max()`.<br><br>It is not always straightforward to determine the appropriate window settings for a dataset. For a dataset like this, it is still possible to estimate by looking at the data and choosing limits.<br>For a larger and/or unstructured dataset, you'll be better off using a more automated method: Python code that does this for you.<br><br>Go ahead and try this method now!

```python
# minima en maxima bepalen

x_min = x.min()
x_max = x.max()

y_min = y.min()
y_max = y.max()

print(x_min, x_max, y_min, y_max)
```

When you read a file and want to visualize the data, you will carry out the necessary steps consecutively. You will need to look at the intermediate results from time to time to know how to proceed. For example, you will need to look at the column headers and the minimum and maximum values.<br>Here you'll find all the code listed again.

```python
# data uit bestand inlezen

csv = pd.read_csv("data/punten.csv")                      # inlezen van tabel, je kan zelf de variabelenaam kiezen!
csv                                                       # data tonen, sleutels (keys, hoofding kolommen) bekijken
```

```python
# data uit bestand omzetten naar NumPy-lijsten

x = csv["x"]                                         # 2e kolom naam geven m.b.v. sleutel x 
y = csv["y"]                                         # 3e kolom naam geven m.b.v. sleutel y

x = np.array(x)                                      # NumPy-lijst aanmaken met gewenste data
y = np.array(y)

print(x)                                             # NumPy-lijst weergeven
print(y)
```

When you want to visualize loaded data, you can use the following code:

```python
# minima en maxima bepalen

x_min = x.min()
x_max = x.max()

y_min = y.min()
y_max = y.max()

print(x_min, x_max, y_min, y_max)
```

```python
# grafische voorstelling

plt.figure()                                       # creëer tekenvenster

plt.axis(xmin=..., xmax=..., ymin=..., ymax=...)    # vensterinstellingen
plt.xticks(np.arange(..., ..., step=...))
plt.yticks(np.arange(..., ..., step=...))

plt.scatter(x, y)                                  # plot punten

...                                                # titel en labels

plt.show()                                         # toon grafiek
```

# Combining all features

Up until now, we have only predicted the crop field using a single feature at a time. We can however expect that we can create a better model if we let it use all the available features.
Based on what you've learned in the previous notebooks, try to do this yourself now. The general steps are:

1. Import the necessary classes and functions 
2. Load the data. 
3. Visualise the relationship between the different features (soil_moisture, temperature, sunlight, fertilizer) and the targeted outcome (crop yield) using scatter plots. This allows you to decide whether to use linear regression or polynomial regression (and if the latter, which degree).
4. Split the data into a train and a test set.
5. Fit the selected model.
6. Quantitatively and qualitatively evaluate your model. Use the calculated error metrics to compare this model (with all features) to the models that used only a single feature before.

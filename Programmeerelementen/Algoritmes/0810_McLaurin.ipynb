### Nodige modules importeren

```python
# modules importeren

import numpy as np
import matplotlib.pyplot as plt
```

### Grafiek van de exponentiële functie laten tekenen 

Exponentiële functie met voorschrift $f(x) = e^x$. 

```python
# grafiek van gegeven functie

x = np.arange(-8, 8, 0.1)                        # kies x-coördinaten van punten die geplot worden        
y = np.e**x                                      # overeenkomstige y-coördinaten voor exponentiële functie

plt.figure(figsize=(8,8))                        # creëer figuur

plt.axis(xmin=-4, xmax=4, ymin=-4, ymax=4)       # vensterinstellingen
plt.xticks(np.arange(-4, 4, step=0.5))
plt.yticks(np.arange(-4, 4, step=0.5))

plt.hlines(0, -4, 4, color="black")              # x-as
plt.vlines(0, -4, 4, color="black")              # y-as

plt.grid(True)                                   # plot rooster

plt.plot(x, y, color="green")                    # plot exponentiële functie

plt.title("Oorspronkelijke functie")

plt.show()                                       # toon figuur
```

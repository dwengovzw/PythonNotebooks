ALs je een stelsel manipuleert a.d.h.v. elementaire rij-operaties, dan zal de overeenkomstige matrix mee veranderen. Het doel is om de matrix te diagonaliseren zodat je de oplossingen van het stelsel onmiddellijk kunt aflezen.
</div>

<div>
    <font color=#690027 markdown="1">
        <h2>1. Inleiding</h2> 
    </font>
</div>

Eerst zal je enkele functies definiëren, die overeenkomen met de elementaire rij-operaties, om die (eventueel) achteraf te gebruiken. 

```python
def verwissel(matrix, rij_1, rij_2):
    """Twee rijen van plaats verwisselen."""
    tijdelijk = matrix[rij_1]
    matrix[rij_1] = matrix[rij_2]
    matrix[rij_2] = tijdelijk
```

```python
def deel_door(rij, deler):
    """Deel rij door een getal."""
    for i in range(len(rij)):
        rij[i] /= deler
```

```python
def trek_veelvoud_af(rij_1, rij_2, factor):
    """Trek een veelvoud van een rij af van een andere rij."""
    for i in range(len(rij_1)):
        rij_1[i] -= rij_2[i] * factor
```

<div>
    <font color=#690027 markdown="1">
        <h2>2. Diagonaliseren</h2> 
    </font>
</div>

Beschouw het volgende stelsel van vergelijkingen: 
$$\begin{cases} 3 x + y - 4 z = 3 \\ 3 y - 5 z = -11 \\ x - 5 y - 2 z = 11 \end{cases}$$

De overeenkomstige matrix is: $$\begin{bmatrix} 3 & 1 & -4 & 3 \\ 0 & 3 & -5 & -11 \\ 1 & -5 & -2 & 11 \end{bmatrix}$$

```python
matrix = [[3, 1, -4, 3], [0, 3, -5, -11], [1, -5, -2, 11]]
```

```python
# matrix diagonaliseren

n = len(matrix)

for k in range (0, n):   # k is kolomnummer
    print("kolom", k)
    # zorg dat er een 1 komt op plaats matrix[k][k]
    deel_door(matrix[k], matrix[k][k])
    print(matrix)
    # zorg in kolom k voor nullen op alle rijen, behalve rij k
    for r in range (0, n):
        if r != k:
            trek_veelvoud_af(matrix[r], matrix[k], matrix[r][k])
            print(matrix)  

# oplossing
for r in range (0,n):
    print(matrix[r][n])
```

<div>
    <font color=#690027 markdown="1">
        <h2>3. Functie om te diagonaliseren</h2> 
    </font>
</div>

```python
def diagonaliseren(matrix):
    "Matrix diagonaliseren."
    n = len(matrix)
    for k in range (0, n):   # k is kolomnummer
        deel_door(matrix[k], matrix[k][k])     # zorg dat er een 1 komt op plaats matrix[k][k]
        for r in range (0, n):
            if r != k:
                trek_veelvoud_af(matrix[r], matrix[k], matrix[r][k])   # zorg in kolom k voor nullen op alle rijen, behalve rij k  
    # oplossing
    oplossing = []
    for r in range (0, n):
        oplossing.append(matrix[r][n])
    return oplossing
```

```python
matrix = [[3, 1, -4, 3], [0, 3, -5, -11], [1, -5, -2, 11]]
diagonaliseren(matrix)
```

Geef de oplossing van deze stelsels:
$\begin{cases} x + y + z = 27 \\ x - 2 y = 0 \\ - y + z = 3 \end{cases}$ en $\begin{cases} 3 x - y = 0 \\ y - 2 z = 2 \\ - x + z = 7 \end{cases}$

```python
matrix2 = [
    [ 1,  1, 1, 27],
    [ 1, -2, 0, 0],
    [ 0, -1, 1, 3]
]
diagonaliseren(matrix2)
```

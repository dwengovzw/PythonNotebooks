<div class="alert alert-box alert-success">
Python est un langage orienté objet. Tout est un <b>objet</b> : chaque nombre, chaque morceau de texte... Un objet occupe une certaine place dans la mémoire et a une certaine valeur. Chaque objet a aussi un <b>type</b>, p. ex. <em>entier</em> et <em>chaîne</em>. <br>Dans ce notebook, vous allez découvrir quelques types.<br>Tu apprends comment afficher quelque chose à l’écran en utilisant une <b>instruction print</b> et tu te familiarises avec la <b>conversion de type</b>.</div>

<div>
    <font color=#690027 markdown="1">  
        <h2>1. String, integer</h2>    </font>
</div>

<div class="alert alert-box alert-info">
Le type <b>str</b> est le type de texte. C’est une suite de caractères, comme des lettres, des signes de ponctuation et des espaces. On reconnaît une chaîne aux <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">" "</span>.<br>Le type <b>int</b> est le type d'un nombre entier.</div>

Exécutez à chaque fois les cellules de code.

```python
bloem = "iris"
```

`bloem` est une variable qui fait référence à l'objet "iris". On dit que tant "iris" que la variable `bloem` ont le type *string*.

```python
type(bloem)
```

```python
getal = -3
```

`getal` est une variable qui fait référence à l'objet `-3`. On dit que tant `-3` que la variable `getal` sont de type *int*.

```python
type(getal)
```

<div class="alert alert-box alert-info">
Une variable a le même type que l'objet auquel elle fait référence.<br> Le type s'obtient avec la fonction <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">type()</span>.</div>

<div>
    <font color=#690027 markdown="1">  
        <h2>2. Floating-point number</h2>    </font>
</div>

Outre les nombres entiers, vous pouvez également utiliser des nombres décimaux.
Vous saisissez un nombre à virgule (un nombre sous forme décimale) - vous utilisez pour cela un point au lieu d'une virgule - et vous demandez son type :

```python
tweedegetal = 3.6
type(tweedegetal)
```

<div class="alert alert-box alert-info"> 
Le type <b>float</b> est le type d'un nombre décimal.</div>

<h2>3. La fonction print()</h2>    </font>
</div>

Avec la fonction `print()`, vous pouvez afficher quelque chose à l'écran.

### Exemple 3.1

```python
tekst = "bloemen"
aantal = 20
print(aantal, tekst)
```

<div class="alert alert-box alert-info"> 
Deux objets ont été affichés à l’écran : un entier et une chaîne de caractères. Ceux-ci étaient séparés par un espace.</div>

### Exemple 3.2

```python
tweedetekst = "auto"
derdetekst = "mooie"
vierdetekst = "bus"
```

```python
print(derdetekst, tweedetekst)
```

<div class="alert alert-box alert-info"> 
Deux objets ont été affichés à l'écran : deux chaînes de caractères. Elles étaient séparées par un espace.</div>

```python
print(tweedetekst + vierdetekst)
```

<div class="alert alert-box alert-info"> 
Un objet a été affiché à l'écran : la chaîne <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">autobus</span>. <br> Les deux chaînes, <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">auto</span> et <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">bus</span>, ont en effet d'abord été concaténées en 1 nouvelle chaîne, qui a ensuite été affichée à l'écran. <br><br>Assembler des chaînes pour en former une nouvelle s'appelle <b>concaténer</b>. Pour cela, on utilise l'opérateur <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>+</b></span>.    </div>

```python
print(derdetekst, tweedetekst + vierdetekst)
```

<div class="alert alert-box alert-info"> 
2 objets ont été affichés à l'écran : deux chaînes de caractères, séparées par un espace.</div>

    <font color=#690027 markdown="1">  
        <h2>4. Typecasting</h2>    </font>
</div>

<div class="alert alert-box alert-info"> 
Vous pouvez convertir le type d’un objet en un autre type. Ainsi, vous pouvez par exemple convertir un nombre en chaîne de caractères. <br> Ce concept s’appelle le <b>typecasting</b>.</div>

### Exemple 4.1

```python
gegeven = "23"       # 23 staat tussen aanhalingstekens, dus string
```

```python
type(gegeven)
```

```python
omgevormdestring = int(gegeven)
type(omgevormdestring)
```

```python
print(omgevormdestring)
```

### Exemple 4.2

```python
kommagetal = 20.5
type(kommagetal)
```

```python
omgevormdefloat = str(kommagetal)
type(omgevormdefloat)
```

```python
print(omgevormdefloat)
```

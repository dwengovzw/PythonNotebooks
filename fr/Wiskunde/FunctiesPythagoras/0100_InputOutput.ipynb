<div class="alert alert-box alert-success">
Dans un notebook, vous pouvez demander une saisie à l'utilisateur. Ce notebook vous montre comment faire.</div>

Vous pouvez, à l'aide de la fonction `input()`, demander des informations à l'utilisateur.<br> Lors de l'exécution de cette fonction, une *invite* apparaît à l'écran. Vous pouvez alors saisir quelque chose. En appuyant ensuite sur 'Entrée', vous avez saisi un objet.<br>

<div>
    <font color=#690027 markdown="1">  
        <h2>1. Demander des données en entrée</h2>    </font>
</div>

### Exemple 1.1

Regarde la cellule de code suivante. Lors de l'exécution de l'instruction, `"Tu t'appelles : "` apparaît à l'écran. C'est l'*invite*. <br>Tu tapes quelque chose et appuies sur 'Entrée'.<br> La variable `naam` fait alors référence à l'objet que tu viens de saisir.

Exécutez l'instruction:

```python
naam = input("Jij heet: ")
print("Dag " + naam + "!")
```

<div class="alert alert-block alert-info">
Tu as tapé ton nom et as ainsi fourni une <b>input</b> ou <b>entrée</b>. La salutation qui a ensuite été affichée à l'écran est une <b>output</b> ou <b>sortie</b>.<img src="images/invoerverwerkinguitvoerdwengo.png" alt="Banner" align="left" width="100"/>
</div>

Demandez, à l'aide de la cellule de code suivante, de quel type est `naam`.

```python
type(naam)
```

```python
print(naam)
```

La variable `naam` fait donc référence à un objet de type string.

### Exemple 2.1

Conversion de type: l'instruction `float(schoenmaat)` *convertit* cette chaîne *en un float*. <br> La variable `schoenmaat` fait donc référence, après l'exécution de cette instruction, à un objet de type *float*. <br>Exécutez la cellule de code et consultez le résultat.

La variable `schoenmaat` fait toutefois référence à un objet de type *string*. <br>
C'est un peu embêtant, mais on peut y remédier à l'aide d'une *conversion de type*:L'instruction `float(schoenmaat)` *convertit* cette chaîne de caractères *en un float*. <br> La variable `schoenmaat` fait donc référence, après l'exécution de cette instruction, à un objet de type *float*. <br>Exécutez la cellule de code et consultez le résultat.

```python
schoenmaat = float(schoenmaat)
```

```python
type(schoenmaat)
```

```python
print(schoenmaat)
```

Exécutez la cellule de code suivante.

```python
print("Dag " + naam + "! Jouw schoenmaat is groter dan", schoenmaat-5, "en kleiner dan die van tante Sidonia.")
```

Notez que dans la cellule de code précédente, la fonction `print()` a trois arguments, séparés par un espace dans la sortie.<br> Deux arguments sont de type *string* et l'argument du milieu est de type *float*.

### Exemple 2.2

Vous pouvez également obtenir le même résultat d'une manière plus courte.<br>Exécutez la cellule de code et consultez le résultat.

```python
lengte = int(input("Jouw lengte in centimeter is: "))
print("Jij bent", lengte/100, "meter groot.")
```

La variable `lengte` fait référence à un objet de type entier.<br><br>Désormais, utilisez la méthode courte !

<div class="alert alert-block alert-info">
La fonction <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>input()</b></span> permet de demander des informations à l'utilisateur. L'<b>entrée</b> est stockée par Python sous forme de <b>chaîne de caractères</b>.  </div>

<div class="alert alert-block alert-info">
Convertir explicitement un type en un autre s'appelle la <b>conversion de type</b>. Avec la fonction <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>int()</b></span>, on convertit vers le type entier, avec <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>float()</b></span> vers le type flottant et avec la fonction <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>str()</b></span> on convertit vers le type chaîne de caractères.</div>

<div class="alert alert-block alert-info">
Tu ne peux concaténer une chaîne qu’avec des chaînes. Si tu veux afficher, en plus d’une chaîne, un nombre à l’aide de la <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>print()</b></span>-fonction, alors passe ce nombre comme objet distinct - tu utilises alors une virgule - ou convertis-le en chaîne au moyen d’un transtypage (typecasting). <br> Tu utilises le concept de <b>décomposition</b> de la pensée computationnelle lorsque tu réfléchis aux différents éléments d’une instruction print.<img src="images/decompositiedwengo.png" alt="Banner" align="left" width="100"/>
</div>

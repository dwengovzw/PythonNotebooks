Voici des exemples montrant comment définir vous-même une fonction en Python.

### Doubler

Exécutez les trois cellules de code suivantes. Comprenez-vous ce qui se passe ?

```python
def verdubbelen(x):
    """Verdubbelen."""
    dubbel = 2 * x
    return dubbel
```

```python
verdubbelen(10)
```

```python
verdubbelen(math.pi)
```

<div class="alert alert-box alert-success">
Lors de la rédaction de la définition, faites attention aux <b>deux-points</b> et à l’<b>indentation</b> (le retrait de quatre caractères). La fonction prend un paramètre x en <b>entrée</b> et renvoie le double de x en <b>sortie</b>. <br>Le texte entre les guillemets est une <b>docstring</b> et sert à clarifier à quoi sert la fonction. Une docstring commence par une majuscule et se termine par un point.</div>

### Diminuer de 3

```python
def verminderen(x):
    """Verminderen met 3."""
    resultaat = x - 3
    return resultaat
```

```python
verminderen(3)
```

```python
verminderen(2.5)
```

        <h2>3. Application : fonction pour calculer l'aire d'un rectangle</h2>    </font>
</div>

L'aire d'un rectangle de largeur $b$ et de longueur $l$ est le produit de $l$ et $b$. On passe donc deux paramètres à la fonction `oppervlakterechthoek()`.

```python
def oppervlakterechthoek(l, b):
    """Oppervlakte rechthoek met lengte l en breedte b."""
    oppervlakte = l * b
    return oppervlakte
```

```python
# toepassen
# rechthoek met lengte 7 en breedte 3
oppervlakterechthoek(7, 3)
```

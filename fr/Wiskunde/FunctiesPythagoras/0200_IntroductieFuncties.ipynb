### Périmètre d'un carré

Le périmètre d’un carré de côté $z$ est quatre fois la longueur du côté, donc $4z$. Le périmètre est donc déterminé par la longueur du côté. $z$ est donc passé en paramètre à la fonction `omtrekvierkant()`.

```python
def omtrekvierkant(z):
    """Omtrek vierkant met zijde z."""
    omtrek = 4 * z
    return omtrek
```

```python
# toepassen
# vierkant met zijde 6
omtrekvierkant(6)
```

### Aire du rectangle

La surface d’un rectangle de largeur $b$ et de longueur $l$ est le produit de $l$ et $b$. Deux paramètres sont donc passés à la fonction `oppervlakterechthoek()`.

```python
def oppervlakterechthoek(l, b):
    """Oppervlakte rechthoek met lengte l en breedte b."""
    oppervlakte = l * b
    return oppervlakte
```

```python
# toepassen
# rechthoek met lengte 7 en breedte 3
oppervlakterechthoek(7, 3)
```

### Volume d'une pyramide ou d'un cône en utilisant l'aire de base et la hauteur

Le volume d'une pyramide ou d'un cône est le tiers du produit de l'aire de la base et de la hauteur. Définissez pour cela une fonction.

```python
def inhoudpir(s, h):
    """Inhoud piramide of kegel met oppervlakte grondvlak s en hoogte h."""
    inhoud = s * h /3
    return inhoud
```

Cela signifie que, pour pouvoir utiliser cette formule, tu dois connaître l'aire de la base. Tu peux éventuellement la calculer avec une autre fonction.

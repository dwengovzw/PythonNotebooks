### Exercice 1.1.1

Écrivez un script dans lequel les longueurs des deux côtés de l’angle droit d’un triangle rectangle sont données. Le script calcule la longueur de l’hypoténuse. Le programme se termine par l’affichage de la longueur de l’hypoténuse.<br><br>
Testez le script et améliorez-le si nécessaire.

```python
# gegeven
a = 3                            # eerste rechthoekszijde
b = 4                            # tweede rechthoekszijde

# verwerking
c = math.sqrt(a**2 + b**2)       # schuine zijde berekenen

# uitvoer
print("De schuine zijde van de driehoek is:", c)       # uitvoer
```

### Exercice 1.1.2

Écrivez un script dans lequel vous définissez une **fonction** `pythagoras`. <br>Les longueurs des deux côtés de l’angle droit d’un triangle rectangle sont données. Via la fonction `pythagoras`, le script calculera la longueur de l’hypoténuse. Le programme se termine par l’affichage de la longueur de l’hypoténuse.<br><br>
Testez le script et améliorez-le si nécessaire.

```python
def pythagoras(a, b):
    """Schuine zijde berekenen van rechthoekige driehoek met rechthoekszijden a en b."""
    schuinezijde = math.sqrt(a**2 + b**2)
    return schuinezijde

# gegeven
rhz1 = 3                            # eerste rechthoekszijde
rhz2 = 4                            # tweede rechthoekszijde

# verwerking
c = pythagoras(rhz1, rhz2)          # schuine zijde berekenen

# uitvoer 
print("De schuine zijde van de driehoek is:", c)       # uitvoer
```

### Exercice 1.1.3

Écrivez un script dans lequel vous définissez une fonction `pythagoras`. <br>Le script demande à l’**utilisateur** de **saisir** les longueurs des deux côtés adjacents à l’angle droit d’un triangle rectangle. À l’aide de la fonction `pythagoras`, le script calculera la longueur de l’hypoténuse. Le programme se termine par l’affichage de la longueur de l’hypoténuse à l’utilisateur.<br><br>
Testez le script et améliorez-le si nécessaire.

```python
def pythagoras(a, b):
    """Schuine zijde berekenen van rechthoekige driehoek met rechthoekszijden a en b."""
    schuinezijde = math.sqrt(a**2 + b**2)
    return schuinezijde

# invoer 
rhz1 = float(input("De eerste rechthoekszijde: "))      # eerste rechthoekszijde
rhz2 = float(input("De tweede rechthoekszijde: "))      # tweede rechthoekszijde

# verwerking
c = pythagoras(rhz1, rhz2)          # schuine zijde berekenen

# uitvoer
print("De schuine zijde van de driehoek is:", c)       # uitvoer
```

### Exemple 2.1

Calculez l’aire d’un triangle rectangle dont un côté de l’angle droit mesure 8 cm et dont l’hypoténuse mesure 14 cm.

```python
import math

# oppervlakte kan berekend worden met functie uit vorige notebook
def oppdriehoek(b, h):
    """Oppervlakte van driehoek met basis b en hoogte h."""
    opp = b * h / 2
    return opp

# oppervlakte driehoek kan je berekenen als de basis en de hoogte gekend zijn, dus hier de twee rechthoekszijden
# andere rechthoekszijde berekenen met Pythagoras2
def pythagoras2(a, c):
    """Rechthoekszijde rechthoekige driehoek met schuine zijde c en rechthoekszijde a."""
    rechthoekszijde = math.sqrt(c**2 - a**2)
    return rechthoekszijde

# gegeven
rhz1 = 8                            
sz = 14                            

# verwerking
rhz2 = pythagoras2(rhz1, sz)       # andere rechthoekszijde
opp = oppdriehoek(rhz1, rhz2)      # gezochte oppervlakte 

# uitvoer
print("De oppervlakte van de driehoek is", opp, "cm².")       
```

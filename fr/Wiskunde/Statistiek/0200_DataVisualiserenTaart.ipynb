<h2>2. Diagramme circulaire</h2>    </font>
</div>

C'était monotone et peut-être pas un travail agréable. De plus, regarder une série de nombres n'est pas très attrayant, et cela demande un certain effort pour estimer correctement les proportions. En visualisant les informations du jeu de données dans un diagramme, beaucoup d'informations peuvent déjà devenir rapidement claires.

Il existe de nombreuses possibilités pour tracer des diagrammes avec Python. Une méthode simple dans ce contexte consiste à appliquer la fonction `plot()` du module pandas à un tableau de fréquences absolues. <br>Vous pouvez alors, via le paramètre `kind`, indiquer quel type de diagramme vous souhaitez tracer. En renseignant `kind="pie"`, vous dessinez par exemple un **diagramme circulaire** (**camembert**).

### Exemple 2.1

```python
plt.figure()

absolute_frequentie_vastgoed.plot(kind="pie")  # teken cirkeldiagram van de tevredenheid over vastgoedprijzen

plt.show()
```

En un coup d’œil, vous pouvez désormais extraire beaucoup d’informations des données. Dans ce diagramme, vous pouvez par exemple constater que le nombre de personnes ayant répondu 'happy' est minoritaire.

### Exemple 2.2

Vous pouvez encore améliorer la présentation en supprimant l'étiquette située à gauche et en ajoutant un titre en haut. Vous pouvez également agrandir la figure.<br>Attribuez pour cela des valeurs aux paramètres `title` et `label` de la fonction `plot()`.

```python
plt.figure(figsize=(8,8))

absolute_frequentie_vastgoed.plot(kind="pie", title="Tevredenheid over vastgoedprijzen", label="")  # Teken hetzelfde cirkeldiagram met titel

plt.show()
```

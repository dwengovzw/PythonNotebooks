<div>
    <font color=#690027 markdown="1">
        <h2>1. Le graphique de la loi normale en Python</h2>    </font>
</div>

La fonction de distribution de probabilité correspondant à la loi normale a l'expression suivante :  
$$f(x) = \frac{1}{\sigma\sqrt{2\pi}}e^{-\frac{1}{2}\frac{{x-\mu}^2}{\sigma^2}}.$$
Ici, $\mu$ est la **moyenne** et $\sigma $ l'**écart-type**.
Le graphique est une **courbe en cloche** (la soi-disant **courbe de Gauss**). <br>Selon les valeurs des deux paramètres, moyenne et écart-type, on obtient une courbe en cloche différente.

<div class="alert alert-block alert-success"> 
Dans une loi normale centrée réduite, la moyenne est $\mu = 0$ et l'écart-type est $\sigma = 1$.    </div>

Le code ci-dessous trace un graphique d'une loi normale standard. Exécutez le code.

### Exemple 1.1: le graphique de la loi normale standard

```python
# grafiek standaardnormale verdeling
# functie en haar parameters, originelen en beelden
gemiddelde = 0
standaardafwijking = 1

x = np.linspace(-5, 5, 100)                    # x-waarden genereren om grafiek te tekenen
y = norm.pdf(x, gemiddelde, standaardafwijking) # bepaal functiewaarden van normale verdeling met gegeven gemiddelde en standaardafwijking

# grafiek
plt.figure()

plt.plot(x, y)                                   # teken grafiek van  normale verdeling

plt.show()
```

En Python, vous pouvez donc faire calculer les valeurs de la fonction de densité de la loi normale via la fonction `pdf()` du sous-module 'norm' du module 'scipy.stats'. Cette fonction prend la moyenne et l'écart-type comme paramètres.

<div class="alert alert-block alert-info"> 
<b>pdf</b> signifie <b><em>probability density function</em></b>; c'est le terme anglais pour une <b>fonction de densité de probabilité</b>.</div>

<h2>2. La fonction de répartition cumulée de la loi normale</h2>    </font>
</div>


Vous pouvez obtenir la **fonction de répartition cumulée** de la loi normale via la fonction `cdf()` du même module 'norm'.

### Exemple 2.1 : le graphique de la loi normale standard

À l'aide de la cellule de code suivante, faites tracer non seulement la loi normale standard, mais aussi sa fonction de répartition cumulée. Exécutez le code.

```python
# grafiek standaardnormale verdeling en cumulatieve
# functies met parameters, originelen en beelden
gemiddelde = 0
standaardafwijking = 1

x = np.linspace(-5, 5 ,100)                         # x-waarden genereren om grafiek te tekenen
y_pdf = norm.pdf(x, gemiddelde, standaardafwijking)  # bepaalfunctiewaarden van normale verdeling
y_cdf = norm.cdf(x, gemiddelde, standaardafwijking)  # bepaal functiewaarden van cumulatieve dichtheidsfunctie 

# grafiek
plt.figure(figsize=(12,8))

plt.plot(x, y_pdf)   # teken grafiek van normale verdeling
plt.plot(x, y_cdf)   # teken grafiek van cumulatieve dichtheidsfunctie

plt.show()
```

<div class="alert alert-block alert-info"> 
    Par analogie, <b>cdf</b> signifie <b><em>cumulative density function</em></b>. C'est le terme anglais pour une <b>fonction de densité cumulée</b>.</div>

<div class="alert alert-block alert-success"> 
La fonction de répartition cumulative est une autre représentation de la loi de probabilité. La probabilité qu'une variable aléatoire soit inférieure à une certaine valeur peut être calculée comme suit à l'aide de la fonction de répartition cumulative :    $$\textit{cdf}(a) = P(X < a),$$  en d’autres termes, la surface entre la densité de probabilité $f$ et l’axe des abscisses sur l’intervalle $]-∞,a[$.    
$\textit{cdf}(10) = P(X < 10)$ correspond par exemple à l'aire entre le graphe de $f$ et l'axe des abscisses sur l'intervalle $]-∞, 10[$.  
</div>

### Exemple 2.2

- Avec la cellule de code suivante, vous calculez l'aire entre la courbe en cloche et l'axe des x.

```python
totale_kans = 2 * norm.cdf(6, gemiddelde, standaardafwijking)
print(totale_kans)
```

- Vous pouvez aussi approximer numériquement cette surface par la limite d'une suite obtenue en déplaçant la borne toujours plus loin vers l'infini positif.

```python
for grens in range(6, 50, 2):
    oppervlakte = norm.cdf(grens, gemiddelde, standaardafwijking)
    print(oppervlakte)
```

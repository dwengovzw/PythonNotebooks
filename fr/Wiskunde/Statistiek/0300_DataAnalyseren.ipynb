<b>Slicing</b> est une technique dans pandas pour découper des données ordonnées, comme ici une colonne d’un DataFrame, en morceaux. Cela peut se faire en faisant suivre la <b>clé</b> de la colonne de <b>[m:n]</b>. Ainsi, seule une partie de la colonne est considérée : la portion allant de l’élément d’indice m jusqu’à l’élément d’indice n-1. On peut aussi omettre m ou n afin d’isoler depuis le début ou jusqu’à la fin de la colonne. <br>Quelques exemples :<ul>
<li><code>kolom[2:7]</code> contient les éléments du troisième au septième inclus de <code>kolom</code>;</li><li><code>kolom[:10]</code> contient les éléments du premier au dixième (inclus) de <code>kolom</code>;</li><li><code>kolom[10:]</code> contient tous les éléments à partir du onzième élément jusqu'à la fin de <code>kolom</code>.</li></ul>
    <br>Vous pouvez également utiliser une <b>condition</b>, <b> une expression logique</b>, pour filtrer, via le slicing, certains éléments d'une colonne. <br>    Par exemple:<ul>
<li><code>kolom[kolom < 7]</code> contient toutes les valeurs de <code>kolom</code> qui sont inférieures à 7;</li><li><code>kolom[90 < kolom]</code> contient toutes les valeurs de <code>kolom</code> qui sont supérieures à 90;</li><li><code>kolom[kolom == "hallo"]</code> contient toutes les valeurs de <code>kolom</code> qui sont égales au texte "hallo".</li>    
</div>

        <h2>3. Quartiles et écart interquartile</h2>    </font>
</div>

<div class="alert alert-box alert-success">
Il existe d'autres <b>mesures de tendance centrale</b> : le premier et le troisième quartile (le deuxième quartile est égal à la médiane).Supposons que les nombres soient ordonnés du plus petit au plus grand. Alors la médiane divise les nombres en deux groupes : la partie inférieure ou égale à la médiane et la partie supérieure ou égale à la médiane. Le <b>premier quartile</b> est la médiane de la première partie ; le <b>troisième quartile</b> est la médiane de la deuxième partie. <br><br>    L'<b>écart interquartile</b> est la différence entre le premier et le deuxième quartile et constitue une <b>mesure de dispersion</b>.<br>L'<b>étendue</b> est la différence entre le plus petit et le plus grand nombre et est aussi une <b>mesure de dispersion</b>.    </div>

### Exemple 3.1Vous pouvez calculer le premier et le troisième quartile avec la méthode `quantile()`. Dans l'exemple ci-dessous, vous calculez le premier quartile de la durée d'une vidéo en minutes.

```python
mrbeast_videoduur_minuten.quantile(0.25)  # eerste kwartiel
```

La méthode `quantile()` prend un nombre réel comme argument. Ici, c’est `0.25` pour indiquer que vous voulez connaître le premier quartile, c’est-à-dire la valeur pour laquelle 25 % des autres données sont inférieures.

<div class="alert alert-block alert-info"> 
La moyenne et la médiane sont des exemples de <b>mesures de tendance centrale</b>. Elles cherchent à donner une idée du milieu ou « centre » des données dans un jeu de données. L’écart interquartile est un exemple de <b>mesure de dispersion</b>. Les mesures de dispersion illustrent à quel point les valeurs s’écartent les unes des autres.</div>

### Exemple 4.1

Avec le module pandas, tu peux également tracer un **boxplot**. Exécute la cellule de code suivante.

```python
mrbeast_videoduur_gefilterd_minuten.plot(kind='box')
```

Un diagramme en boîte s’interprète comme suit :* La ligne verte représente la médiane.* Les lignes supérieure et inférieure du rectangle représentent le premier et le troisième quartile.* La ligne supérieure et la ligne inférieure sur la barre verticale représentent le minimum et le maximum.* Les points sont des valeurs aberrantes. Avec pandas, elles sont détectées automatiquement et placées à une distance appropriée de la 'box' du boxplot.

<div class="alert alert-box alert-success">
Il existe d’autres mesures de tendance centrale et de dispersion : la <b>mesure de tendance centrale</b> <b>mode</b> et la <b>mesure de dispersion</b> <b>écart-type</b>.<br>La <b>mode</b> est la donnée ayant la fréquence (absolue ou relative) la plus élevée. Si vous travaillez avec des classes, alors la <b>classe modale</b> est la classe ayant la fréquence la plus élevée.<br>    L'<b>écart-type</b> est l'écart quadratique moyen par rapport à la moyenne. <br>    
</div>

<div class="alert alert-box alert-info">
À l'aide du module pandas, vous calculez la <b>mode</b> avec la méthode <code>mode()</code> et l'<b>écart-type</b> avec la méthode <code>std()</code> (std de l'anglais <em>standard deviation</em>).<br>
Lors de l'utilisation de la méthode <code>mode()</code>, pandas choisit automatiquement une division en classes aussi précise que possible.</div>

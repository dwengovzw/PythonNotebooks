        <h3>1.1. Pandas DataFrames</h3>    </font>

La statistique est la science qui étudie un ensemble de données de manière mathématique. Un **jeu de données** est un tel ensemble de données. Le module 'pandas' fournit de nombreuses fonctionnalités pratiques pour travailler avec des jeux de données à l'aide de Python.

```python
import pandas as pd  # importeer pandas-module
```

Tout d'abord, 'pandas' permet de lire un jeu de données avec Python. L'instruction ci-dessous charge le jeu de données `datasetvoorbeeld.csv`, qui est stocké dans le dossier `data` - donc à l'emplacement `data/datasetvoorbeeld.csv`- et y fait référence avec une variable nommée `voorbeeld`.

<div class="alert alert-block alert-warning"> 
Dans le notebook 'Diagramme de dispersion - Données à partir d'un fichier', il est expliqué comment travailler en Python avec un fichier CSV.</div>

```python
voorbeeld = pd.read_csv("data/datasetvoorbeeld.csv")
voorbeeld
```

Le jeu de données est présenté sous forme de tableau, plus précisément dans la structure de données 'pandas' **DataFrame**.<br>Un DataFrame est composé de lignes et de colonnes. Chaque **élément** du DataFrame est une donnée.<br>Ces éléments peuvent être des nombres entiers, des nombres réels, du texte, des symboles, des horodatages ...
Les colonnes d’un DataFrame peuvent contenir les valeurs de la **variable d’échantillon**, telles que la masse, la vitesse, la température ou un score. Chaque donnée dans une colonne correspond à une valeur particulière de la variable d’échantillon. Le nom d’une colonne est appelé la **clé** de la colonne.<br>Une ligne d’un DataFrame contient les valeurs de plusieurs variables d’échantillon pour un élément donné de la population étudiée.

<div class="alert alert-box alert-danger">
En statistique, on utilise le terme 'variable d'échantillon'. <br>Par exemple : si vous mesurez la vitesse d'une voiture à 5 instants, vous disposez de 5 valeurs de la variable d'échantillon « vitesse de la voiture ».    
Dans un langage de programmation, on parle également d’une « variable ». C’est-à-dire lorsque l’on veut stocker quelque chose dans la mémoire de l’ordinateur et s’y référer.    
Pour éviter toute confusion, dans ce qui suit, on emploiera systématiquement le terme <b>variable d'échantillon</b> pour les données d'une colonne d'un DataFrame et <b>variable</b> dans l'autre cas.</div>

<div>
    <font color=#690027 markdown="1">
        <h3>1.2. Taille d'un jeu de données</h3>    </font>
</div>

Le nombre de lignes dans un jeu de données s'appelle la **taille** d'un jeu de données. Vous pouvez obtenir la taille avec la fonction `len()`. C'est l'une des nombreuses opérations possibles sur un jeu de données.

```python
voorbeeld_grootte = len(voorbeeld)  # sla grootte van dataset voorbeeld op in variabele voorbeeld_grootte
voorbeeld_grootte                   # toon grootte op scherm
```

'pandas' vous permet d'isoler une colonne d'un jeu de données et d'y effectuer des calculs. Vous utilisez la *clé* de la colonne pour désigner la colonne souhaitée.

```python
kolom2 = voorbeeld["Kolom 2"]       # zonder Kolom 2 af en verwijs ernaar met variabele kolom2
kolom2                              # toon Kolom 2 op scherm
```

Sur une colonne, vous pouvez également effectuer toutes sortes d'opérations. La méthode `sum()` renvoie la somme de toutes les données de la colonne.

```python
kolom2_som = kolom2.sum()  # verwijs met variabele kolom2_som naar som van gegevens uit Kolom 2 
kolom2_som                 # toon som van gegevens uit Kolom 2 op het scherm
```

Vous pouvez également effectuer des opérations arithmétiques (+, -, . , ÷) sur toutes les données d'une colonne simultanément.

```python
kolom2_dubbel = 2 * kolom2  # verdubbel alle gegevens uit Kolom 2 en verwijs naar nieuwe kolom met kolom2_dubbel
kolom2_dubbel               # toon kolom2_dubbel op scherm
```

La méthode `value_counts()` sert à compter combien de fois chaque valeur apparaît dans une colonne donnée. Le résultat est un **tableau des fréquences absolues**.
Dans la cellule de code suivante, la méthode `value_counts()` est appliquée à la variable `kolom2`. Comprends-tu la sortie ?

```python
# absolutefrequentietabel
absolutefrequentietabel_kolom2 = kolom2.value_counts()  # tel hoeveel keer elk gegeven in Kolom 2 voorkomt
absolutefrequentietabel_kolom2                          # toon absolutefrequentietabel van Kolom 2 op scherm
```

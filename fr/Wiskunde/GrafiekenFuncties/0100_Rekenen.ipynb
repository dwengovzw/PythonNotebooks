        <h2>2. Le module standard math</h2>    </font>

<div class="alert alert-block alert-success"> 
Un <b>module</b> contient de nombreuses fonctions que des informaticiens expérimentés ont déjà programmées pour toi. Ainsi, l'utilisation de Python devient très accessible. Pour pouvoir utiliser un module, tu dois d'abord l'<b>importer</b>.  </div>

Souvent, ces opérations arithmétiques simples ne suffisent pas. Parfois, vous souhaitez par exemple calculer une racine carrée.<br>Dans le *module* math, de nombreuses fonctions mathématiques sont disponibles. En important le module, vous pouvez utiliser toutes ces fonctions. <br>Pour importer, exécutez l'instruction suivante.

```python
import math                      # module importeren
```

<div class="alert alert-block alert-success"> 
Avec <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>#</b></span>, vous pouvez ajouter des commentaires qui clarifient ce que fait le code. Les commentaires rendent le code plus facile à lire. <br>Lors de l’exécution du code, le commentaire est ignoré par Python.</div>

Outre les racines carrées, tu peux désormais, par exemple, également travailler avec le nombre $\pi$ et avec des fonctions trigonométriques.Essaie un peu.

```python
# constante pi wordt opgeroepen uit module 'math' en wordt afgedrukt
print(math.pi) 

# functie sqrt() wordt opgeroepen uit module 'math'
# vierkantswortel van 16 wordt berekend en afgedrukt
print(math.sqrt(16))                     
```

Vous pouvez aussi le faire comme suit. Vous n'avez alors plus besoin de mentionner le nom du module pour appeler une fonction ou une constante.

```python
from math import sqrt, pi              # nodige constante en functies importeren uit module math

# .............
print(pi)      

# .............
print(sqrt(2))
```

<div class="alert alert-block alert-info"> 
    <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">sqrt()</span> est une fonction du module math.</div>

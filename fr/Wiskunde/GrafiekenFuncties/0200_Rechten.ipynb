Une fonction du premier degré s’écrit toujours sous la forme $f(x) = ax + b$ ou $y = ax + b$. Le graphique correspondant à une telle expression est toujours une droite.

### Importer les modules nécessaires

```python
import numpy as np
import matplotlib.pyplot as plt
```

<div>
    <font color=#690027 markdown="1">
        <h2>1. Tracer un repère</h2>    </font>
</div>

Avant de tracer tes droites, revois brièvement comment tracer facilement l'axe des x et l'axe des y. Ceux-ci sont en effet très importants pour interpréter les graphiques de droites !<br><br>Pour cela, utilisez les fonctions `vlines()` et `hlines()` du module Matplotlib.<br>Exécutez la cellule de code ci-dessous.

```python
# ASSENSTELSEL TEKENEN

plt.figure()                                     # creëer tekenvenster

plt.hlines(0, -10, 10)                           # plot horizontale rechte van x = -10 tot x = 10 waarvoor y = 0 (x-as)
plt.vlines(0, -10, 10)                           # plot verticale rechte van y = -10 tot y = 10 waarvoor x = 0 (y-as)
 
plt.show()                                       # toon figuur
```

<div class="alert alert-box alert-success">
Désormais, ajoute toujours les axes x et y lorsque tu traces une droite !</div>

Une droite représente visuellement la relation linéaire entre deux variables. Une droite n’est donc rien d’autre qu’un ensemble de points. L’équation d’une droite (qui n’est pas parallèle à l’axe des y) sera toujours donnée ici sous la forme $y = ax + b$.<br><br>Le schéma ci-dessus montre le problème que tu dois aborder, mais décomposé en sous-problèmes (**décomposition**). Comme on le voit dans le schéma, pour pouvoir tracer une droite, tu as besoin de points que tu visualiseras dans une fenêtre de dessin.

### Exemple 2.1

Considérons la droite d’équation $y =2x-3$.- Pour pouvoir tracer le graphique de cette droite, Matplotlib a besoin de coordonnées x, données par une liste NumPy.<br>- Matplotlib a également besoin des coordonnées y correspondantes. Pour cela, tu donnes l'instruction "Calculer la coordonnée y pour chaque coordonnée x, en utilisant l'équation de la droite". En code Python, c'est `y = 2 * x - 3`, à condition que la liste des coordonnées x ait déjà été saisie.
Note que l'ordre des instructions est très important. Tu as déjà rencontré ce principe de programmation, appelé **séquence**, dans les notebooks.
Exécutez la cellule de code suivante.

```python
# GRAFIEK VAN RECHTE MET GEGEVEN VERGELIJKING: gegevens invoeren 

x = np.arange(-8, 8, 1)                     # kies x-coördinaten van punten die geplot worden        

y = 2 * x - 3                               # overeenkomstige y-coördinaten voor rechte

# coördinaten printen
print ("x = ", x)                           
print ("y = ", y)
```

Les listes nécessaires sont disponibles ! <br> Il ne te reste plus qu'à appliquer ce que tu as appris dans les notebooks 'Diagramme de dispersion'.

```python
# GRAFIEK VAN EEN RECHTE MET GEGEVEN VERGELIJKING: tekenvenster

x = np.arange(-8, 8, 1)                          # kies x-coördinaten van punten die geplot worden        
y = 2 * x - 3                                    # overeenkomstige y-coördinaten voor rechte

plt.figure()                                     # creëer figuur

plt.axis(xmin=-10, xmax=10, ymin=-25, ymax=20)   # vensterinstellingen
plt.xticks(np.arange(-10, 11, step=1))
plt.yticks(np.arange(-25, 25, step=5))

plt.grid(True)                                   # plot rooster

plt.scatter(x, y)                                # plot punten
 
plt.show()                                       # toon figuur
```

Ajoutez déjà l'axe des x et l'axe des y :

```python
# GRAFIEK VAN EEN RECHTE MET GEGEVEN VERGELIJKING: tekenvenster

x = np.arange(-8, 8, 1)                          # kies x-coördinaten van punten die geplot worden        
y = 2 * x - 3                                    # overeenkomstige y-coördinaten voor rechte

plt.figure()                                     # creëer figuur

plt.axis(xmin=-10, xmax=10, ymin=-25, ymax=20)   # vensterinstellingen
plt.xticks(np.arange(-10, 11, step=1))
plt.yticks(np.arange(-25, 25, step=5))

plt.hlines(0, -10, 10)                           # plot horizontale rechte van x = -10 tot x = 10 waarvoor y = 0 (x-as)
plt.vlines(0, -10, 10)                           # plot verticale rechte van y = -10 tot y = 10 waarvoor x = 0 (y-as)

plt.grid(True)                                   # plot rooster

plt.scatter(x, y)                                # plot punten
 
plt.show()                                       # toon figuur
```

<h2>2. Deux droites d'équations données se coupent</h2>    </font>
</div>

Le module SymPy convient pour résoudre des équations, calculer des coordonnées, déterminer les racines et résoudre des systèmes. Pour déterminer le point d’intersection de deux droites, il faut résoudre un système d’équations. Ainsi, pour calculer avec Python le point d’intersection de deux droites, tu utilises le module Sympy. Les éléments nécessaires ont déjà été importés.<br>
Outre les éléments énumérés ci-dessus, SymPy est également pratique car il n'approxime pas les nombres rationnels mais les laisse sous forme de fractions. SymPy conserve également les radicaux, éventuellement simplifiés.

### ExempleÀ l'aide de la cellule de code suivante, tu calcules le point d'intersection des deux droites sécantes de la question 1 avec Python. <br>Pour déterminer le point d'intersection des deux droites, il faut résoudre un système de deux équations à deux inconnues x et y :$$\left\{    \begin{array}{ll}        -3 x + 5 - y = 0 \\        2 x + 16 - y = 0    \end{array}\right..$$**Tu devras également indiquer en Python le rôle de x et y en tant qu’inconnues.**<br>**Notez que les équations sont réduites à zéro.**<br>Examinez attentivement le code dans la cellule et exécutez-le.

```python
# TWEE RECHTEN MET GEGEVEN VERGELIJKING SNIJDEN

# x en y beschouwen als onbekenden
x = Symbol("x")         
y = Symbol("y") 

# vergelijking van de eerste rechte: y = -3x + 5
# vergelijking van de tweede rechte: y = 2x + 16

# snijpunt berekenen door het stelsel met vergelijkingen −3 𝑥 + 5 − 𝑦 = 0  en 2 𝑥 + 16 − 𝑦 = 0 op te lossen naar x en y
snijpunt = solve((-3 * x + 5 - y, 2 * x + 16 - y), x, y)

print(snijpunt)
```

La fonction `solve()` renvoie un *dictionnaire*. Attention aux accolades, qui indiquent que l'ordre dans un dictionnaire n'a pas d'importance. Pour le problème ici, cet ordre est bien entendu important : l'ordre est mis en évidence par les *clés* du dictionnaire, à savoir x et y.<br>

<div class="alert alert-box alert-warning">
Vous trouverez plus d'explications sur les dictionnaires dans les notebooks de Programmeerelementen sur les types de données.</div>

```python
# betere weergave van berekende snijpunt
print("Het snijpunt van de twee rechten is: ", (snijpunt[x], snijpunt[y]))   
```

```python
# SNIJPUNT PLOTTEN

plt.figure()

plt.plot(snijpunt[x], snijpunt[y], color="plum", marker="o")

plt.show()
```

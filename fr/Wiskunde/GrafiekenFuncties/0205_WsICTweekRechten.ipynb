Une fonction affine est toujours donnée par une formule de la forme $f(x) = ax + b$ ou $y = ax + b$. La représentation graphique correspondant à une telle formule est une droite.
Une droite est un ensemble de points.

### Importer les modules nécessaires

```python
import numpy as np
import matplotlib.pyplot as plt
```

<div>
    <font color=#690027 markdown="1">
<h2>2. Tracer un repère</h2>    </font>
</div>

On peut tracer un axe des x et un axe des y à l'aide des fonctions `vlines()` et `hlines()` du module Matplotlib.<br>Vous pouvez également indiquer l'étalonnage et ajuster la taille de l'écran.<br>Exécutez la cellule de code ci-dessous.

```python
# ASSENSTELSEL TEKENEN

plt.figure(figsize=(15,10))                      # creëer tekenvenster

plt.hlines(0, -10, 20)                           # plot horizontale rechte van x = -10 tot x = 20 waarvoor y = 0 (x-as)
plt.vlines(0, -10, 10)                           # plot verticale rechte van y = -10 tot y = 10 waarvoor x = 0 (y-as)
 
plt.axis(xmin=-10, xmax=20, ymin=-10, ymax=10)   # vensterinstellingen
plt.xticks(np.arange(-10, 21, step=1))           # labels op de x-as
plt.yticks(np.arange(-10, 11, step=1))           # labels op de y-as

plt.grid(True)                                   # plot rooster

plt.show()                                       # toon figuur
```

Une droite représente visuellement la relation linéaire entre deux variables. Une droite n'est donc rien d'autre qu'un ensemble de points. L'équation d'une droite (qui n'est pas parallèle à l'axe des y) sera toujours donnée ici sous la forme $y = ax + b$.<br><br>Le schéma ci-dessus représente le problème que tu dois aborder, mais divisé en sous-problèmes (**décomposition**). Comme on peut le voir dans le schéma, pour pouvoir tracer une droite, tu as besoin de points que tu visualiseras dans une fenêtre de dessin.

### Exemple 3.1

Considérez la droite d’équation $y =0,8 \; x-5$.- Pour pouvoir tracer le graphique de cette droite, Matplotlib a besoin de coordonnées x, fournies par une liste NumPy.<br>- Matplotlib a également besoin des coordonnées y correspondantes. Pour cela, vous donnez l'instruction "Calculez la coordonnée y pour chaque coordonnée x, en utilisant l'équation de la droite". En code Python, cela donne `y = 0.8 * x - 5`, à condition que la liste des coordonnées x soit déjà saisie.
Exécutez les cellules de code suivantes.

```python
# GRAFIEK VAN RECHTE MET GEGEVEN VERGELIJKING: gegevens invoeren 

x = np.arange(-4, 12, 1)                     # kies x-coördinaten van punten die geplot worden        

# x-coördinaaten laten zien
print ("x = ", x)                           
```

```python
# GRAFIEK VAN RECHTE MET GEGEVEN VERGELIJKING: gegevens invoeren 

x = np.arange(-4, 12, 1)                     # kies x-coördinaten van punten die geplot worden        

y = 0.8 * x - 5                              # overeenkomstige y-coördinaten voor rechte

# coördinaten laten zien
print ("x = ", x)                           
print ("y = ", y)
```

La taille de la figure et l'utilisation de titres et d'étiquettes sont des moyens de rendre les graphiques plus clairs. <br>Comme tu l'auras déjà remarqué dans les notebooks précédents, il est également possible d'ajouter ici et là un petit morceau de code dans le code existant pour modifier la mise en forme. Les options dont tu disposes pour cela sont assez étendues. <br>Voici déjà les plus importantes.

Cette cellule de code vous montrera à quoi ressemble un graphique sans mise en forme supplémentaire. Exécutez-la.

```python
# koppels (O,3), (1,24), (2,12), (3,48), (4,6) ingeven m.b.v. Numpy lijsten
x = np.array([0, 1, 2, 3, 4])                   # NumPy-lijst met x-coördinaten          
y = np.array([3, 24, 12, 48, 6])                # NumPy-lijst met corresponderende y-coördinaten, volgorde is van belang 

plt.figure(figsize=(10,10))                     # creëer tekenvenster van bepaalde grootte

# plot punten  
plt.plot(x, y)                                  # punten verbinden
      
# toon venster
plt.show()
```

Dans les notebooks précédents, les points étaient affichés au moyen de la fonction `scatter()` de Matplotlib. Ici, vous utilisez la fonction `plot()`. Le graphique n’affiche plus seulement les points : ils sont également reliés entre eux par des **segments**.

Exécutez la cellule de code ci-dessous. Observez à présent les modifications de la mise en forme via les arguments de la fonction `plot()` de Matplotlib.

```python
# koppels (O,3), (1,24), (2,12), (3,48), (4,6) ingeven m.b.v. Numpy lijsten
x = np.array([0, 1, 2, 3, 4])                   # NumPy-lijst met x-coördinaten          
y = np.array([3, 24, 12, 48, 6])                # NumPy-lijst met corresponderende y-coördinaten, volgorde is van belang 

plt.figure(figsize=(10,10))                     # creëer tekenvenster van bepaalde grootte

plt.plot(x, y, color="orange", marker="o", linestyle="dashed", linewidth="3")  # plot punten en lijnstukken in bepaalde stijl
      
plt.show()                                      # toon grafiek
```

Notez que si vous utilisez la fonction `plot()` sans arguments supplémentaires, le graphique adopte une *mise en forme par défaut*. Cette mise en forme se compose d'une ligne bleue continue d'épaisseur 1 et de points représentés par un point.

### 1.1 Ajuster la couleur : `color`Pour ajuster la couleur d’un graphique, ajoutez l’attribut `color` comme argument. Complétez-le ensuite avec la couleur de votre choix.<br>Voici quelques exemples :- <em>black</em>;- <em>gray</em>;- <em>red</em>;- <em>orange</em>;- <em>yellow</em>;- <em>green</em>;- <em>blue</em>;- <em>indigo</em>;- <em>violet</em>.
Vous trouverez la liste complète des couleurs parmi lesquelles vous pouvez choisir sur <a href="https://matplotlib.org/stable/gallery/color/named_colors.html">liste des couleurs</a>.

Vous pouvez essayer ces couleurs ci-dessous.

```python
# koppels (O,3), (1,24), (2,12), (3,48), (4,6) ingeven m.b.v. Numpy arrays
x = np.array([0, 1, 2, 3, 4])                           # NumPy-lijst met x-coördinaten          
y = np.array([3, 24, 12, 48, 6])                        # NumPy-lijst met corresponderende y-coördinaten, volgorde is van belang 

plt.figure(figsize=(10,10))                             # creëer tekenvenster van bepaalde grootte

plt.plot(x, y, color="darkgoldenrod")                   # plot punten en lijnstukken in bepaalde kleur
       
plt.show()                                              # toon grafiek
```

### 1.2 Mise en forme des points : `marker`Pour modifier le symbole utilisé pour tracer les points, ajoutez l’option `marker`. Complétez-la ensuite avec le symbole de votre choix.<br>Voici quelques exemples :- <em>.</em>- <em>,</em>- <em>o</em>- <em>+</em>- <em>x</em>
La liste complète des marqueurs parmi lesquels vous pouvez choisir se trouve sur <a href="https://matplotlib.org/stable/api/markers_api.html">liste des marqueurs</a>.

Vous pouvez tester ces caractères ci-dessous.

```python
# koppels (O,3), (1,24), (2,12), (3,48), (4,6) ingeven m.b.v. Numpy arrays
x = np.array([0, 1, 2, 3, 4])                           # NumPy-lijst met x-coördinaten          
y = np.array([3, 24, 12, 48, 6])                        # NumPy-lijst met corresponderende y-coördinaten, volgorde is van belang 

plt.figure(figsize=(10,10))                             # creëer tekenvenster van bepaalde grootte

plt.plot(x, y, marker="o")                              # plot punten in bepaalde stijl 
      
plt.show()                                              # toon grafiek
```

### 1.3 Mise en forme des lignes : `linestyle` et `linewidth`Pour ajuster le style d’un segment de ligne, ajoutez le paramètre `linestyle` en tant qu’argument. Vous le complétez ensuite avec un style de votre choix.<br>Voici quelques exemples :- <em>solid</em> (la valeur par défaut);- <em>dotted</em> (ligne pointillée);- <em>dashed</em> (ligne en pointillés);- <em>dashdot</em> (alternativement _ . _ . _ ).
Vous trouverez la liste complète des styles parmi lesquels vous pouvez choisir sur <a href="https://matplotlib.org/stable/gallery/lines_bars_and_markers/linestyles.html">liste des styles</a>.

Pour ajuster l’épaisseur d’une ligne ou d’une droite, ajoutez l’attribut `linewidth` comme argument. Complétez-le ensuite avec une épaisseur de votre choix.<br>L’épaisseur est un nombre au choix.

Vous pouvez essayer ces styles ci-dessous.

```python
# koppels (O,3), (1,24), (2,12), (3,48), (4,6) ingeven m.b.v. Numpy arrays
x = np.array([0, 1, 2, 3, 4])                           # NumPy-lijst met x-coördinaten          
y = np.array([3, 24, 12, 48, 6])                        # NumPy-lijst met corresponderende y-coördinaten, volgorde is van belang 

plt.figure(figsize=(10,10))                             # creëer tekenvenster van bepaalde grootte

plt.plot(x, y, linestyle="solid", linewidth="4")        # plot lijnstukken in bepaalde stijl 
      
plt.show()                                              # toon grafiek
```

        <h2>3. Esquisse d'un graphique</h2>    </font>
</div>

Une manière spéciale de mettre en forme un graphique est <em>xkcd</em>. <br>Vous pouvez implémenter cette approche à l’aide de la fonction `xkcd()` de Matplotlib. Assurez-vous de placer cette fonction avant l’instruction `plot()` !

```python
# koppels (O,3), (1,24), (2,12), (3,48), (4,6) ingeven m.b.v. Numpy arrays
x = np.array([0, 1, 2, 3, 4])                           # NumPy-lijst met x-coördinaten          
y = np.array([3, 24, 12, 48, 6])                        # NumPy-lijst met corresponderende y-coördinaten, volgorde is van belang 

plt.figure(figsize=(10,10))                             # creëer tekenvenster van bepaalde grootte

plt.xkcd()                                              # xkcd lay-out oproepenn
plt.plot(x, y)                                          # plot lijnstukken

plt.show()                                              # toon grafiek
```

Il s’agit d’un style de mise en forme qui vous permet de donner aux éléments l’apparence d’un dessin fait à la main.

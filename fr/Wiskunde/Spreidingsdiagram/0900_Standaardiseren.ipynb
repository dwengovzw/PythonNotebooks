Le degré de relation linéaire entre les coordonnées x et y des points donnés peut être examiné à l'aide du *coefficient de corrélation R*.

<div class="alert alert-box alert-info">
Le coefficient de corrélation, un nombre réel R, se situe toujours entre -1 et 1. Plus R est proche de 0, plus la corrélation linéaire est faible. <br>Pour un R positif, on parle d'une corrélation linéaire positive, pour un R négatif d'une corrélation linéaire négative.</div>

    <font color=#690027 markdown="1">
        <h2>5. Standardiser</h3>    </font>
</div>

Les données ne peuvent être comparées que si elles sont standardisées. Pour cela, on calcule le *Z-score* des données : on soustrait la moyenne puis on divise par l'écart-type.

```python
x1 = (x1-np.mean(x1))/np.std(x1)
y1 = (y1-np.mean(y1))/np.std(y1)
x2 = (x2-np.mean(x2))/np.std(x2)
y2 = (y2-np.mean(y2))/np.std(y2)
x3 = (x3-np.mean(x3))/np.std(x3)
y3 = (y3-np.mean(y3))/np.std(y3)
```

Vous pouvez maintenant tracer les nuages de points qui correspondent à ces données modifiées.

```python
# gestandaardiseerde puntenwolk

plt.figure()

plt.scatter(x1, y1, color="red", marker="o")      
plt.scatter(x2, y2, color="yellow", marker=">")
plt.scatter(x3, y3, color="lightblue", marker="<")

plt.title("Gestandaardiseerd")
plt.xlabel("breedte bloemblad (mm)")
plt.ylabel("lengte bloemblad (mm)")
plt.xlim(-2, 2)
plt.ylim(-2, 2)

plt.show()
```

La corrélation linéaire est pratiquement la même pour les trois tableaux. Après standardisation, les points se superposent presque.<br>

Vous pouvez trouver plus d'explications sur ces notions dans le cours 'Corrélation : méthodes exploratoires : feuille de travail pour l'élève'. [1]

<div class="alert alert-box alert-info">
Comparer la corrélation de plusieurs jeux de données entre eux se fait au mieux lorsque les données ont d'abord été standardisées. La standardisation n'a d'ailleurs aucun effet sur la valeur du coefficient de corrélation.<br>La corrélation entre <b>données bivariées</b>, par exemple, peut être estimée visuellement en regardant le <b>nuage de points</b> correspondant. Cependant, sa forme n'est fiable que si les données sont standardisées. <br><b>Standardiser</b> des variables consiste à ramener les variables à une échelle permettant de comparer ou de mettre en relation des variables, par exemple, d’ordres de grandeur différents ou exprimées dans des unités différentes.</div>

<div class="alert alert-box alert-info">
Après la standardisation, la plupart des valeurs se situent entre 0 et 1, ce qui est avantageux pour les calculs effectués par l’ordinateur.<br>Le calcul avec des nombres relativement grands mène rapidement à des nombres encore plus grands et à une instabilité numérique ; c’est une raison supplémentaire pour laquelle les données sont standardisées. <br>De plus, certains algorithmes d’apprentissage automatique ne sont utilisables que si les données sont standardisées, car ces algorithmes sont conçus ainsi.</div>

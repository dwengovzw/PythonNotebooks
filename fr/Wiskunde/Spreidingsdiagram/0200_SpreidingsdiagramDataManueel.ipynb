Lorsqu'on le fait, on obtient un *diagramme de dispersion* ou *nuage de points*. La figure 1 présente un exemple d'un tel nuage de points pour deux variables.

<h3>2.3 Tracer des données</h3></div>

Avec la fonction `scatter()` de pyplot, vous tracez les points. <br>Lors du traçage, Matplotlib associera les x et les y ayant la même position dans la liste (voir Figure 2). <br><br>

<img src="images/overeenkomst.png" alt="Banner" width="200"/><br>

Dans la cellule de code suivante, les instructions nécessaires aux sous-problèmes sont regroupées. Exécutez la cellule de code.

```python
# SPREIDINGSDIAGRAM

# koppels (O,3), (1,24), (2,12), (3,48), (4,6) ingeven m.b.v. NumPy-lijsten
x = np.array([0, 1, 2, 3, 4])            # NumPy-lijst met x-coördinaten          
y = np.array([3, 24, 12, 48, 6])         # NumPy-lijst met corresponderende y-coördinaten, volgorde is van belang 

plt.figure()                             # creëer tekenvenster

plt.scatter(x, y)                        # plot punten
      
plt.show()                               # toon grafiek
```

Un tel graphique est donc un diagramme de dispersion. <br>Comme vous pouvez le voir, Python utilise par défaut des marqueurs bleus pour représenter les points.

Notez que `axis()` de pyplot n'est pas nécessaire si vous avez des points de données à afficher. Python choisira alors lui-même les paramètres de la fenêtre afin d'afficher tous les points de données.

    <font color=#690027 markdown="1">
        <h2>3. Mise en forme du graphique</h2>    </font>
</div>

Une fois que vous avez tracé le graphique dans la fenêtre de dessin, vous pouvez envisager de le mettre en forme. <br>Quelques options sont :- ajustement des limites de la fenêtre de tracé;- modification de la taille de la fenêtre de dessin;- ajout d'une grille.

Pour ce faire, vous devez ajouter des instructions dans le code : `plt.axis()` pour la plage, `figsize=(a, b)` pour la taille, et pour la grille `plt.grid(True)`.

Dans la cellule de code ci-dessous, le code du diagramme de dispersion ci-dessus a été étoffé afin de le rendre un peu plus soigné.<br>Il faut notamment tenir compte des valeurs x et y minimales et maximales lors du choix de l'intervalle.

### Exemple 3.1

```python
# VOORBEELD 3.1

# koppels (O,3), (1,24), (2,12), (3,48), (4,6) ingeven m.b.v. Numpy-lijsten
x = np.array([0, 1, 2, 3, 4])            # NumPy-lijst met x-coördinaten          
y = np.array([3, 24, 12, 48, 6])         # NumPy-lijst met corresponderende y-coördinaten, volgorde is van belang

plt.figure(figsize=(9,5))                       # creëer tekenvenster

# vensterinstellingen
plt.axis(xmin=-1, xmax=5, ymin=-10, ymax=70)    # bereik x-as en y-as

plt.grid(True)                                  # rooster

plt.scatter(x, y)                               # plot punten  

plt.show()                                      # toon grafiek
```

### Exemple 3.2

Vous pouvez également ajuster manuellement l’espacement des graduations sur les axes horizontal et vertical. Comme celui de l’axe horizontal est correct, n’ajustez que celui de l’axe vertical : au lieu d’intervalles de 10, utilisez des intervalles de 5. Pour cela, utilisez la fonction `yticks()` de pyplot et la fonction `arange()` de NumPy.

```python
# VOORBEELD 3.2

# koppels (O,3), (1,24), (2,12), (3,48), (4,6) ingeven m.b.v. Numpy-lijsten
x = np.array([0, 1, 2, 3, 4])                    # NumPy-lijst met x-coördinaten          
y = np.array([3, 24, 12, 48, 6])                 # NumPy-lijst met corresponderende y-coördinaten, volgorde is van belang

plt.figure(figsize=(9,5))                        # creëer tekenvenster

# vensterinstellingen
plt.axis(xmin=-1, xmax=5, ymin=-10, ymax=70)     # bereik x-as en y-as
plt.yticks(np.arange(-10, 71, step=5))           # verdeling y-as (halfopen interval [-10, 71[ verdelen in stappen van 5)

plt.grid(True)                                   # rooster
 
plt.scatter(x, y)                                # plot punten

plt.show()                                       # toon figuur
```

<div class="alert alert-box alert-info">
Avec la fonction arange() de NumPy, une liste NumPy est générée. <br>Ici, l'instruction np.arange(-10, 71, step=5) donne la liste: <br>      <center>  [-10,  -5,   0,   5,  10,  15,  20,  25,  30,  35,  40,  45,  50, 55,  60,  65,  70].</center></div>

<h2>3. Régression linéaire</h2>    </font>
</div>

Une droite (oblique) a une équation de la forme $y =  a x + b$, autrement dit, c’est le graphe d’une fonction $f$ définie par $f(x) = a x + b$. Dans le cas où $ a = 0$, la droite est horizontale. <br><br>Pour trouver avec Python la droite qui s'ajuste le mieux aux données, la **droite de régression**, vous utilisez deux fonctions définies par l'utilisateur : `rechte()` et `linreg()`.- La fonction `rechte()` indique comment est construite l'équation d'une droite (oblique) ;- la fonction `linreg()` recherche les coefficients dans ce modèle.
La fonction `linreg()` fait elle-même appel à la fonction `curve_fit()` du module SciPy. `curve_fit()` renvoie deux éléments, auxquels on fait référence sous les noms `popt` et `pcov`.- `popt` renvoie à une liste des paramètres optimaux (donc les coefficients de l’équation de la droite la mieux ajustée) ;- `pcov` fait référence à 'l'erreur', la mesure dans laquelle les données s'écartent de la droite trouvée.
Vous vous limiterez ici à la recherche des coefficients. Vous n'avez donc besoin que de `popt` : la liste à laquelle `popt` fait référence contient les coefficients $a$ et $b$.

```python
# regressielijn is rechte

# ingeven hoe vergelijking rechte is opgebouwd 
def rechte(x, a, b):
    """Voorschrift (schuine) rechte met variabele x en coëfficiënten a en b."""
    return a * x + b

# rechte zoeken die beste past bij bepaalde data, vergelijking tonen en coëfficiënten teruggeven
def linreg(x, y):
    """Rechte best passend bij data x en y."""
    popt, pcov = curve_fit(rechte, x, y)            # curve_fit() kijkt in def rechte() hoe functievoorschrift eruitziet
    # curve_fit() geeft twee zaken terug, waaraan gerefereerd wordt met popt en pcov
    # enkel eerste nodig, popt, die a en b van gezochte rechte geeft
    a, b = popt                                     # coëfficiënten
    print("y = ", a, "x +", b)                      # toon vergelijking regressielijn
    return a, b                                     # geeft coëfficiënten terug van vergelijking regressielijn
```

Dans la cellule de code suivante, la fonction `linreg()` est appelée en utilisant comme valeurs pour les paramètres x et y les données x et y fournies au début de ce notebook. La fonction affiche l’équation de la droite de régression recherchée et renvoie les coefficients afin que vous puissiez les utiliser plus loin dans le notebook.

```python
# coëfficiënten regressielijn bij gegeven punten
a, b = linreg(x, y)
print(a, b)
```

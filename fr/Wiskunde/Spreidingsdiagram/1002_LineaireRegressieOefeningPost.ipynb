### TâcheDans le graphique, on voit que le nombre de lettres qu’un Belge envoie par an a fortement diminué sur la période 2010-2015. Le nombre de lettres envoyées présente donc une **tendance à la baisse**. Cela a eu pour conséquence que le facteur ne distribue plus vos lettres quotidiennement.
Vous examinez le nombre de lettres envoyées par Belge et par an en fonction de l'année. <br>Extrayez les données dont vous avez besoin de la figure. Saisissez les données comme suit :- les valeurs de x correspondent à l'année;- les valeurs de y correspondent au nombre de lettres par Belge et par an.
Attention à la formulation '**en fonction de**' ! Tu en déduis quelle grandeur va sur l'axe des abscisses et laquelle sur l'axe des ordonnées.
Ensuite, vous déterminez :le degré de liaison entre l'année et le nombre de lettres à l'aide du coefficient de corrélation ;- l'équation de la **ligne de tendance**. <br>
Enfin, représentez clairement les données fournies et la ligne de tendance sur un graphique.
Astuce : Vous pouvez améliorer la représentation de la ligne de tendance en utilisant des valeurs x supplémentaires. Ainsi, vous pouvez indiquer comment les chiffres pourraient évoluer à l'avenir.

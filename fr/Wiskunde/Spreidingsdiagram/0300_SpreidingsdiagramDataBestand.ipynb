```python
# data omzetten naar NumPy-lijsten

x = np.array(x)
y = np.array(y)
```

### Exemple 2.4Vous pouvez supprimer des éléments d'une liste NumPy via l'*index* à l'aide de la fonction `delete()`.

<div class="alert alert-block alert-warning"> 
Dans le notebook <em>'Diagramme de dispersion - Saisie manuelle des données'</em>, tu as déjà appris comment réaliser un diagramme de dispersion. Tu peux éventuellement y jeter un coup d'œil.</div>

L'étendue de l'axe des x est facile à voir car les x sont ordonnés dans la liste. <br>Pour les y, ce n’est pas le cas. Pour déterminer la plage sur l’axe des y, il est utile de connaître la plus petite et la plus grande valeur de y. Pour cela, on utilise les méthodes `min()` et `max()`. Comme ce sont des méthodes, elles se placent dans l’instruction après l’objet y, séparées par un point, par exemple `y.max()`.<br><br>Il n’est pas toujours aussi simple de déterminer les réglages de fenêtre appropriés pour un jeu de données. Pour un jeu de données comme celui-ci, il est encore possible d’estimer en regardant les données et en choisissant des limites.<br>Pour un jeu de données plus volumineux et/ou non ordonné, il est toutefois préférable d'utiliser une méthode plus automatisée : du code Python qui le fera pour vous.<br><br>Essayez dès maintenant cette méthode !

```python
# minima en maxima bepalen

x_min = x.min()
x_max = x.max()

y_min = y.min()
y_max = y.max()

print(x_min, x_max, y_min, y_max)
```

Maintenant que tu connais les limites du jeu de données, tu peux les saisir pour créer une fenêtre graphique adaptée.

Il est toujours préférable de ne pas créer de graphiques trompeurs et donc d’afficher également 0 sur l’axe des ordonnées. Sinon, les hausses et les baisses éventuelles sont fortement amplifiées.</div>

        <h2>4. Pratique</h2>     </font>
</div>

Lorsque vous importez un fichier et souhaitez visualiser les données, vous exécuterez les étapes nécessaires successivement. Vous devrez de temps en temps examiner les résultats intermédiaires afin de savoir comment poursuivre. Par exemple, vous devrez consulter l’en-tête des colonnes ainsi que les valeurs minimales et maximales.<br>Vous trouverez ici tout le code récapitulé une nouvelle fois.

```python
# data uit bestand inlezen

csv = pd.read_csv("data/punten.csv")                      # inlezen van tabel, je kan zelf de variabelenaam kiezen!
csv                                                       # data tonen, sleutels (keys, hoofding kolommen) bekijken
```

```python
# data uit bestand omzetten naar NumPy-lijsten

x = csv["x"]                                         # 2e kolom naam geven m.b.v. sleutel x 
y = csv["y"]                                         # 3e kolom naam geven m.b.v. sleutel y

x = np.array(x)                                      # NumPy-lijst aanmaken met gewenste data
y = np.array(y)

print(x)                                             # NumPy-lijst weergeven
print(y)
```

Lorsque vous souhaitez visualiser des données importées, vous pouvez utiliser le code suivant :

```python
# minima en maxima bepalen

x_min = x.min()
x_max = x.max()

y_min = y.min()
y_max = y.max()

print(x_min, x_max, y_min, y_max)
```

```python
# grafische voorstelling

plt.figure()                                       # creëer tekenvenster

plt.axis(xmin=..., xmax=..., ymin=..., ymax=...)    # vensterinstellingen
plt.xticks(np.arange(..., ..., step=...))
plt.yticks(np.arange(..., ..., step=...))

plt.scatter(x, y)                                  # plot punten

...                                                # titel en labels

plt.show()                                         # toon grafiek
```

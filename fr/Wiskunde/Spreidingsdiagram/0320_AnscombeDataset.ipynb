<div class="alert alert-box alert-info">
Le coefficient de corrélation R est un nombre réel qui est toujours compris entre -1 et 1. Plus R est proche de 0, plus la corrélation est faible. <br>Pour un R positif, on parle d'une corrélation positive; pour un R négatif, d'une corrélation négative.<br> Si le coefficient de corrélation est proche de 1 ou de -1, il y a une forte corrélation.<br>Vous pouvez déterminer le coefficient de corrélation avec la fonction `corrcoef()` du module NumPy. Cette fonction renvoie en réalité un tableau à 2 lignes et 2 colonnes, dont le coefficient de corrélation se trouve sur la première ligne dans la deuxième colonne.    </div>

Vous pouvez déterminer le coefficient de corrélation avec la fonction `corrcoef()` du module NumPy. Cette fonction renvoie en fait un tableau de 2 lignes et 2 colonnes, dont le coefficient de corrélation se trouve sur la première ligne, dans la deuxième colonne.

- Comparez les coefficients de corrélation des trois nuages de points :

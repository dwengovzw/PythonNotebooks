### Importer les modules nécessaires

<div class="alert alert-block alert-info"> 
Python est un langage de programmation très accessible, souvent intuitif. Des informaticiens expérimentés ont déjà programmé de nombreuses fonctions ; celles-ci sont regroupées dans des <b>modules</b>.<br>Vous pouvez utiliser ces modules ; ainsi, vous pouvez faire beaucoup de choses rapidement avec peu de code. Pour pouvoir utiliser un module, vous devez d'abord l'<b>importer</b>. </div>

<div class="alert alert-box alert-info">
Les fonctions permettant de créer et de mettre en forme une fenêtre pour des graphiques se trouvent dans le module Matplotlib, plus précisément dans le sous-module pyplot. Les fonctions de pyplot permettent de créer des graphiques et de les personnaliser, comme p. ex. ajouter une grille, afficher une partie déterminée du système d’axes, ajouter des couleurs …</div>

```python
import matplotlib.pyplot as plt
```

<div>
    <font color=#690027 markdown="1">
        <h2>1. Création de la fenêtre de dessin</h2>    </font>
</div>

Avant de pouvoir tracer un graphique, vous devez d'abord créer un emplacement où celui-ci peut être tracé. Cet emplacement est 'la fenêtre de dessin'.

Exécutez la cellule de code ci-dessous.

```python
# Tekenvenster

plt.figure()                                       # creëer tekenvenster

plt.axis(xmin=-1, xmax=10, ymin=-1, ymax=10)       # bepaal assen
      
plt.show()                                         # toon venster
```

<div class="alert alert-block alert-info"> 
Avec <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>#</b></span> vous pouvez ajouter des commentaires qui expliquent ce que fait le code. Les commentaires rendent le code plus lisible. <br>Lors de l'exécution du code, les commentaires sont ignorés par Python.</div>

<div>
    <font color=#690027 markdown="1">
        <h2>2. Mise en forme de la fenêtre graphique</h2>    </font>
</div>

Il n'est pas nécessaire de prévoir beaucoup de mise en forme pour la fenêtre de tracé. Un titre pour l'ensemble et l'étiquetage des axes suffisent. Pour cela, utilisez les fonctions `title()` et `xlabel()` de pyplot.<br>Vous pouvez également définir les dimensions de la fenêtre de tracé en ajoutant `figsize=()` lors de l'initialisation (création) de la figure.
Exécutez la cellule de code ci-dessous.

```python
# Tekenvenster

plt.figure(figsize=(15,10))                       # creëer venster met bepaalde grootte

plt.axis(xmin=-1, xmax=10, ymin=-1, ymax=10)      # vensterinstellingen
      
plt.title("Titel")                                # titel geven aan grafiek
plt.xlabel("horizontaal")                         # omschrijving bij horizontale as
plt.ylabel("verticaal")                           # omschrijving bij verticale as    
 
plt.show()                                        # toon venster
```

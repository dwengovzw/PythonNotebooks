Pour créer à partir de **fichiers csv** - csv signifie **comma separated values** - des tableaux clairs, on utilise le module 'pandas'.<br>Pour convertir les colonnes d'un tel tableau en une liste et pouvoir effectuer des calculs, importez le module 'NumPy'.<br>Pour créer un diagramme de dispersion des données, on utilise le sous-module 'pyplot' du module 'Matplotlib'.

<div>
    <font color=#690027 markdown="1">
<h2>1. Importer des données</h2>    </font>
</div>

Vous importez les modules nécessaires en exécutant la cellule de code suivante.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

Le fichier CSV a déjà été placé **sur le serveur**. Cependant, pour utiliser le fichier CSV, vous devez d'abord l'**importer** dans le notebook.<br>Lisez donc le fichier en exécutant la cellule de code suivante :

```python
csv = pd.read_csv("data/punten.csv")                     # inlezen bestand
```

Et récupérez le résultat :

```python
csv
```

Si vous souhaitez voir uniquement le début du tableau, ou seulement ses dernières lignes, vous pouvez le faire à l'aide des méthodes `head()` et `tail()`.

```python
csv.head()
```

```python
csv.tail()
```

Le tableau a 15 lignes et 2 colonnes. <br>

Pour représenter ces données sur un graphique, vous avez besoin de deux listes NumPy : l'une pour les coordonnées x et l'autre pour les coordonnées y. Avec 'pandas', c'est assez simple, car vous pouvez utiliser l'en-tête du tableau comme **clé** *(key)*.

```python
# data opsplitsen

x = csv["x"]     # hoofding van kolom is sleutel, die speelt rol van index, x-coördinaat komt op de x-as
y = csv["y"]     # y-coördinaat komt op y-as
```

Voyez à quoi ressemble une telle liste :

```python
print(x)
```

C'est toujours une sorte de tableau. En fait, pour x, vous n'avez besoin que de la deuxième colonne de ce tableau.Vous pouvez obtenir cela en utilisant une conversion de type : en convertissant ce tableau pandas en une liste NumPy, vous obtenez le résultat souhaité. **C'est une étape très importante.**

```python
# data omzetten naar NumPy-lijsten

x = np.array(x)
y = np.array(y)
```

```python
# data bekijken

print(x)
print(y)
```

<div>
    <font color=#690027 markdown="1">
        <h2>2. Visualiser les données</h2>    </font>
</div>

<div class="alert alert-block alert-warning"> 
Dans le notebook <em>'Diagramme de dispersion - Saisir les données manuellement'</em>, tu as déjà appris comment créer un diagramme de dispersion. N'hésite pas à y revenir.</div>

L’étendue de l’axe des x est facile à voir car les x sont ordonnés dans la liste. <br>Pour les y, ce n’est pas le cas. Pour déterminer l’étendue sur l’axe des y, il est utile de connaître la plus petite et la plus grande valeur de y. Pour cela, vous utilisez les méthodes `min()` et `max()`. Comme il s’agit de méthodes, elles se placent dans l’instruction après l’objet y, séparées par un point, par exemple `y.max()`.<br><br>Il n'est pas toujours aussi simple de déterminer les paramètres de fenêtrage adaptés pour un jeu de données. Pour un jeu de données comme celui-ci, il est encore possible de les estimer en regardant les données et en choisissant des limites.<br>Pour un jeu de données plus volumineux et/ou non structuré, il est préférable d'utiliser une méthode plus automatisée : du code Python qui le fera pour vous.<br><br>Essayez dès maintenant cette méthode !

```python
# minima en maxima bepalen

x_min = x.min()
x_max = x.max()

y_min = y.min()
y_max = y.max()

print(x_min, x_max, y_min, y_max)
```

Tes estimations correspondent-elles à ce que le script Python a trouvé ?

Maintenant que vous connaissez les limites de l'ensemble de données, vous pouvez les saisir afin de créer une fenêtre de tracé appropriée.

<div class="alert alert-block alert-danger">
Il est toujours préférable de ne pas créer de graphiques trompeurs et donc d'afficher également 0 sur l'axe des ordonnées. Sinon, les hausses et les baisses éventuelles sont fortement amplifiées.</div>

Pour indiquer quelles données sont affichées sur le graphique, placez une étiquette sur les axes x et y à l'aide des fonctions `xlabel()` et `ylabel` de pyplot.<br>Donnez également au graphique le titre 'Diagramme de dispersion' et, si vous le trouvez un peu petit, vous pouvez bien sûr toujours utiliser l’argument `figsize=(a,b)` pour agrandir le graphique.<br><br>Complétez le code et exécutez-le.

```python
# oefening

plt.figure()                                       # creëer tekenvenster

plt.axis(xmin=25, xmax=120, ymin=..., ymax=...)    # vensterinstellingen
plt.xticks(np.arange(0, 120, step=10))
plt.yticks(np.arange(..., ..., step=10))

plt.scatter(x, y)                                  # plot punten

plt.title(...)                                     # titel
plt.xlabel("x")                                    # omschrijving bij horizontale as
plt.ylabel("y")                                    # omschrijving bij verticale as    

plt.show()                                         # toon grafiek
```

Assurez-vous que tous les points sont correctement positionnés ! Vérifiez bien le dernier point (113, 100) !<br><br>

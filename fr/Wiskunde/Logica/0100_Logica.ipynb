<div class="alert alert-box alert-success">
Dans ce notebook, vous apprendrez à générer des tables de vérité avec Python.</div>

<div class="alert alert-box alert-info">
Une proposition est représentée par $p, q,  r \;...$ <br>Une proposition est <b>vraie</b> ou <b>non vraie (fausse)</b>. Cela correspond aux valeurs booléennes <b>True</b> et <b>False</b>. <br> Une proposition vraie est souvent indiquée par <b>$1$</b> et une proposition fausse par <b>$0$</b>.<br><br>Vous pouvez également composer des propositions, par exemple au moyen de <b>et</b> ($\land$) et de <b>ou</b> ($\lor$). Vous pouvez aussi utiliser <b>non</b> (~ ou $\neg$); ceux-ci correspondent à <b><span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">and</span></b>, <b><span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">or</span></b> et <b><span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">not</span></b> (donc et, ou et non).<br><br>Les flèches sont également importantes : la flèche d'implication ($\Rightarrow$) et la flèche d'équivalence ($\Leftrightarrow$), qui correspondent à <b><span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">implies</span></b>et <b><span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">equivalentto</span></b> (donc <b>implique</b> et  <b>est équivalent à</b>).</div>

<div class="alert alert-box alert-info">
<b>Une proposition $p \wedge q$ n'est vraie que si $p$ et $q$ sont toutes deux vraies.</b> <br>Donc, si l'une des deux est fausse ou si toutes deux sont fausses, alors $p \wedge q$ est faux.<br><br><b>Une proposition $p \vee q$ n'est fausse que si $p$ et $q$ sont toutes deux fausses.</b><br>Donc si l'une des deux est vraie ou si les deux sont vraies, alors $p \vee q$ est vraie.<br>Ce <b>ou</b> est un <b>ou inclusif</b>, et diffère donc de <em>ou exclusif</em>.<br><br><b>L'énoncé $\neg p$ est vrai si $p$ est faux et inversement.</b><br><br><b>L'énoncé $p \Rightarrow q$ n'est faux que si $p$ est vrai et $q$ est faux.</b> <br> Si $p$ est faux, alors $p \Rightarrow q$ est vrai quelle que soit la valeur de vérité de $q$.<br><br><b>L'énoncé $p \Leftrightarrow q$ est vrai chaque fois que $p$ et $q$ ont la même valeur de vérité.</b><br> Donc $p \Leftrightarrow q$ est vrai si $p$ et $q$ sont tous deux vrais ou tous deux faux.</div>

<div class="alert alert-box alert-success">
En Python, vous disposez par défaut des opérateurs logiques <b>and</b>, <b>or</b> et <b>not</b>. Des notations comme ~ et $\neg$, ainsi que des flèches comme $\Rightarrow$ et $\Leftrightarrow$, ne sont pas disponibles par défaut. Mais vous avez par exemple besoin de ces flèches si vous voulez générer une table de vérité pour certaines propositions.</div>

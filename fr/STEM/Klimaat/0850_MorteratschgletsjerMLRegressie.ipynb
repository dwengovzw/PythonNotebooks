Depuis la révolution industrielle, la concentration des gaz à effet de serre dans l'atmosphère n'a cessé d'augmenter. Depuis 1880, la température moyenne mondiale a augmenté d'environ 0,85 °C. Ce réchauffement s'accompagne d'un réchauffement des océans, d'une élévation du niveau de la mer de 20 cm, d'une fréquence accrue des phénomènes météorologiques extrêmes et d'une diminution de 40 % de la banquise arctique. La glace des glaciers fond également, presque partout dans le monde.<br><br>Les eaux de fonte provenant des glaciers de montagne détermineront dans une large mesure de combien le niveau de la mer s’élèvera à l’avenir. Des scénarios possibles évoquent une hausse allant jusqu’à 30 cm due à la fonte des glaciers de montagne. <br>De plus, les glaciers ont un impact sur les ressources en eau locales et sont importants pour le tourisme.<br>La vitesse à laquelle le volume d’un glacier diminue sous l’effet de l’augmentation globale de la température varie d’un glacier à l’autre. Des facteurs locaux jouent en effet un rôle : par exemple l’orientation du glacier, la mesure dans laquelle le glacier est à l’ombre ... [1].

Le glacier de Morteratsch, long de 6 km, se trouve en Suisse et est à l'ombre des sommets environnants pendant une grande partie de l'année [1].

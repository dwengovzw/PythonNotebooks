Le Dr. Pieter Tans est le chef du 'Carbon Cycle Greenhouse Gases Group' du centre de recherche 'National Oceanic & Atmospheric Administration (NOAA)' au Colorado. Son groupe de recherche effectue, au Mauna Loa à Hawaï, des mesures de la concentration de CO$_{2}$ dans l'atmosphère. Les données sont disponibles sur le site web de la NOAA [1].

Pieter Tans [2]: "David Keeling du 'Scripps Institute of Oceanography' à San Diego a commencé en 1958 les premières mesures de précision du CO$_{2}$ sur le volcan Mauna Loa dans une cabane du 'U.S. Weather Bureau', qui a maintenant été intégré à la NOAA. Les mesures de CO$_{2}$ de la NOAA ont commencé en 1973. Depuis lors, il existe deux programmes de mesure indépendants : par Scripps et par la NOAA.<br>Les concentrations de CO$_{2}$ de la période antérieure au début des mesures à Mauna Loa peuvent être déterminées à partir d'échantillons d'air naturels, des bulles d'air emprisonnées dans la glace profonde. Le CO$_{2}$ des 2000 dernières années, qui était emprisonné dans la glace profonde, a été extrait en Antarctique."

Le Dr. Pieter Tans nous a fourni des données issues de ses recherches avec les **carottes de glace** de Law Dome [2]. <br>Dans ce notebook, vous allez travailler avec les données issues des carottes de glace ainsi qu'avec les données de Mauna Loa.

La concentration de CO$_{2}$ est exprimée en **ppm** : parties par million.

Le tableau comporte 203 lignes et 2 colonnes. <br>La première colonne contient l'année de la mesure, la deuxième colonne la concentration de CO$_{2}$ en ppm.

Les données obtenues à partir des carottes de glace donnent la concentration de CO$_{2}$ de l'an 1 à l'an 1996.

Pour tracer ces données dans un graphique, choisissez *deux variables* : x pour l'année et y pour la concentration de CO$_{2}$. <br>En d'autres termes, vous créez une liste des années et une liste des concentrations de CO$_{2}$. Avec pandas, c'est assez simple, car vous pouvez utiliser l'en-tête du tableau comme **clé** (*key*).

```python
x = antarctica["jaartal"]                # hoofding van kolom is sleutel, die speelt rol van index, jaartal komt op x-as
y = antarctica["molfractie CO2 (ppm)"]   # concentratie CO2 komt op y-as
```

Voyez à quoi ressemble une telle liste :

```python
print(x)
```

C'est encore une sorte de tableau. En fait, pour x, tu n'as besoin que de la deuxième colonne.<br> Tu peux obtenir cela en utilisant la *conversion de type* : en convertissant ce tableau Pandas en un *tableau NumPy*, tu obtiens le résultat souhaité. C'est une étape très importante.

```python
# data omzetten naar NumPy array
x = np.array(x)
y = np.array(y)
```

```python
print(x)
print(y)
```

Pour déterminer l'intervalle sur l'axe des y, il est intéressant de savoir quelle est la plus petite et quelle est la plus grande valeur de y. <br>Vous utilisez pour cela les méthodes `min()` et `max()`. <h2>3. Le petit âge glaciaire</h2>    </font>
</div>

Les carottes de glace fournissent également des informations importantes sur la période précédant la révolution industrielle. Le CO$_{2}$ atmosphérique était alors principalement déterminé par des processus naturels, mais était peut-être déjà quelque peu soumis aux activités humaines telles que la déforestation, l'agriculture et la guerre [4].

Le Petit Âge glaciaire est une période allant du XVIe siècle jusqu'au milieu du XIXe siècle. Des périodes plus froides sont survenues dans le monde entier, mais elles ne se sont pas produites de manière synchrone. Surtout au XVIIe siècle, il faisait plus froid en Europe du Nord-Ouest. La période se caractérisait par exemple par l'expansion des glaciers de montagne. <br>De nombreuses récoltes ont échoué. On a signalé davantage de faim et de maladies, ainsi qu’une mortalité infantile plus élevée [5].<img src="images/bruegeljagersindesneeuw.jpg" alt="Banner" width="400"/>

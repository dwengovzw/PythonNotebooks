Depuis la révolution industrielle, la concentration de gaz à effet de serre dans l’atmosphère a augmenté de façon régulière. Depuis 1880, la température moyenne mondiale a augmenté d’environ 0,85 °C. Ce réchauffement s’accompagne d’un réchauffement des océans, d’une élévation du niveau de la mer de 20 cm, d’une occurrence plus fréquente de phénomènes météorologiques extrêmes et d’une diminution de 40 % de la banquise arctique. La glace des glaciers fond également, presque partout dans le monde.<br><br>L’eau de fonte provenant des glaciers de montagne déterminera dans une large mesure de combien le niveau de la mer augmentera à l’avenir. Des scénarios possibles évoquent une hausse pouvant atteindre 30 cm due à la fonte des glaciers de montagne. <br>En outre, les glaciers ont un impact sur les ressources en eau locales et ils sont importants pour le tourisme.<br>La vitesse à laquelle le volume d’un glacier diminue sous l’influence de l’augmentation de la température globale varie d’un glacier à l’autre. Des facteurs locaux y jouent en effet un rôle: p. ex. l’orientation du glacier, la mesure dans laquelle le glacier est à l’ombre ... [1].

Le glacier de Silvretta, long de 3 km, est situé dans le nord-est de la Suisse. Depuis 1954, le glacier a déjà perdu 400 m de longueur [3].

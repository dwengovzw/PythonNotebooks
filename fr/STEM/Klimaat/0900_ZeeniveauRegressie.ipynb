Le niveau de la mer est influencé notamment par l'augmentation du volume de la masse d'eau due à la hausse des températures, la fonte des calottes glaciaires et des glaciers, ainsi que par les modifications du stockage des eaux de surface et des eaux souterraines. Le changement climatique mondial entraînera encore au cours de ce siècle une élévation du niveau de la mer de 18 à 59 cm [1].<br>
**Nous examinons l’évolution du niveau de la mer le long de la côte belge depuis 1951. Ostende est le point de mesure sur notre côte avec la plus longue série de mesures ininterrompue.**

L'altitude d'un point est mesurée par rapport au niveau de la mer. <br>Cependant, le niveau de la mer n'est pas toujours au même niveau. Les marées entraînent une différence d'environ quatre mètres sur la côte belge. <br>On a donc besoin d'un point de référence. Le point de référence TAW se situe 2,33 mètres en dessous du point de référence NAP. Pour comparer les mesures altimétriques nationales entre elles, il faut tenir compte des différents points de référence [2].<br><br>**Le niveau de la mer est exprimé en mm RLR (Référence locale révisée); les données par rapport à la référence locale sont converties en données par rapport au niveau de référence international.**

Les données sont **standardisées** de la manière suivante : pour chaque donnée d'entraînement, on soustrait la moyenne puis on divise le résultat par l'écart-type. En d'autres termes, on calcule le score Z pour l'ensemble des données d'entraînement. <br> Ainsi, la plupart des valeurs se situent entre -1 et 1. <br>Notez que l'on standardise l'ensemble du jeu de données de la même manière. On fait donc exactement la même chose avec les données de test : on utilise aussi la moyenne et l'écart-type des **données d'entraînement**.

<div class="alert alert-box alert-info">
Effectuer des calculs avec des nombres assez grands conduit rapidement à des nombres encore plus grands et à une instabilité numérique, c’est l’une des raisons pour lesquelles les données sont standardisées. La standardisation des variables consiste à mettre les variables à l’échelle de telle manière que l’on puisse comparer ou relier entre elles des variables, p. ex., d’ordres de grandeur différents ou exprimées dans des unités différentes. La corrélation entre des données bivariées, p. ex., peut être estimée visuellement en examinant le nuage de points correspondant. Cependant, la forme de ce nuage n’est fiable que si les données sont standardisées. De plus, certains algorithmes d’apprentissage automatique ne sont utilisables que si les données sont standardisées, car ces algorithmes sont conçus ainsi.    </div>

<h2>5. Courbe polynomiale de degré 10 comme courbe de régression</h2>    </font>

```python
def tiendegr(x, a, b, c, d, e, f, g, h, i, j, k):
    """Voorschrift van veeltermfunctie van tiende graad met parameters a, b, c en d."""
    return a * x**10 + b * x**9 + c * x**8 + d * x**7 + e * x**6 + f * x**5 + g * x**4 + h * x**3 + i * x**2 + j * x + k

def tiendegraadsreg(x, y):
    popt, pcov = curve_fit(tiendegr, x, y)
    a, b, c, d, e, f, g, h, i, j, k = popt
    return a, b, c, d, e, f, g, h, i, j, k
```

Dr. Pieter Tans est le responsable du 'Carbon Cycle Greenhouse Gases Group' du centre de recherche 'National Oceanic & Atmospheric Administration (NOAA)' au Colorado. Son groupe de recherche effectue à Mauna Loa, à Hawaï, des mesures de la concentration de CO$_{2}$ dans l'atmosphère. Les données sont disponibles sur le site web de la NOAA [1].

Pieter Tans [2]: "David Keeling du 'Scripps Institute of Oceanography' à San Diego a commencé en 1958 les premières mesures de précision du CO2 sur le volcan Mauna Loa dans une cabane du 'U.S. Weather Bureau', qui fait maintenant partie de la NOAA. Les mesures de CO$_{2}$ de la NOAA ont commencé en 1973. Depuis, il existe deux programmes de mesure indépendants : par Scripps et par la NOAA.<br>Les concentrations de CO$_{2}$ de la période antérieure au début des mesures à Mauna Loa peuvent être déterminées à partir d'échantillons d'air naturels,  des bulles d'air emprisonnées dans la glace profonde. Le CO$_{2}$ des 2000 dernières années, qui était emprisonné dans la glace profonde, a été extrait en Antarctique."

Le Dr Pieter Tans nous a fourni des données issues de ses recherches avec les **carottes de glace** du Law Dome [2]. <br>Dans ce notebook, vous allez travailler avec les données issues des carottes de glace.

La concentration de CO$_{2}$ est exprimée en **ppm** : parties par million.

### Importer les modules nécessaires

Pour pouvoir créer des tableaux clairs à partir de **fichiers csv** - csv signifie **comma separated values** -  vous utilisez le module pandas. <br>Pour convertir les colonnes de ce tableau en une liste et pouvoir effectuer des calculs, importez le module NumPy. <br>Pour réaliser des représentations graphiques des données, on utilise l'interface pyplot du module Matplotlib. <br><br>Vous importez les modules nécessaires en exécutant la cellule de code suivante.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

### Vous examinez la variation du dioxyde de carbone atmosphérique et l'évolution de la température moyenne mondiale pour la période de 1880 à 1996.

    <font color=#690027 markdown="1">
<h2>1. Variation du dioxyde de carbone atmosphérique</h2>    </font>

<div>
    <font color=#690027 markdown="1">
<h3>1.1 Lecture des données de dioxyde de carbone issues des carottes de glace</h2>    </font>
</div>

Les données nous ont été fournies sous forme de fichier CSV. C’est un format de fichier très utilisé.Cela se présente comme suit :

<img src="images/csvantarctica.jpg" alt="Banner" width="250"/>

Lisez le fichier en exécutant la cellule de code suivante :

```python
antarctica = pd.read_csv("data/antarcticaCO2.csv")         # data Tans (NOAA)
```

Et récupérez le résultat :

```python
antarctica
```

Le tableau a 203 lignes et 2 colonnes. <br>La première colonne contient l’année de la mesure, la deuxième colonne la concentration de CO$_{2}$ en ppm.

Les données issues des carottes de glace donnent la concentration de CO$_{2}$ de l’an 1 jusqu’à l’an 1996.

Pour représenter ces données dans un graphique, choisissez deux variables : x pour l'année et y pour la concentration de CO$_{2}$. <br>Vous créez une liste des années et une liste des concentrations de CO$_{2}$. Avec Pandas, cela se fait assez facilement parce que vous pouvez utiliser l'en-tête du tableau comme **clé** (*key*).

```python
x = antarctica["jaartal"]                # hoofding van kolom is sleutel, die speelt rol van index, jaartal komt op x-as
y = antarctica["molfractie CO2 (ppm)"]   # concentratie CO2 komt op y-as
```

```python
# data omzetten naar NumPy array
x = np.array(x)
y = np.array(y)
```

```python
print(x)
print(y)
```

<div>
    <font color=#690027 markdown="1">
<h3>1.2 Données de concentration atmosphérique de CO$_{2}$ pour la période 1882-1996</h3>    </font>
</div>

Vous avez donc besoin de la concentration de CO$_{2}$ pour la période 1880-1996.<br> Pour cela, vous devez restreindre à cette période à la fois le tableau NumPy des années et le tableau NumPy des concentrations de CO$_{2}$.

Tu commences par déterminer l’indice de l’élément dans les tableaux qui correspond à l’année 1880.

```python
np.where(x == 1880)
```

Oups, aucun résultat. Va jeter un coup d'œil à la ligne affichée. Tu verras alors que 1880 en tant que tel n'y figure pas, mais bien 1.8770e+03 et 1.8820e+03. <br>Essayez avec une année autour de 1880.

```python
np.where(x == 1882)
```

Vérifier à nouveau:

```python
print(x[107])
```

```python
# data beperken tot de gevraagde periode
x_CO2 = x[107:]
y_CO2 = y[107:]
```

```python
print(x_CO2, y_CO2)
```

<div>
    <font color=#690027 markdown="1">
<h3>1.3 Afficher les données de dioxyde de carbone sur un graphique</h2>    </font>
</div>

Il est toujours bon de visualiser les données. Cela rend les données un peu plus concrètes.

La plage nécessaire sur l'axe des x est facile à voir. Pour déterminer la plage sur l'axe des y, il est intéressant de connaître la plus petite et la plus grande valeur de y. <br>Pour cela, vous utilisez les méthodes `min()` et `max()`.

```python
y_CO2_min = np.min(y_CO2)
y_CO2_max = np.max(y_CO2)
print(y_CO2_min, y_CO2_max)
```

Il est toujours préférable de ne pas créer de graphiques trompeurs et donc d'afficher également 0 sur l'axe des ordonnées. Sinon, les hausses et les baisses éventuelles sont fortement amplifiées.

```python
# grafiek

plt.figure()                                    # grafiekscherm

plt.xlim(1850, 2010)                            # bereik op x-as
plt.ylim(280, 370)                              # bereik op y-as
plt.title("CO2 uit ijskernen voor 1882-1996")    # titel geven aan grafiek
plt.xlabel("jaartal")                            # omschrijving geven bij x-as
plt.ylabel("CO2 (in ppm)")                       # omschrijving geven bij y-as

plt.plot(x_CO2, y_CO2, color="blue")   # puntenwolk, kleur en vorm van punten vastleggen

plt.show()                                       # grafiek tonen
```

        <h2>2. Évolution de la température</h2>    </font>
</div>

<div>
    <font color=#690027 markdown="1">
        <h3>2.1 Données de température</h3>    </font>
</div>

Vous avez également besoin de données sur la température moyenne mondiale pour la période 1880-1996.

Les données que vous allez utiliser sont disponibles sur le site web de la NASA/GISS [3]. Ce ne sont cependant pas les températures mesurées qui constituent les données. Les données indiquent dans quelle mesure la température mesurée, en °C, s’écarte par rapport à une certaine période de référence. On parle d’une **anomalie de température**. Pour les données de la NASA/GISS, la période de référence est 1951-1980.

```python
globaletemp = pd.read_csv("data/globaletemperatuur.csv")           # data NASA
```

```python
print(globaletemp)
```

Notez que ces mesures s'étendent jusqu'en 2018, et vous n'en avez besoin que jusqu'en 1996.

<div>
    <font color=#690027 markdown="1">
        <h3>2.2 Données sur l’anomalie de la température globale pour la période 1880-1996</h3>    </font>
</div>

```python
x_globaletemp = globaletemp["jaar"]
y_globaletemp = globaletemp["temperatuur"]
```

```python
x_globaletemp = np.array(x_globaletemp)
y_globaletemp = np.array(y_globaletemp)
```

```python
np.where(x_globaletemp == 1996)
```

```python
print(x_globaletemp[116])
```

```python
# data beperken tot de gevraagde periode
x_temp = x_globaletemp[:117]
y_temp = y_globaletemp[:117]
```

```python
print(x_temp, y_temp)
```

<div>
    <font color=#690027 markdown="1">
        <h3>2.3 Afficher les données d'anomalie de température pour 1880-1996 sur un graphique</h3>    </font>
</div>

```python
y_temp_min = np.min(y_temp)
y_temp_max = np.max(y_temp)
print(y_temp_min, y_temp_max)
```

```python
# grafiek

plt.figure()                                  # grafiekscherm

plt.xlim(1850, 2010)                          # bereik op x-as
plt.ylim(-0.6, 0.6)                               # bereik op y-as
plt.title("Temperatuuranomalie voor 1880-1996")    # titel geven aan grafiek
plt.xlabel("jaartal")                              # omschrijving geven bij x-as
plt.ylabel("temperatuuranomalie (°C)")             # omschrijving geven bij y-as

plt.plot(x_temp, y_temp, color="green")   # puntenwolk, kleur en vorm van punten vastleggen

plt.show()                                         # grafiek tonen
```

<div>
    <font color=#690027 markdown="1">
<h2>3. Évolution de l'anomalie de température et variation du dioxyde de carbone atmosphérique pour la période 1880-1996 sur un seul graphique</h2>    </font>
</div>

```python
# grafiek met twee krommen met verschillende schaal op y-as 

# twee subgrafieken op eenzelfde figuur
fig, ax1 = plt.subplots(figsize=(8, 6))      # grafiek met een eerste assenstelsel
ax2 = ax1.twinx()                            # een tweede assenstelsel, met een gemeenschappelijke x-as met het eerste

# bereik x-as
plt.xlim(1850, 2010)
ax1.set_xlabel("jaartal")

# bereik y-assen
ax1.set_ylim(280, 370)
ax2.set_ylim(-0.6, 0.6)
ax1.set_ylabel("CO$_{2}$ (ppm)", color="blue")
ax2.set_ylabel("temperatuur anomalie (°C)\n(relatief t.o.v. 1951-1980)", color="green")

plt.title("Temperatuuranomalie en CO$_{2}$ (1882-1996)")

ax1.plot(x_CO2, y_CO2, color="blue")
ax2.plot(x_temp, y_temp, color="green")

plt.show()
```

<div class="alert alert-box alert-success">
Le niveau de la mer à Ostende est mesuré depuis 1951. Dans ce notebook, l'<em>évolution du niveau de la mer</em> à Ostende est examinée.<br></div>

Le niveau de la mer est influencé notamment par l'augmentation du volume de la masse d'eau due à une hausse de la température, la fonte des calottes glaciaires et des glaciers, la modification du stockage des eaux de surface et des eaux souterraines. Le changement climatique mondial entraînera au cours de ce siècle une hausse du niveau de la mer de 18 à 59 cm [1].<br>
**Nous examinons l'évolution du niveau de la mer le long de la côte belge depuis 1951. Ostende est le point de mesure sur notre littoral qui présente la plus longue série de mesures ininterrompue.**

L'altitude d'un point est mesurée par rapport au niveau de la mer. <br>Cependant, le niveau de la mer n’est pas toujours au même niveau. Les marées engendrent une différence d’environ quatre mètres sur la côte belge. <br>Il faut donc un point de référence. Le niveau moyen de la mer à marée basse à Ostende est utilisé comme niveau zéro : le Deuxième Nivellement Général (TAW). Aux Pays-Bas, on a pris le niveau moyen de la mer entre la marée basse et la marée haute : le Normaal Amsterdams Peil (NAP). Le point de référence TAW se situe 2,33 mètres en dessous du point de référence NAP. Pour comparer les mesures altimétriques nationales entre elles, il faut tenir compte des différents points de référence [2].<br><br>**Le niveau de la mer est exprimé en mm RLR (Revised Local Reference); les données par rapport à la référence locale sont converties en données par rapport au niveau de référence international.**

Depuis la révolution industrielle, la concentration des gaz à effet de serre dans l’atmosphère a augmenté progressivement. Depuis 1880, la température moyenne mondiale a augmenté d’environ 0,85 °C. Ce réchauffement s’accompagne d’un réchauffement des océans, d’une élévation du niveau de la mer de 20 cm, d’une fréquence accrue des phénomènes météorologiques extrêmes et d’une diminution de 40 % de la banquise arctique. La glace des glaciers fond elle aussi, presque partout dans le monde.<br><br>L'eau de fonte provenant des glaciers de montagne déterminera en grande partie de combien le niveau de la mer augmentera à l'avenir. Des scénarios possibles évoquent une hausse pouvant atteindre 30 cm due à la fonte des glaciers de montagne. <br>De plus, les glaciers ont un impact sur les ressources en eau locales et sont importants pour le tourisme.<br>La vitesse à laquelle le volume d’un glacier diminue sous l’effet de l’augmentation globale de la température diffère d’un glacier à l’autre. Des facteurs locaux y jouent en effet un rôle : par exemple l’orientation du glacier, son degré d’ombrage ... [1].

Le glacier de Silvretta, long de 3 km, est situé dans le nord-est de la Suisse. Depuis 1954, le glacier a déjà perdu 400 m de longueur [3].

Lander Van Tricht (VUB) nous a fourni des données issues de ses recherches sur l'évolution du glacier de la Silvretta. Depuis 1919, on enregistre de combien de mètres le glacier s'amincit chaque année [2].

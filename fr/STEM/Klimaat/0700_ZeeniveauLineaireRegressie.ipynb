Le niveau de la mer est influencé notamment par l’augmentation du volume de la masse d’eau avec la hausse de la température, la fonte des calottes glaciaires et des glaciers, la modification du stockage des eaux de surface et des eaux souterraines. Le changement climatique mondial entraînera au cours de ce siècle une élévation du niveau de la mer de 18 à 59 cm [1].<br>
**Nous examinons l'évolution du niveau de la mer le long de la côte belge depuis 1951. Ostende est le point de mesure sur notre côte avec la plus longue série de mesures ininterrompue.**

L'altitude d'un point est mesurée par rapport au niveau de la mer. <br>Cependant, le niveau de la mer n'est pas toujours le même. La marée basse et la marée haute provoquent un écart d'environ quatre mètres sur la côte belge. <br>Il faut donc un point de référence. Le niveau moyen de la mer à marée basse à Ostende est utilisé comme niveau zéro : le Deuxième Nivellement Général (TAW). Aux Pays-Bas, on a pris le niveau moyen de la mer entre la marée basse et la marée haute : le Normaal Amsterdams Peil (NAP). Le point de référence TAW se situe 2,33 mètres en dessous du point de référence NAP. Pour comparer les mesures d'altitude nationales entre elles, il faut tenir compte des différents points de référence [2].<br><br>**Le niveau de la mer est exprimé en mm RLR (Référence locale révisée) ; les données par rapport à la référence locale sont converties en données par rapport au niveau de référence international.**

Le niveau de la mer à Ostende est mesuré depuis 1951. Les valeurs de ces mesures peuvent être consultées sur le site web de la Vlaamse Milieumaatschappij [3]. <br><br>Pour ce notebook, les données sont disponibles dans le fichier `zeeniveau.csv` dans le dossier `data`. <br>Les données comprennent, pour chaque enregistrement, le niveau moyen annuel de la mer (en mm RLR) à Ostende et l’année correspondante.

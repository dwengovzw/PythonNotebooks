Depuis la révolution industrielle, la concentration de gaz à effet de serre dans l’atmosphère n’a cessé d’augmenter. Depuis 1880, la température moyenne mondiale a augmenté d’environ 0,85 °C. Ce réchauffement s’accompagne d’un réchauffement des océans, d’une élévation du niveau de la mer de 20 cm, d’une fréquence accrue des phénomènes météorologiques extrêmes et d’une diminution de 40 % de la banquise arctique. La glace des glaciers fond également, presque partout dans le monde.<br><br>L'eau de fonte provenant des glaciers de montagne déterminera en grande partie de combien le niveau de la mer augmentera à l'avenir. Des scénarios possibles évoquent une hausse pouvant atteindre 30 cm due à la fonte des glaciers de montagne. <br>De plus, les glaciers ont un impact sur les ressources en eau locales et ils sont importants pour le tourisme.<br>La vitesse à laquelle le volume d’un glacier diminue sous l’effet de la hausse globale des températures varie d’un glacier à l’autre. Des facteurs locaux y jouent en effet un rôle : p. ex. l’orientation du glacier, la mesure dans laquelle le glacier est à l’ombre… [1].

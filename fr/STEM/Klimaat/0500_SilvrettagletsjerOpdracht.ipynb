Depuis la révolution industrielle, la concentration de gaz à effet de serre dans l’atmosphère a augmenté de façon constante. Depuis 1880, la température moyenne mondiale a augmenté d’environ 0,85 °C. Ce réchauffement s’accompagne d’un réchauffement des océans, d’une élévation du niveau de la mer de 20 cm, d’une fréquence accrue des phénomènes météorologiques extrêmes et d’une diminution de 40 % de la banquise arctique. La glace des glaciers fond également, presque partout dans le monde.<br><br>L'eau de fonte issue des glaciers de montagne déterminera dans une large mesure de combien le niveau de la mer s'élèvera à l'avenir. Des scénarios possibles évoquent une hausse pouvant atteindre 30 cm due à la fonte des glaciers de montagne. <br>De plus, les glaciers ont un impact sur les ressources en eau locales et sont importants pour le tourisme.<br>La vitesse à laquelle le volume d’un glacier diminue sous l’influence de l’augmentation de la température globale varie d’un glacier à l’autre. Des facteurs locaux y jouent en effet un rôle : par exemple l’orientation du glacier, la mesure dans laquelle le glacier se trouve à l’ombre ... [1].

Le glacier de Silvretta, long de 3 km, se situe dans le nord-est de la Suisse. Depuis 1954, le glacier a déjà perdu 400 m de longueur [3].

Lander Van Tricht (VUB) nous a fourni des données de son étude sur l'évolution du glacier de Silvretta. Depuis 1919, on a enregistré de combien de mètres le glacier s'amincit chaque année [2].

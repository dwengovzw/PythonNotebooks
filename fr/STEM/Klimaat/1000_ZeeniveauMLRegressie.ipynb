Dans ce notebook, le niveau futur de la mer à Ostende est prédit à l'aide d'une <em>régression</em>.<br>Pour ce faire, on applique une technique issue de <em>l'apprentissage automatique</em>. Ce qui est typique de cette méthode, c'est qu'on divise les données en <em>données d'entraînement et données de test</em>. Les données sont également toujours <em>standardisées</em>. On examinera, à l'aide des données de test, si le système prédit avec une précision suffisante. <br>Les fonctionnalités du module Python <em>scikit-learn</em> sont utilisées. <br>Le phénomène de <em>surapprentissage</em> est illustré.</div>

Le niveau de la mer est influencé, entre autres, par l’augmentation du volume de la masse d’eau avec la hausse de la température, la fonte des calottes glaciaires et des glaciers, ainsi que par la modification du stockage des eaux de surface et des eaux souterraines. Le changement climatique global entraînera d’ici la fin de ce siècle une élévation du niveau de la mer de 18 à 59 cm [1].<br>
**Nous examinons l'évolution du niveau de la mer le long de la côte belge depuis 1951. Ostende est le point de mesure de notre littoral qui dispose de la plus longue série de mesures ininterrompue.**

L'altitude d'un point est mesurée par rapport au niveau de la mer. <br>Cependant, le niveau de la mer n'est pas toujours au même niveau. Les marées provoquent une différence qui atteint environ quatre mètres sur la côte belge. <br>Il faut donc un point de référence. Le niveau moyen de la mer à marée basse à Ostende est utilisé comme niveau zéro : le Deuxième Nivellement Général (TAW). Aux Pays-Bas, on a pris le niveau moyen de la mer entre la marée basse et la marée haute : le Normaal Amsterdams Peil (NAP). Le point de référence TAW se situe 2,33 mètres en dessous du point de référence NAP. Pour comparer les mesures altimétriques nationales, il faut tenir compte des différents points de référence [2].<br><br>**Le niveau de la mer est exprimé en mm RLR (Référence locale révisée) ; les données par rapport à la référence locale sont converties en données par rapport au niveau de référence international.**

Le niveau de la mer à Ostende est mesuré depuis 1951. Les valeurs de ces mesures peuvent être consultées sur le site web de l’Agence flamande de l’environnement [3]. <br><br>Pour ce notebook, les données sont disponibles dans le fichier `zeeniveau.csv` dans le dossier `data`. <br>Les données contiennent pour chaque année le niveau moyen annuel de la mer (en mm RLR) à Ostende et l'année correspondante.

<div class="alert alert-box alert-success">
Ce notebook contient un exemple concret d’un <em>jeu de données étiqueté important</em> d’images en niveaux de gris de chiffres manuscrits. </div>

Le jeu de données MNIST contient 70 000 images en niveaux de gris au format 28x28 pixels ; il y a 10 classes avec des étiquettes de 0 à 9 inclus. Ces images ont été collectées par le National Institute of Standards and Technology (le NIST dans l'acronyme MNIST) dans les années 80.
Cet ensemble de données se compose de 60 000 images pour entraîner un réseau de neurones et  10 000 images pour tester le réseau. <br>
Le jeu de données est notamment disponible via le module Python <em>Keras</em>.

### Importer les modules nécessaires

```python
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.datasets import mnist   # MNIST-dataset
```

<div style='color: #690027;' markdown="1">
    <h2>1. Chargement des données</h2></div>

Les données MNIST sont constituées d’un couple. De plus, chacun des deux éléments du couple est lui-même un couple. 
Les données sont structurées comme suit :-  (trainingdata, catégories correspondantes) dans le premier élément;-  (données de test, catégories correspondantes) dans le deuxième élément.
Vous chargez le jeu de données avec l'instruction `mnist.load_data()`. <br>`load_data()` est une fonction de `mnist`. <br>Entre-temps, vous nommez quatre variables qui renvoient chacune à un type de données particulier dans le jeu de données.

```python
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()      # elementen van tupels krijgen juiste naam
```

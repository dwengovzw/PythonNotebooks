<h3>2.2 Multiplication scalaire</h3>    </font>
</div>

La multiplication scalaire est la multiplication par un scalaire, ou autrement dit par un nombre. <br>Si l’on multiplie une matrice par un nombre, alors chaque élément de la matrice est multiplié par ce nombre.<br> Le résultat est donc une matrice de même dimension (mathématique) que la matrice d’origine.

<h3>2.3 Augmenter chaque élément de 50</h3>    </font>
</div>

Deux matrices de même dimension (mathématique) peuvent être additionnées : tous les éléments correspondants sont alors additionnés.<br> Le résultat est alors une matrice ayant à nouveau la même dimension (mathématique).<br>Vous pouvez donc additionner à une matrice une matrice de même taille dont tous les éléments ont la même valeur, par exemple 50, comme dans l’exercice suivant.

#### Exercice 2.3-  Que se passera-t-il avec les couleurs dans boom2 si vous augmentez toutes les valeurs de pixels de 50 ?

Réponse: 

- Pour vérifier cela avec Python, exécutez la cellule de code suivante.

```python
print(boom2 + 50)                     # elk element van matrix vermeerderd met zelfde getal
plt.imshow(boom2 + 50, vmin=0, vmax=255, cmap="gray") 
```

Notez que dans le code il n’y a que `+ 50`, et non pas une matrice dont tous les éléments sont égaux à 50. Python l’interprète toutefois comme une matrice dont tous les éléments valent 50 et ayant les mêmes dimensions (mathématiques) que la matrice à laquelle on doit l’ajouter.

        <h2>3. Transposer </h2>    </font>
</div>

Une matrice transposée est la matrice obtenue en transformant les lignes en colonnes et les colonnes en lignes, tout en conservant l'ordre.

Exemple : considérez la matrice $\begin{bmatrix} -1 & 0 & 0 & 3 \\ 2 & -5 & 12 & 4 \\ 0 & 4 & -2 & 7\end{bmatrix} $.<br>La matrice transposée $\begin{bmatrix} -1 & 0 & 0 & 3 \\ 2 & -5 & 12 & 4 \\ 0 & 4 & -2 & 7\end{bmatrix}^{\large{T}} = \begin{bmatrix} -1 & 2 & 0 \\ 0 & -5 & 4 \\ 0 & 12 & -2 \\ 3 & 4 & 7 \end{bmatrix} $.

Vous pouvez voir ci-dessous l'effet de la transposition sur l'image correspondant à la matrice.

```python
print(boom.T)                     # getransponeerde matrix: rijen worden kolommen en kolommen worden rijen
plt.imshow(boom.T, cmap="gray") 
```

<div class="alert alert-block alert-info">
Pour l’humain, c’est le même arbre, mais tourné de 90° dans le sens inverse des aiguilles d’une montre et symétrisé par rapport à un axe horizontal. Pour l’ordinateur, il s’agit toutefois d’une matrice tout à fait différente.<br> Dans l’apprentissage automatique, on utilise de telles techniques pour créer des données supplémentaires (<em>augmentation de données</em>). Le jeu de données dont on dispose est complété par les mêmes images, mais réfléchies, tournées, etc.  </div>

<h2>4. Mise en miroir</h2></div>

Dans la cellule de code suivante, l’arbre est *réfléchi* par rapport à un axe horizontal.

```python
boom_x = np.flipud(boom)     # spiegeling om horizontale as (upside - down)

plt.figure(figsize=(6,9))
plt.subplot(1, 2, 1)
plt.imshow(boom, cmap="gray")
plt.subplot(1, 2, 2)
plt.imshow(boom_x, cmap="gray")
```

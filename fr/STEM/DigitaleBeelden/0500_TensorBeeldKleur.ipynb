        <h3>2.2 RGB</h3>    </font>
</div>

Dans une image en couleurs, la couleur de chaque pixel est représentée en **code RVB** : trois nombres représentant respectivement l'intensité du rouge, du vert et du bleu.<br>Si vous 'assemblez' maintenant ces trois matrices en les plaçant côte à côte, vous obtenez un **tenseur** qui représente l'arbre en couleur. La première matrice détermine l'intensité du rouge, la deuxième matrice l'intensité du vert et la troisième matrice l'intensité du bleu.<br> Pour déterminer les couleurs d'un pixel, vous formez des triplets avec les éléments correspondants des trois matrices.<br><br>Par exemple:-  Les pixels de la ligne du bas ont alors $RGB(0,0,0)$ et sont donc noirs.-  Le pixel en haut à droite a pour valeur $RGB(0,0,255)$ et est donc bleu.-  Les pixels de la cime ont $RGB(0,255,0)$ et sont donc verts.

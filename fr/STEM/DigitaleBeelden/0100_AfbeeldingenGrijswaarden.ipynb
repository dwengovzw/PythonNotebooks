<div class="alert alert-block alert-info"> 
Un <b>bit</b> est une unité d'information. Le terme provient de <b>binary digit</b>. C'est une unité qui ne peut prendre que les valeurs 0 et 1. <br>Huit bits constituent un <b>octet</b>. Il existe 256 combinaisons possibles de 0 et 1 qui forment ensemble un octet.</div>

        <h2>5. Répartition des teintes</h2>    </font>
</div>

Exécutez les cellules de code suivantes.

```python
# kleinste, grootste en gemiddelde grijswaarde van 'linkerbovenhoek'
print(linkerbovenhoek.min())
print(linkerbovenhoek.max())
print(linkerbovenhoek.mean())
```

```python
# verdeling grijswaarden bij 'linkerbovenhoek'
plt.hist(linkerbovenhoek.ravel(), bins=130) # ravel zet alle elementen van de matrix naast elkaar, omdat hist met rijmatrix werkt
```

Signification de l'*histogramme*:<br>- On utilise 131 nombres binaires, le plus petit est 74 et le plus grand 204.- Notez que 204-74 = 130. Ce 130 se trouve dans le code : `bins=130`, et correspond au nombre d'intervalles.- Deux matrices ligne sont affichées : la seconde avec tous les nombres naturels de 74 à 204 inclus et la première avec le nombre de fois qu’ils apparaissent.<br>Par ex., la valeur 74 apparaît deux fois et la valeur 75 apparaît zéro fois.- Il y a plus de teintes claires que de très foncées.

Vous pouvez également regrouper les valeurs en intervalles :

```python
plt.hist(linkerbovenhoek.ravel(), bins=11)   # waarden verdelen over 11 intervallen
```

### Exemples

Exécutez les cellules de code suivantes.

```python
# voorbeeld 1
afb = np.full((15, 18), 140, dtype="uint8")    # 15x18-matrix met allemaal zelfde elementen, nl. 140
print(afb)
plt.imshow(afb, vmin=0, vmax=255, cmap="gray")  # vmin en vmax geven het bereik van de waarden die de kleuren weergeven
```

```python
# voorbeeld 2
afb2 = np.zeros((8, 6), dtype="uint8")          # matrix met enkel nullen
print(afb2)
```

```python
plt.imshow(afb2, vmin=0, vmax=255, cmap="gray")
```

```python
# voorbeeld 3
vierpixels = np.array(([50, 140], [140, 50]), dtype="uint8")
print(vierpixels)
```

```python
plt.imshow(vierpixels, vmin=0, vmax=255, cmap="gray")
```

```python
print(vierpixels[1,1])       # element op eerste rij, eerste kolom
```

```python
print(vierpixels.ravel())
```

```python
# voorbeeld 4
zespixels = np.array(([70, 160], [160, 70], [110, 190]), dtype="uint8")
print(zespixels)
```

```python
plt.imshow(zespixels, vmin=0, vmax=255, cmap="gray")
```

```python
print(zespixels.ravel())
```

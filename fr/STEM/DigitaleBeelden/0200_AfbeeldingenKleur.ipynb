### Que sais-tu déjà ?

Dans le notebook « Matrices et images en niveaux de gris », il a été montré comment un ordinateur « regarde » une image.

<div class="alert alert-box alert-info">
Une image numérique est une grille rectangulaire de pixels. <br>Chaque pixel a une certaine couleur. Dans une image en noir et blanc, un pixel est noir ou blanc. Dans une image en niveaux de gris, c'est une valeur de gris. Dans une image en couleur, c'est une couleur.</div>

<div class="alert alert-box alert-info">
On peut enregistrer une image en niveaux de gris de manière à ce qu’elle corresponde à une matrice dont les éléments sont des entiers naturels compris entre 0 et 255. On peut faire en sorte que chacun de ces entiers soit stocké sur un octet en choisissant le type uint8.</div>

<div>
    <font color=#690027 markdown="1">  
        <h2>1. Couleurs</h2>    </font>
</div>

Les ondes électromagnétiques possèdent un large spectre de longueurs d’onde, chaque longueur d’onde correspondant à une couleur différente. La lumière que l’être humain peut voir, la lumière visible, occupe seulement une petite partie du spectre. La lumière visible aux plus petites longueurs d’onde est bleue, celle aux plus grandes longueurs d’onde est rouge.

<img src="images/spectrum.png" alt="Banner" width="800"/>
<center>Figure 2 : Spectre électromagnétique.</center>

En mélangeant la lumière rouge, la verte et la bleue, on peut simuler presque toutes les couleurs. Les photos en couleur peuvent alors être enregistrées dans un *système RVB* (RVB = rouge, vert, bleu).<br>Dans ce cas, trois matrices sont nécessaires : une pour les teintes rouges, une pour les teintes vertes et une pour les teintes bleues. Ces trois grilles sont ensuite empilées pour former une sorte de bloc de nombres. En informatique, on parle d’un *tenseur*.

<div class="alert alert-box alert-info">
<b>Les photos couleur sont stockées dans un système RVB.</b> Dans ce cas, trois matrices sont nécessaires qui, ensemble, forment un <b>tenseur</b>.</div>

### Exemple

Reprenez la photo avec le stomate du notebook 'Matrices et images en niveaux de gris'.

```python
begonia_RGB = plt.imread("images/begoniamicroscoop.jpg")  
plt.imshow(begonia_RGB)
```

```python
print(begonia_RGB)   
print(begonia_RGB.shape)
print(begonia_RGB.dtype)
print(begonia_RGB.ndim)
```

La dimension (mathématique) du tenseur est 181x263x*3*. Cela signifie que trois matrices de dimension 181x263 ont été empilées pour former une *sorte de bloc*.

Ce tenseur est un *tableau 3D*. Le tenseur se compose de trois tableaux 2D, à savoir les trois matrices : la première pour les teintes rouges, puis les teintes vertes et enfin les teintes bleues.<br>Les matrices sont numérotées : 0, 1 et 2.

La matrice rouge est la première matrice dans le tenseur. Avec l'instruction suivante, vous extrayez cette matrice rouge du tenseur.

```python
begonia_R = begonia_RGB[:,:,0]
print(begonia_R)
```

Avec la cellule de code suivante, vous pouvez afficher l'image.

```python
plt.imshow(begonia_R) 
```

Peut-être que tu t’attendais à une image avec beaucoup de rouge.

En Python, vous pouvez utiliser ce qu'on appelle des ***colormaps***. Pour afficher des teintes rouges, vous pouvez par exemple utiliser la colormap *Reds*.

```python
plt.imshow(begonia_R, cmap="Reds")   # autumn, summer en winter zijn mogelijke colormaps voor resp. rood, groen en blauw 
```

### Exercice 2.1À quelle couleur t’attendrais-tu dans le coin supérieur gauche ? Une foncée ou une claire ? Et au niveau du stomate ?

Réponse:

**Les couleurs sont apparemment inversées**: au lieu d'afficher une valeur de la matrice, c'est '255 moins cette valeur particulière' qui est affiché.

Pour mieux comprendre ce qui se passe ici, il est préférable d'en savoir un peu plus sur les valeurs des pixels.<br>Vous pouvez obtenir les plus petites et les plus grandes valeurs de pixels avec les méthodes `min()` et `max()`.

```python
begonia_R.min(),  begonia_R.max()  
```

Pour le vert et le bleu, vous pouvez utiliser les colormaps *Greens* et *Blues*, mais dans l'image les couleurs seront alors également inversées. <br>Tu peux éventuellement essayer ça.

Pour éviter l’inversion, vous pouvez utiliser **d’autres colormaps** qui n’inversent pas les couleurs, comme autumn, summer et winter, en tant que colormaps possibles pour, respectivement, le rouge, le vert et le bleu.

```python
plt.imshow(begonia_R, cmap="autumn")   # autumn, summer en winter zijn mogelijke colormaps voor resp. rood, groen en blauw 
```

Python ramènera l’intervalle des valeurs de pixels à [0, 255], à moins que vous ne l’empêchiez vous-même.<br>Pour éviter la mise à l’échelle des couleurs, indiquez explicitement que les couleurs sont représentées avec une valeur de pixel minimale possible de 0 et une valeur de pixel maximale possible de 255.

```python
plt.imshow(begonia_R, cmap="autumn", vmin=0, vmax=255)    
```

<h2>3. Composer vos propres couleurs</h2>    </font>
</div>

Le rose est très présent dans l'image. Une telle teinte de rose est également composée de teintes rouge, verte et bleue. <br>Pour obtenir un carré de quatre pixels roses, il nous faut trois matrices 'composées' : une pour la teinte rouge, une pour la teinte bleue et une pour la teinte verte. Le code RVB du rose est (230, 125, 156).

```python
rozepixels_R = np.array(([230, 230], [230, 230]), dtype="uint8")
rozepixels_G = np.array(([125, 125], [125, 125]), dtype="uint8")
rozepixels_B = np.array(([156, 156], [156, 156]), dtype="uint8")
print(rozepixels_R)
print(rozepixels_G)
print(rozepixels_B)
```

Assemblez ces trois matrices de sorte qu'elles forment un tenseur.

```python
# matrices samenstellen tot een tensor
rozepixels = np.dstack([rozepixels_R, rozepixels_G, rozepixels_B])
print(rozepixels.shape)
print(rozepixels.dtype)
print(rozepixels.ndim)
```

Regardez les images correspondant aux trois matrices et au tenseur construit.

```python
plt.figure()

plt.subplot(1, 4, 1)
plt.imshow(rozepixels_R, cmap="autumn", vmin=0, vmax=255)
plt.subplot(1, 4, 2)
plt.imshow(rozepixels_G, cmap="summer", vmin=0, vmax=255)
plt.subplot(1, 4, 3)
plt.imshow(rozepixels_B, cmap="winter", vmin=0, vmax=255)
plt.subplot(1, 4, 4)
plt.imshow(rozepixels, vmin=0, vmax=255)

plt.show()
```

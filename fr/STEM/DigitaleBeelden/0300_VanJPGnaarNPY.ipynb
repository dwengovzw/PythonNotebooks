<div class="alert alert-box alert-success">
Dans ce notebook, un exemple est donné montrant comment convertir un fichier jpg en une image en niveaux de gris, représentée par une matrice, et comment enregistrer cette image sous forme de fichier npy.   Vous utilisez pour cela le module scikit-image (skimage) qui fournit des algorithmes pour le traitement d’images numériques et la <em>vision par ordinateur</em>.</div>

### Importer les modules nécessaires

```python
import numpy as np
import matplotlib.pyplot as plt
import scripts.helper as helper
import skimage.color as skc
```

<div class="alert alert-block alert-warning">
Dans les notebooks 'Matrices et images en niveaux de gris' et 'Tenseurs et RVB', on explique comment, respectivement, une image en niveaux de gris et une image en couleur sont représentées pour qu'un ordinateur les comprenne.</div>

<div>
    <font color=#690027 markdown="1">  
        <h2>1. Photo en couleur</h2>    </font>
</div>

Une photo a été prise avec un smartphone d’un chiffre 4 écrit à la main : un 4 écrit au feutre noir sur une feuille blanche. <br>Cette photo a été importée dans Paint et recadrée. Ensuite, le format de la photo a été converti dans Paint en 28 x 28 pixels.<br>Le résultat a été enregistré sous `vier.jpg`.

Chargez la photo et récupérez la dimension mathématique. Affichez la photo.

```python
vier = plt.imread("images/vier.jpg")     
```

```python
vier.shape                    # jpg dus RGB en 3D-tensor
```

```python
plt.imshow(vier)              # zwart op witte achtergrond, 28 op 28 pixels
```

Regardez les trois matrices du tenseur.

```python
print(vier[:,:,0])     # roodtinten
print(vier[:,:,1])     # groentinten
print(vier[:,:,2])     # blauwtinten
```

Notez que les valeurs vont de 0 à 255 inclus.

<div>
    <font color=#690027 markdown="1">  
        <h2>2. Convertir une photo en couleur en image en niveaux de gris</h2>    </font>
</div>

Convertissez la photo en une image en niveaux de gris et examinez l'effet sur la dimension mathématique.

```python
vier_grijs = skc.rgb2gray(vier)         # jpg (RGB) omzetten naar grijswaardenafbeelding
vier_grijs.shape                        # gerepresenteerd door matrix i.p.v. tensor
```

```python
plt.imshow(vier_grijs, cmap="gray")    
```

Regardez la matrice.

```python
print(vier_grijs)
```

Notez que les valeurs sont comprises entre 0 et 1. C’est-à-dire que toutes les valeurs de 0 à 255 inclus ont été divisées par 255. Les valeurs sont **normalisées**, une mise à l’échelle a donc été effectuée. 0 correspond au noir et 1 au blanc.

<div>
    <font color=#690027 markdown="1">  
<h2>3. Inverser les couleurs</h2>    </font>
</div>

En effectuant une opération sur la matrice, vous pouvez inverser les couleurs. En remplaçant 0 par 1, 1 par 0, 0,3 par 0,7 ... vous obtenez un quatre blanc sur un fond noir. <br>Testez ceci en exécutant la cellule de code suivante.

```python
vier_zw = 1 - vier_grijs    # grijswaarden inverteren
print(vier_zw.shape) 
```

```python
plt.imshow(vier_zw, cmap="gray")
```

Examinez la matrice et voyez comment les premiers éléments ont été modifiés.

```python
print(vier_zw)
```

Facultatif : Vous pouvez éventuellement ramener les valeurs à des nombres dans [0,255]. Pour ce faire, procédez comme suit :

```python
vier_zw_0tot255 =  vier_zw * 255   
print(vier_zw_0tot255)
vier_zw_uint = vier_zw_0tot255.astype("uint") 
print(vier_zw_uint)
```

<div>
    <font color=#690027 markdown="1">  
        <h2>4. Enregistrer et charger une image</h2>    </font>
</div>

Avec les bonnes fonctions, vous pouvez stocker la matrice en tant que matrice, donc sans en changer la dimension mathématique.<br>Si vous enregistriez la matrice au format `jpg`, la dimension mathématique changerait : vous obtiendriez alors à nouveau un tenseur.

Vous pouvez toutefois enregistrer le fichier au format npy.

Pour enregistrer dans le dossier `data`, procédez comme suit :

```python
np.save("data/vier.npy", vier_zw)         # opslaan als npy-bestand
```

Voici comment récupérer le fichier npy :

```python
vier2 = np.load("data/vier.npy")          # inlezen van een npy-bestand
```

Vous pouvez maintenant montrer le résultat.

```python
plt.figure()

plt.subplot(1,2,1)
plt.imshow(vier, cmap="gray")
plt.subplot(1,2,2)
plt.imshow(vier2, cmap="gray")

plt.show()
```

```python
# controle formaat
print(vier2.shape)
```

Vous pouvez à nouveau consulter le fichier npy :

```python
print(vier2)
```

En effet, cela n'a pas changé.

    <h1>MESSAGE CACHÉ</h1></div>

<div class="alert alert-block alert-success">
Ce notebook illustre comment un message peut être caché dans une image. En détectant un motif inhabituel dans l'image, on peut découvrir le message.</div>

```python
import numpy as np
import matplotlib.pyplot as plt
import scripts.helper as helper
```

```python
figuurmetboodschap = np.load("data/verborgen.npy")
```

```python
plt.imshow(figuurmetboodschap, cmap="gray")
```

```python
figuurmetboodschap[figuurmetboodschap %2 != 0] = 255    # boodschap verborgen in even pixelwaarden, dus maak oneven pixelwaarden wit
plt.imshow(figuurmetboodschap, cmap="gray")
```

<div class="alert alert-block alert-info">
L'image en niveaux de gris d'origine ne contient que des valeurs de pixels impaires.<br>Ensuite, quelques pixels ont été augmentés d'une unité : ces valeurs de pixels sont donc paires et constituent ensemble le message.<br>En remplaçant enfin toutes les valeurs de pixel impaires par la valeur 255, et donc en les colorant en blanc, seul le message n’est pas blanc et peut être lu.</div>

Comment procédez-vous ?

Visuellement, la différence avec l'image d'origine n'est pas perceptible.<br><br>
Vous fournissez également une image du message : un texte blanc sur un fond noir, le blanc est représenté par la valeur 1 et le noir par 0. Cette image est représentée par une matrice ayant les mêmes dimensions que la première image. <br><br>
Vous créez une nouvelle image en additionnant les deux matrices. Cette nouvelle image ressemble à l'image d'origine mais le message y est caché. En effet, tous les pixels sont inchangés, sauf précisément ceux qui contiennent le texte du message, lesquels sont augmentés d'une unité. Tous les pixels de la nouvelle image sont donc impairs, à l'exception de ceux du message, qui sont pairs.</div>

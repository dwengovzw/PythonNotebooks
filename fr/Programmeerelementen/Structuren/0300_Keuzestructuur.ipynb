### Exemple 1.1Exécutez le code suivant.

```python
# keuzestructuur

getal = float(input("Geef een getal: "))     # na input() steeds aangezien als string, dus typecasting gebruiken

if getal > 0: 
    print(getal, "is strikt positief.")
elif getal < 0:    
    print(getal, "is strikt negatief.")
else:
    print(getal, "is nul.")
```

<div class="alert alert-box alert-info">
La combinaison des mots-clés <b>if - elif - else</b> correspond à <b>si - sinon si - sinon</b>.</div>

### Exemple 2.1Exécutez le code suivant.

```python
# geneste structuur: herhalings- en keuzestructuur

getallen = [3, 7, 1, 13, 2, 4, 5, 11, 15] 

for getal in getallen:
    if getal % 2 == 0:
        print("Het getal", getal, "is even.")
    else: 
        print("Het getal", getal, "is oneven.")
```

### Exemple 2.2Dans la cellule de code suivante, la variable `wiskundigen' fait référence à une liste de noms de mathématiciens et de mathématiciennes. Notez que les noms des femmes commencent par un 'N' ou un 'L', alors que ceux des hommes non. Cette information est utilisée dans le code pour compter le nombre de femmes dans la liste.<br>Exécutez le code suivant.

```python
# geneste structuur

wiskundigen = ["Fermat", "Gauss", "Euler", "Fibonacci", "Galois", "Noether", "Nightingale", "Lovelace"]
aantal_vrouwen = 0 

for persoon in wiskundigen: 
    geslacht = "vrouwelijk"
    if persoon[0] in ["N", "L"]:
        aantal_vrouwen += 1
    else:
        geslacht = "mannelijk"    
    print(persoon, "is van het geslacht:", geslacht)
    
print("Er zijn", aantal_vrouwen, "vrouwelijke wiskundigen in de lijst.")
```

Procédure : Lorsqu’en Python une boucle for parcourt une liste, Python utilisera l’indice d’un élément de cette liste.Si vous supprimez un élément, les éléments de la liste reçoivent un nouvel indice. Ainsi, l’élément suivant l’élément supprimé reçoit l’indice déjà utilisé. Dans la boucle for, la variable (l’itérateur) est cependant incrémentée, de sorte que cet élément suivant est ignoré. <br>Pour éviter cela, il faut d'abord stocker les éléments à supprimer dans une nouvelle liste, puis supprimer un à un tous les éléments de cette liste.<br><br>Vous pouvez également ajouter des instructions print supplémentaires dans votre script afin de voir ce qui se passe. Le cas échéant, vous pouvez ensuite les « supprimer » en plaçant un # devant.

        <h2>1. Boucle for</h2>    </font>
</div>

<div class="alert alert-block alert-info">
Une <b>boucle for</b> est une structure de répétition, plus précisément une répétition bornée.  </div>

#### Exemple 1.1Dans l'exemple suivant, on continue à travailler avec la liste à laquelle la variable `gemengdelijst` fait référence.<br>Te rappelles-tu que `gemengdelijst = ["koekoek", 2.5, -1, [5, "ren"], "?!@", " "]`? <br>En exécutant la cellule de code ci-dessous, chaque élément sera affiché successivement et l'un en dessous de l'autre à l'écran.

```python
gemengdelijst = ["koekoek", 2.5, -1, [5, "ren"], "?!@", " "]
for element in gemengdelijst:
    print(element)
```

Chaque élément de la liste a été parcouru un par un. La même opération est donc répétée. Cette répétition est limitée puisqu'elle s'arrête une fois que tous les éléments de la liste ont été parcourus.

#### Exemple 1.2Dans l'exemple suivant, on continue à travailler avec la liste à laquelle fait référence la variable `ontbijt`.<br>Te rappelles-tu que `ontbijt = ["boter", "koffie", "ochtend", "choco"]`? <br>Une nouvelle liste `ontbijt2` de même longueur est introduite ainsi qu'une liste vide `ontbijt3`.<br>Ensuite, les éléments correspondants des deux listes sont concaténés (fusionnés) ; la liste `ontbijt3` en est remplie. Pour travailler avec des éléments correspondants, tu utilises la méthode `zip()` (pense à une « fermeture éclair »).

```python
ontbijt = ["boter", "koffie", "ochtend", "choco"]
ontbijt2 = ["koek", "kan", "humeur", "pot"]        # lijst met vier elementen

ontbijt3 = []                                      # lege lijst om op te vullen

for woord1, woord2 in zip(ontbijt, ontbijt2):      # beschouw achtereenvolgens de overeenkomstige elementen uit beide lijsten 
    woord = woord1 + woord2
    ontbijt3.append(woord)                         # voeg string waarnaar variabele 'woord' verwijst toe aan lijst 'ontbijt3'  
print(ontbijt3)                                    # toon lijst 'ontbijt3' 
```

        <h2>2. If ... else ... </h2>    </font>
</div>

<div class="alert alert-block alert-info">
Une structure de sélection est une structure qui, dans le code, se reconnaît à <b>if ...</b> ou <b>if ... else ...</b></div>

#### Exemple 2.1Dans cet exemple, on continue à travailler avec la chaîne de caractères `"Hello, world"`, à laquelle faisait référence la variable `zin`.Le script Python marquera les voyelles dans la chaîne.

Pour pouvoir travailler avec la couleur, tu dois d'abord importer un module. Un module contient de nombreuses fonctions que des informaticiens expérimentés ont déjà programmées pour toi.

```python
# module importeren
from colorama import Fore, Back         # om in kleur te kunnen printen
```

```python
# antwoord
zin = "Hello, world"
klinkers = "aeiou"

aangepastezin = ""            # lege string om op te vullen
for karakter in zin:
    if karakter in klinkers:
        aangepastezin = aangepastezin + Back.YELLOW + karakter + Back.RESET  # achtergrond geel maken en na letter weer resetten
    else:
        aangepastezin = aangepastezin + karakter
        
print(aangepastezin)
```

```python
# Alternatief antwoord
zin = "Hello, world"
klinkers = "aeiou"

for karakter in zin:
    if karakter in klinkers:
        zin = zin.replace(karakter, Back.YELLOW + karakter + Back.RESET)  # achtergrond geel maken en na letter weer resetten
        
print(zin)
```

#### Exemple 2.2

Pour les tâches répétitives, une fonction peut également être utile.

```python
# functie opstellen
def opkuisen(tekst):
    """Leestekens en code in een tekst verwijderen."""
    leestekens = string.punctuation
    for code in ["\n", "\ufeff"]:
        tekst = tekst.replace(code, "")
    for karakter in leestekens:
        tekst = tekst.replace(karakter, "")
    return tekst
```

```python
# uittesten
werkwijze = "Verwarm de wraps in de microgolfoven,gedurende dertig seconden.Geniet erna van een bietenwrap met de (zelfbereide) tonijnsla!"
werkwijze_opgekuist = opkuisen(werkwijze)
print(werkwijze)
print(werkwijze_opgekuist)
```

<div>
    <font color=#690027 markdown="1">
        <h2>1. Boucle for</h2>    </font>
</div>

La boucle for est utilisée pour une *répétition bornée*: une instruction est exécutée un certain nombre de fois. Le nombre de fois que l'instruction est exécutée est explicitement indiqué ou est borné par, par exemple, le nombre d'éléments dans une liste.

### Exemple 1.1

Dans le script suivant, il y a deux listes. Une liste `wiskundigen`, dans laquelle les femmes ont été oubliées, et une liste `wiskundige_vrouwen` avec des mathématiciennes.<br> Le script ajoutera automatiquement toutes les mathématiciennes à la liste wiskundigen. <br>La boucle est limitée par le nombre d'éléments dans la liste `wiskundige_vrouwen`.<br>Exécute le script Python. Comprends-tu ?

```python
# begrensde herhaling

wiskundigen = ["Fermat", "Gauss", "Euler", "Fibonacci", "Galois"]
wiskundige_vrouwen = ["Noether", "Nightingale", "Lovelace"]

for vrouw in wiskundige_vrouwen:
    wiskundigen.append(vrouw)

print(wiskundigen)
```

Vous avez probablement déjà remarqué que Python n'utilise pas de `;` ni de `{}`.Python utilise bien `:`. <br>L'utilisation de l'*indentation* est très particulière. À certains endroits, on effectue une indentation de quatre espaces. Ainsi, le code dans la boucle for est 'groupé'. <br>La variable `vrouw` fait référence à un élément de la liste `wiskundigen` ; cette variable n'a pas été déclarée à l'avance.

<div class="alert alert-block alert-info"> 
Les instructions d'une boucle à exécuter suivent après <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">:</span>  et sont regroupées par indentation. On appelle cela <b> indentation</b>. <br>Lorsque l'indentation s'arrête, la boucle a été exécutée une première fois.</div>

### Exemple 1.2

Vous pouvez aussi utiliser une *répétition bornée* où vous indiquez explicitement le nombre de fois que l'instruction est exécutée. Dans la cellule de code suivante, on utilise une boucle for qui s'exécute 20 fois.<br> Exécutez le code suivant.

```python
# begrensde herhaling

getal = 10
print("getal bij de start:", getal)
for i in range(20):                  # lus wordt 20 keer doorlopen; i neemt waarden van 0 t.e.m. 19 aan
    print("i:", i)
    getal = getal + i                # verkorte instructie: getal += i 
    print("\tgetal:", getal)        # \t doet inspringen
```

`range(20)` génère les nombres naturels consécutifs de 0 à 19 inclus.

### Exemple 1.3

Exécutez également le code suivant.

```python
# begrensde herhaling

even = [0, 2, 4, 6, 8] 

# deze lijst aanvullen met volgende zes even getallen
for i in range(6):                       # lus wordt 6 keer doorlopen; i neemt waarden 0, 1, 2, 3, 4 en 5 aan  
    print("i:", i)
    getal = 8 + 2 * (i+1)
    print("\tgetal:", getal)
    even.append(getal)
    print("\tde lijst even is nu:", even)
```

<h2>2. Boucle while</h2>    </font>
</div>

Outre la boucle for, vous pouvez aussi utiliser une boucle while comme structure de répétition. Avec la boucle while, le nombre de fois que la boucle est exécutée est déterminé par une condition. Dans ce cas, vous travaillez avec une *répétition conditionnelle*.

```python
# voorwaardelijke herhaling

lijst = ["A", "B", "C", "D", "E", "F", "G", "H", "I"]
getal = 0

# de eerste 6 letters van de lijst afdrukken
while getal < 6:
    print("getal =", getal, "\nletter =" , lijst[getal])      # \n doet overgaan naar nieuwe lijn
    getal +=1
```

Remarquez que la boucle while est répétée tant que la condition est satisfaite. Cependant, pour pouvoir exécuter la boucle while une première fois, il est nécessaire que la variable `getal` ait déjà été initialisée avant la boucle while : la variable doit être définie et une valeur doit lui être attribuée, ici 0.

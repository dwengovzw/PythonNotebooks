<div class="alert alert-box alert-success">
Python est un langage orienté objet. En Python, tout est un objet : chaque nombre, chaque chaîne de caractères... Les objets et méthodes associés (fonctions spécifiques à ces objets) sont regroupés en classes.<br>Dans ce notebook, vous apprendrez à définir vous-même une classe.</div>

<div class="alert alert-box alert-info">
On déclare une classe avec le mot-clé <b>class</b>. <br>Dans une classe, un certain nombre d'<b>attributs</b> sont définis qui <b>caractérisent</b> les objets de la classe.Grâce à des fonctions spécifiques, <b>méthodes</b>, on peut, par exemple, consulter ou modifier des attributs. <br></div>

### Exemple

Les plantes à fleurs (ou *angiospermes*) sont classées en monocotylédones (*monocotylées*) et dicotylédones (*dicotylées*). <br>Chez les monocotylédones, les stomates se trouvent sur des lignes parallèles, chez les dicotylédones, ils sont répartis de manière aléatoire.

Le script ci-dessous définit la *classe* `Bloem`. Dans cette classe de plantes à fleurs, on peut initialiser différents *objets*, tels qu'une tulipe et un pommier. Chaque objet de la classe possède *trois attributs* : nom de la plante, espèce de la plante et groupe de plantes.

```python
class Bloem:
    """Klasse Bloemplanten.""" 
    
    def __init__(self, plantnaam, plantsoort, plantgroep):
        """Initialisatie, Bloem-object self heeft drie parameters: naam en type van de bloemplant, al dan niet dicotyl (0 of 1)."""
        self.naam = plantnaam
        self.soort = plantsoort
        self.dicot = plantgroep
        
    def kenmerken(self):
        """Geeft attributen weer."""
        print("De bloemplant is: een", self.naam)
        print("Het soort plant is: een", self.soort)
        print("Is het een monocotyl of dicotyl?")
        if self.dicot == 0: 
            print("Monocotyl")
        else:
            print("Dicotyl")
```

Dans cette classe, on peut utiliser une fonction pour initialiser un objet, `__init__()`, et une méthode pour afficher les caractéristiques (*attributs*) d’un objet donné, `kenmerken()`.

On fait référence à un objet particulier de cette classe avec `self`. Cet objet a alors un nom et appartient à une certaine catégorie : fleur, arbre, culture ou arbuste. <br>Certaines plantes à fleurs sont dicotylées et d’autres ne le sont pas (elles sont monocotylées), ce qui est représenté respectivement par 1 ou 0.

        <h2>2. Construction d'objets de la classe</h2>    </font>
</div>

Si vous avez exécuté la cellule de code ci-dessus, alors la classe `Bloem` a été créée. Cependant, la classe ne contient encore aucun objet. Vous devez encore instancier ces objets. <br>Vous le faites par exemple en exécutant l'instruction `Bloem("bruidsbloem", "struik", 1)`. Via `__init__()`, un objet `self` est alors créé avec les caractéristiques données. <br>

### ExemplePour créer un objet de la classe `Bloem`, vous utilisez le *constructeur* `Bloem()`.<br>Dans le script suivant, un objet `self` est créé, la variable `bloem1` fait référence à ce nouvel objet. Exécutez.

```python
bloem1 = Bloem("tulp", "bloem", 1)
```

Récupérez les attributs.

```python
bloem1.kenmerken()
```

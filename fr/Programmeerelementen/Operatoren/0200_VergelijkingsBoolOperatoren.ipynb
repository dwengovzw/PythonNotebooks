<div class="alert alert-box alert-success">
Lors de l’écriture d’un programme, il arrive souvent que vous deviez vérifier si un certain élément appartient à une liste, ou que vous deviez comparer des éléments. Pour cela, vous avez besoin d’<em>opérateurs</em>.<br>Dans ce notebook, tu vas découvrir les opérateurs de comparaison et tu vas également manipuler des opérateurs booléens (ou logiques).</div>

Pour découvrir ces opérateurs, tu travailleras avec des listes.

<div class="alert alert-block alert-warning">
    Tu as déjà appris à travailler avec des listes, par exemple dans les notebooks sur les types de données. Dans ce qui suit, tu reprends brièvement ce dont tu as besoin pour ce notebook.</div>

<div class="alert alert-block alert-info">
Une <b>liste</b> est un ensemble d’éléments, tels que des nombres et des chaînes de caractères, où un même élément peut apparaître plusieurs fois et où l’ordre des éléments est important. Vous pouvez reconnaître une liste à <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">[ ]</span>.</div>

Exemple:
`getallen = [-2, 510, -7, 28.4, -2, 0, 3.14]                         # liste de sept nombres réels `<br>`gemengdelijst = ["koekoek", 2.5, -1, [5, "ren"], "?!@", " "]        # liste de six éléments, dont l'un est lui-même une liste`
- `getallen` et `gemengdelijst` sont des variables;- `[-2, 510, -7, 28.4, -2, 0, 3.14]` est un objet de type <b>list</b>;- `["koekoek", 2.5, -1, [5, "ren"], "?!@", " "]` est un objet de type <b>list</b>.

<div class="alert alert-block alert-info">
La position d'un élément dans une liste est donnée par l'<b>index</b>.</div>

Exemple:
- `getallen[0] = -2`- `gemengdelijst[3] = [5, "ren"]`

Ce qui se trouve dans les cellules de code ci-dessus est appelé une <b>expression logique</b>.<br> Lors de l'exécution des cellules de code, on vérifie à chaque fois si l'expression (logique) donnée est vraie. Si elle est vraie, la sortie est <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">True</span>, sinon, la sortie est <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">False</span>.<br>

<div class="alert alert-block alert-info">
<span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">True</span> et <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">False</span> sont des <b>valeurs booléennes</b>.  <br><br>Une <em>expression logique</em> ne peut prendre que les valeurs <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">False</span> et <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">True</span>.<br>Une telle expression logique est un objet de type <b>bool</b>. On parle aussi d'une <em>expression booléenne</em>.</div>

### Exemple 1.2

```python
type(5>2)
```

```python
type(False)
```

À l'aide des opérateurs **logiques** ou **booléens**, tu peux combiner des expressions (logiques) pour former une nouvelle expression (logique) et vérifier si la combinaison est vraie.

### Exemple 2.1

Utilisez le tableau ci-dessus pour comprendre la sortie des cellules de code suivantes.

```python
ontbijt[0] != "boten" and 5 <= 2
```

```python
ontbijt[0] != "boten" and 5 > 2
```

```python
ontbijt[0] != "boten" or 5 <= 2
```

```python
not(5>2)
```

<div class="alert alert-block alert-info">
Une combinaison d'expressions logiques effectuée avec l'opérateur <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>and</b></span> est vraie si les deux expressions sont vraies. <br>Une combinaison d'expressions logiques faite avec l'opérateur <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>or</b></span> est vraie si l'une des deux expressions est vraie ou si les deux sont vraies. <br>Si une <em>expression logique</em> est vraie, alors l'application de l'opérateur <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>not</b></span> prendra la valeur False; si elle est fausse, alors l'application prendra la valeur True.</div>

        <h2>1. Octet</h2>    </font>
</div>

<div class="alert alert-block alert-info"> 
Un <b>bit</b> est une unité d'information. Le terme provient de <b>binary digit</b>. C'est une unité qui ne peut prendre que les valeurs 0 et 1. <br>Huit bits forment ensemble un <b>octet</b>. Il y a ainsi 256 combinaisons possibles de 0 et 1 qui forment ensemble un octet.</div>

Les nombres naturels sont des entiers qui sont positifs. Il n'est donc pas nécessaire d'indiquer explicitement le signe (il n'est pas nécessaire de noter le +). <br>Les nombres naturels de 0 à 255 inclus peuvent être représentés par un octet. 77, par exemple, correspond à 01001101.

Dans le module NumPy, vous pouvez stocker un nombre naturel de 0 à 255 avec un octet. <br>Dans ce cas, le type `uint8` *(entier non signé sur 8 bits)* est utilisé.<br><br>Dans le point suivant 2. NumPy ndarray: matrix, vous y trouverez des exemples.

<div>
    <font color=#690027 markdown="1">
        <h2>2. NumPy ndarray: matrix</h2>    </font>
</div>

*En mathématiques, un tableau de nombres s'appelle une matrice.*

Exemple: La matrice $\begin{bmatrix} 1 & 2 & 0 \\ 3 & 4 & 5 \end{bmatrix} $ est une matrice $2x3$. C'est une matrice avec $2$ lignes et $3$ colonnes. <br>En mathématiques, on dit que $2x3$ est la *dimension* de la matrice.

Une matrice est un exemple de *tableau 2D*.

Exécutez les cellules de code suivantes.

```python
matrix1 = np.array([[1, 2, 0], [3, 4, 5]])   # matrix met 2 rijen en 3 kolommen, m.a.w. 2X3-matrix
print(matrix1)
```

```python
type(matrix1) 
```

```python
matrix1.ndim
```

`matrix1` est un objet de type *ndarray* (ndarray signifie nD-array). Ici, il s'agit d'un tableau 2D. Cette matrice possède des lignes et des colonnes, d'où n = 2. <br>Attention : à ne pas confondre avec la dimension mathématique d'une matrice (voir le notebook 'Tensoren').<br><br>Vous saisissez la matrice ligne par ligne dans la cellule de code.

Outre l’attribut *ndim*, un objet de type *ndarray* possède d’autres attributs, tels que *shape*, *dtype* et *size*.<br> Voici comment obtenir ces attributs :

```python
matrix1.shape                                # geeft aantal rijen en aantal kolommen, nl. wiskundige dimensie
```

```python
matrix1.dtype                                # geeft type van elementen van matrix
```

```python
matrix1.size                                 # geeft aantal elementen van matrix
```

Vous pouvez également passer de tels objets comme arguments aux fonctions NumPy telles que `sum()` et `mean()`. <br>En exécutant par exemple la cellule de code suivante, vous calculez la somme de tous les éléments de `matrix1`.

```python
np.sum(matrix1)
```

Comme mentionné dans 1. Byte, le module NumPy offre la possibilité de stocker un nombre naturel de 0 à 255 avec un seul octet. <br>Pour ce faire, vous devez utiliser le type `uint8` *(entier non signé sur 8 bits)*.

Exécutez la cellule de code suivante pour voir comment cela fonctionne.

```python
matrix2 = np.array([[7, 0, 1], [5, 1, 2]], dtype="uint8")   # type elementen zelf kiezen
print(matrix2)
```

```python
matrix2.dtype
```

```python
matrix3 = matrix1.astype("uint8")             # type elementen wijzigen
print(matrix3)
```

```python
matrix3.dtype
```

<div class="alert alert-box alert-info">
Une matrice est un exemple de <b>ndarray</b>.<br> Par défaut, NumPy utilise le type <b>int64</b> pour les nombres entiers. L'intervalle de <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">int64</span> va de -9223372036854775808 à 9223372036854775807.  <br>Vous pouvez aussi définir vous-même le type des éléments avec <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">dtype</span> ou modifier le type avec <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">astype()</span>. <br>Pour un ndarray contenant des nombres naturels de 0 à 255, vous pouvez choisir le type <b>uint8</b>.<br>Pour obtenir les attributs, procédez comme suit : le nombre de lignes et de colonnes d'une matrice avec <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">shape</span>, le type des éléments de la matrice avec <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">dtype</span> et le nombre d'éléments de la matrice avec <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">size</span>.Pour un <em>ndarray<em>, on peut aussi obtenir la valeur de $n$. On le fait avec <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">ndim</span>.</div>

<div class="alert alert-block alert-warning"> 
Outre les matrices, avec un <b>ndarray</b> vous pouvez également représenter des <em>listes NumPy</em> et des <em>tenseurs</em>. <br>Vous trouverez davantage d'exemples de ndarrays dans les notebooks sur les listes NumPy et sur les tenseurs dans le <em>parcours Types de données</em> sur https://dwengo.org/.</div>

<h3>Exemple 2.1 : valeur minimale et maximale dans une liste NumPy</h3>    </font>

L'**étendue** d'une liste donnée de nombres est la différence entre la plus grande et la plus petite valeur de la liste. <br>Pour obtenir la valeur la plus grande ou la plus petite dans une liste NumPy, vous pouvez utiliser les méthodes Python intégrées `min()` et `max()`.<br>Mais vous pouvez également utiliser les fonctions `min()` et `max()` de NumPy, comme dans les cellules de code suivantes.

```python
x.min()
```

```python
np.min(x)
```

```python
np.max(y)
```

```python
variatiebreedte = np.max(y) - np.min(x)
print(variatiebreedte)
```

        <h3> Exemple 2.2 : moyenne et somme des éléments d'une liste NumPy</h3>    </font>
</div>

Avec la fonction `sum()` du module NumPy, vous pouvez calculer la somme des éléments d’un tableau NumPy, avec la fonction `mean()` la moyenne et avec `median()` la médiane.

```python
# som berekenen
x_som = np.sum(x)
print("de som van de elementen van x is:", x_som)
```

```python
# gemiddelde berekenen
x_gemiddelde = np.mean(x)                         # gemiddelde van alle elementen van lijst x
y_gemiddelde_middelstevijf = np.mean(y[5:10])     # gemiddelde van y[5] t.e.m. y[9]
print(x_gemiddelde, y_gemiddelde_middelstevijf)   # laat beide resultaten zien, gescheiden door komma 
```

<h3>Exemple 2.3 : variance et écart-type</h3>    </font>
</div>

Avec les fonctions NumPy `var()` et `std()`, vous obtenez la variance et l'écart-type de tous les éléments d'une liste NumPy.

```python
x_variantie = np.var(x)
x_standaardafwijking = np.std(x)
print(x_variantie, x_standaardafwijking)
```

<h3>Exemple 2.6 : histogramme</h3>    </font>
</div>

La distribution des éléments d'une liste NumPy peut être visualisée sous forme d'histogramme. Pour cela, vous utilisez la fonction `hist()` du sous-module pyplot de Matplotlib.

```python
print(y)
print(np.min(y), np.max(y))
```

```python
import matplotlib.pyplot as plt
plt.hist(y,  bins=10)
```

On considère dix intervalles allant de la valeur minimale 33 à la valeur maximale 100. Le point de départ de chaque intervalle est indiqué sur l’axe des abscisses dans l’histogramme. Le premier intervalle est [33; 39,7[, le deuxième intervalle est [39,7; 46,4[, etc. <br>Le premier intervalle contient donc 4 éléments et le deuxième intervalle n’en contient aucun, etc.

<h3>Exemple 3.2 : arange() et linspace()</h3>    </font>
</div>

Avec NumPy, vous pouvez également générer automatiquement des listes en utilisant l’une des fonctions NumPy `linspace()` ou `arange()`.

```python
lijst1 = np.linspace(0, 10, 11)
print(lijst1)
```

```python
lijst2 = np.arange(10)
print(lijst2)
```

```python
lijst3 = np.arange(-10, 15, step=5)
print(lijst3)
```

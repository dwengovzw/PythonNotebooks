Chaque élément d'une liste NumPy possède un **index**. L'index fait référence à la position d'un élément dans la liste NumPy.À l'aide de l'indice, vous pouvez récupérer un élément donné d'une liste NumPy. Ainsi, le premier élément de la liste NumPy `x` est l'élément d'indice $0$, le deuxième élément a l'indice $1$, etc. La manière de récupérer un élément à l'aide de son indice est montrée dans les cellules de code suivantes:

```python
# element opvragen
x[0]                        # element uit lijst x met index 0
```

```python
# element opvragen
x[6]                        # zevende element uit lijst x
```

```python
# element opvragen          # laatste element uit lijst x
x[-1]
```

<h3>Exemple 2.2 : rechercher un élément</h3>    </font>

Avec la fonction `where()` du module NumPy, vous pouvez déterminer l’indice d’un élément donné et vérifier si un élément apparaît dans une liste NumPy donnée.

```python
# index van bepaald element
np.where(x == 60)
```

`where()` renvoie pour l'élément `60` l'indice $9$. C.-à-d. que `60` est le dixième élément de la liste NumPy.

```python
# index van bepaald element
np.where(x == 57)
```

```python
# index van bepaald element
np.where(x == 3)
```

`where()` renvoie deux indices pour l'élément `57`, car `57` apparaît deux fois dans la liste NumPy `x`.<br> `where()` renvoie pour l’élément `3` deux indices, parce que `3` n’apparaît pas dans la liste NumPy `x`.

Vous pouvez, via l’*index*, supprimer des éléments de la liste NumPy `rij2` à l’aide de la fonction `delete()` du module NumPy.

        <h3>Exemple 2.5 : sélectionner une partie d’une liste NumPy</h3>    </font>
</div>

À l’aide de l’*index*, on peut sélectionner une partie d’une liste NumPy. Python utilise pour cela un *intervalle semi-ouvert*.

```python
stukvanrij2 = rij2[3: 7]      # halfopen interval [3, 7[, m.a.w. indices 3, 4, 5 en 6
print(stukvanrij2)
```

```python
eerstehelftrij2 = rij2[:5]    # indices van begin lijst t.e.m. 4
tweedehelftrij2 = rij2[5:]    # indices van index 5 tot einde lijst
print(eerstehelftrij2)
print(tweedehelftrij2)
```

        <h3>Exemple 3.1 : bit et octet</h3>    </font>
</div>

<div class="alert alert-block alert-info"> 
Un <b>bit</b> est une unité d'information. Le terme vient de <b>binary digit</b>. C'est une unité qui ne peut prendre que les valeurs 0 et 1. <br>Huit bits forment un <b>octet</b>. Il existe 256 octets possibles.</div>

Les nombres naturels sont des nombres entiers qui sont positifs. Il n'est donc pas nécessaire d'indiquer explicitement le signe (il n'est pas nécessaire de noter le +). <br>Les nombres naturels de 0 à 255 inclus peuvent être représentés par un octet. 77, par exemple, correspond à 01001101, c’est une représentation dans le système binaire.

### Exercice 3.1.1Explique pourquoi il y a 256 combinaisons possibles pour former un octet.

Réponse:

Dans le *système de numération binaire*, on ne dispose que de deux chiffres, à savoir $0$ et $1$. Cela signifie que les nombres binaires sont formés par une succession de zéros et de uns, comme $1001001$ et $11111$. Ils sont construits à l'aide des puissances de $2$, puisque $2$ est la base du système de numération binaire.<img src="images/binair.jpg" alt="Banner" width="200"/><br>
Concrètement, le nombre $101 = 1.2^2 + 0.2^1 + 1.2^0$.<br>C’est-à-dire que le nombre binaire $101$ correspond au nombre $5$ dans le système décimal.<br>Ce nombre binaire se lit « un zéro un ».

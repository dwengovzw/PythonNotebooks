        <h2>2. Index</h2>    </font>
</div>

Prenons maintenant une autre chaîne : "Hello, world!". Chaque caractère d'une chaîne a une certaine position. En Python, la position d'un caractère dans une chaîne correspond à un **indice**. On commence à compter à partir de 0.<br>
Par exemple:-  La lettre "H" est le premier caractère. L'indice de "H" est 0.-  La lettre "e" dans la chaîne est à la deuxième position. L'index de "e" est 1.-  Le premier "o" se trouve à la cinquième position. L'index du premier "o" est 4.-  L'espace est le septième caractère de cette chaîne. L'index de l'espace est 6.

#### Exemple 3.1

En utilisant l'opérateur +, vous pouvez concaténer des chaînes. Testez ceci en exécutant les cellules de code suivantes.

```python
# strings concateneren
woord1 = "chat"
woord2 = "bot"
woord = woord1 + woord2
print(woord)
```

```python
# strings concateneren
deel1 = "Wil je aan onderz"
deel2 = "oek"
deel3 = " "
deel4 = "d"
deel5 = "oen"
deel6 = "?"
geheel = deel1 + deel2 + deel3 + deel4 + deel5 + deel6
print(geheel)
```

#### Exemple 4.1

Teste le code dans la cellule de code suivante. Comprends-tu ce que fait ce code ?

```python
zin.split()
```

<div class="alert alert-block alert-info">
La méthode <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">split()</span> convertit une chaîne en une list (une liste). La méthode se base sur les espaces pour effectuer le découpage.<br>Dans l'exemple, la chaîne <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">"Hello, world"</span> a été convertie en une liste de deux chaînes, dans le même ordre que celui dans lequel elles apparaissent dans la chaîne.</div>

Notez que la virgule et le point d'exclamation ne sont pas séparés du mot 'Hello' ici, car il n'y a pas d'espace avant ces signes de ponctuation.C'est quelque chose à garder à l'esprit si, par exemple, vous souhaitiez énumérer tous les mots d'un texte.

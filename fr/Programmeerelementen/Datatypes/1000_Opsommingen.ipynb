Les plantes à fleurs sont divisées en monocotylédones et dicotylédones. <br>Chez les monocotylédones, les nervures sont disposées en lignes parallèles, chez les dicotylédones, elles sont disposées de manière aléatoire.

Quelques exemples de monocotylédones et de dicotylédones sont énumérés ci-dessous.

Les objets `monocotyl` et `dicotyl` sont de type *list*. Ce sont des *listes*. <br>Les deux listes `monocotyl` et `dicotyl` contiennent un certain nombre d’*éléments*. Les éléments des deux listes sont de type `string`.

On peut comparer une telle liste à une suite en mathématiques, à ceci près qu'une suite ne contient que des nombres. <br>L'*ordre* des éléments d'une liste est important. Chaque élément a une place déterminée, représentée par son *indice*. <br>Un élément peut apparaître plusieurs fois.

Le premier élément d'une liste a pour indice 0. On commence donc à compter à partir de 0.<br>Les valeurs négatives ont néanmoins aussi leur signification. Découvrez par vous-même lesquelles.<br>Vérifiez quel élément correspond à chaque fois aux instructions suivantes. Le nombre entre `[ ]` est l’indice.

```python
dicotyl[1]
```

```python
dicotyl[-1]
```

Le nombre d'éléments d'une liste peut être obtenu avec la *fonction* `len()` :

```python
len(dicotyl)
```

Vous pouvez aussi demander une partie d'une liste :

```python
dicotyl[2:4]        # Python werkt hier met het halfopen interval [2,4[
```

```python
dicotyl[:2]         # indices van begin lijst t.e.m. 1
```

```python
dicotyl[2:]         # indices van 2 tot einde lijst
```

<h3>1.4 Méthodes des listes</h3>    </font>
</div>

Exécutez le script suivant. <br>Regardez bien ce que font les méthodes `sort()`,  `remove()`, `append()` et `pop()` et la fonction `len()`.

```python
getallen = [3, 7, 1, 13, 2, 4, 5, 11, 15]       # lijst van getallen
aantalgetallen = len(getallen)
print("aantalgetallen:", aantalgetallen)

# deze lijst sorteren van klein naar groot
getallen.sort()
print("getallen:", getallen)

# deze lijst omvormen tot een lijst van 7 priemgetallen
print("Er worden vier instructies uitgevoerd om de lijst om te vormen tot een lijst van 7 priemgetallen.")
# 1, 4 en 15 zijn geen priemgetallen
# er zijn maar 6 priemgetallen over, voeg er nog een toe
getallen.remove(1)
getallen.remove(4)
getallen.pop() 
getallen.append(17)
print("omgevormde lijst van getallen:", getallen)

# tellen hoeveel elementen in lijst getallen
print("aantal nu:", len(getallen))
```

`sort()` est en effet une méthode qui peut être appelée sur tous les objets de type list.Pour une liste qui, p. ex., ne contient que des mots, vous pouvez utiliser `sort()` pour trier la liste par ordre *alphabétique*. <br>L'instruction `getallen.sort()` montre comment une méthode est utilisée dans une instruction : d'abord l'objet sur lequel la méthode doit être appliquée, suivi d'un point et enfin la méthode.

        <h2>3. Range</h2>    </font>
</div>

`range(20)` génère les nombres naturels consécutifs de 0 à 19 inclus. Si vous affichez l’objet `range(20)`, cela ne vous dira pas grand-chose:

```python
x = range(20)     # genereert opeenvolgende natuurlijke getallen van 0 t.e.m. 19
print(x)
```

```python
# type 
type(x)
```

Vous pouvez également utiliser `range()` pour remplir une liste, à condition d'utiliser le *typecasting* pour convertir un objet de type *range* en type *list*. La signification de l'objet devient alors beaucoup plus concrète. <br>Exécutez le code suivant.

```python
lijst = list(range(20))    
print(lijst)
```

```python
y = range(5, 30)   # genereert opeenvolgende natuurlijke getallen van 5 t.e.m. 29
print(list(y))
```

```python
y = range(5, 30, 3)   # genereert natuurlijke getallen van 5 t.e.m. 29 in stappen van 3
print(list(y))
```

        <h2>4. Dictionary</h2>    </font>
</div>

Dans l'exemple ci-dessous, vous trouverez un objet `planten` de type *dictionnaire (dict)*.<br>On pourrait comparer un dictionnaire à un dictionnaire : on cherche un mot-clé (la *clé*) et, en retour, on trouve une ou plusieurs définitions (la *valeur*).<br><br>Un dictionnaire est un ensemble d’éléments; chaque élément est constitué d’une *clé* et d’une *valeur*. <br>L’ordre des éléments n’a pas d’importance. Une clé ne peut apparaître qu’une seule fois. Un dictionnaire se note entre accolades.<br> 
Un ensemble en mathématiques est également noté avec des accolades. Et, dans un ensemble en mathématiques, l'ordre des éléments n'a pas d'importance.

### Exemple

Les plantes produisent elles-mêmes leur nourriture par la photosynthèse. Cependant, la photosynthèse ne se déroule pas de la même manière chez toutes les plantes. Selon le processus, on distingue les plantes C3, C4 et CAM. <br>La plupart des plantes sont des plantes C3.<br>Les plantes en C4 sont des plantes des régions tropicales. Chez ces plantes, le processus de photosynthèse a été adapté afin de lutter contre la déshydratation. Lors du processus, le moins de CO<sub>2</sub> possible est perdu. <br>Les plantes CAM sont des plantes désertiques, chez elles, l'absorption de CO<sub>2</sub> a lieu la nuit lorsqu'il fait moins chaud.

Exécutez les cellules de code suivantes.

```python
planten = {
    "appelboom": "C3", 
    "cactus": "CAM",  
    "maïs": "C4"
    }
print(planten)
print(len(planten))         # len() geeft aantal elementen terug
```

```python
type(planten)
```

La variable `planten` fait donc référence à un objet de type *dict*.<br>`"cactus": "CAM"` est l'élément avec la *clé (index)* `"cactus"` et la *valeur* `"CAM"`.

```python
print(planten["cactus"])
```

Remarque que pour la clé, on utilise aussi `[ ]`, comme pour l'index des listes.

Vous pouvez ajouter des éléments au dictionnaire :

```python
planten["suikerriet"] = "C4"
planten["sedum"] = "CAM"
planten["graan"] = "C3"
```

```python
print(planten)
```

```python
planten["roos"] = "C2"
print(planten)
```

Oups, faute de frappe, une plante C2 n'existe pas du tout. Heureusement, tu peux corriger cette erreur.<br> De plus, tu peux aussi supprimer des éléments d'un dictionnaire. Pour cela, utilise le mot-clé *del*.

```python
planten["roos"] = "C3"
del planten["sedum"]
print(planten)
```

Exécutez les cellules de code suivantes.

```python
"suikerriet" in planten
```

L'expression logique `"suikerriet" in planten` est un objet de type `bool` et de valeur `True`.

```python
"madeliefje" in planten
```

        <h2>1. Dictionaries</h2>    </font>
</div>

On peut comparer un **dictionnaire** à un dictionnaire.<br>Dans un dictionnaire, on cherche un mot et ce qui suit est la définition. Un dictionnaire contient donc des paires mot, définition.<br>Un dictionnaire est également composé de paires : une **clé** avec une **valeur** correspondante. L'ordre des paires n'a pas d'importance. Les clés sont toutes distinctes.

<div class="alert alert-block alert-info">
Un <b>dictionnaire</b> est composé de paires d'une <b>clé</b> et d'une <b>valeur</b> correspondante. <br> L'ordre des paires n'a pas d'importance. Vous pouvez reconnaître un dictionnaire à <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">{ } et :</span>.<br></div>

#### Exemple 1.1
Considérez un dictionnaire composé de paires (mot, genre). Exécutez les cellules de code suivantes.

```python
geslacht_dict = {"meisje": "onzijdig", 
                 "zin": "mannelijk", 
                 "parasol": "mannelijk", 
                 "auto": "mannelijk", 
                 "functie": "vrouwelijk", 
                 "schip": "onzijdig", 
                 "eiland": "onzijdig", 
                 "dienst": "mannelijk", 
                 "waarheid": "vrouwelijk", 
                 "bakkerij": "vrouwelijk"}  
print(geslacht_dict)
```

#### Exemple 2.1À l'aide du code dans la cellule de code suivante, chargez le début du livre 'Pride and Prejudice' de Jane Austen. Le fragment est affiché à l'écran.

```python
# fragment inlezen en tonen op scherm
with open("data/prideandprejudice_fragment.txt", "r", encoding="utf8") as fragment:
    austen_fragment = fragment.read()      # txt-bestand wordt geopend, fragment gelezen en bestand gesloten
print("Austen:\n", austen_fragment)        # met /n betekent nieuwe lijn
```

```python
type(austen_fragment)
```

#### Exemple 2.2De la même manière, tu lis la première partie de 'Hamlet' (William Shakespeare), ainsi que les premières phrases de 'The Hound of the Baskervilles' (Sir Arthur Conan Doyle). <br>Ensuite, un dictionnaire est créé avec les trois fragments. La variable `drieboeken_fragment` fait référence à ce dictionnaire.

```python
with open("data/hamlet_fragment.txt", "r", encoding="utf8") as fragment:
    shakespeare_fragment = fragment.read()
with open("data/thehoundofthebaskervilles_fragment.txt", "r", encoding="utf8") as fragment:
    conandoyle_fragment = fragment.read()
```

```python
print("Shakespeare:\n", shakespeare_fragment)
```

```python
print("Conan Doyle:\n", conandoyle_fragment)
```

```python
# dictionary van de drie fragmenten
drieboeken_fragment = {"Pride and Prejudice": austen_fragment, "Hamlet": shakespeare_fragment, "The Hound of the Baskervilles": conandoyle_fragment}
```

```python
drieboeken_fragment
```

Supprimez les signes de ponctuation tels que '“', '”' et '.', ainsi que les caractères tels que '_'.Graag de Nederlandse tekst die u wilt laten vertalen. Stuur de inhoud (met eventuele HTML/Markdown of Python-code) en ik vertaal die naar het Frans volgens uw richtlijnen.- Regardez à nouveau l'extrait.

```python
austen_fragment
```

```python
austen_fragment_opgekuist1 = austen_fragment.replace("\ufeff", "")
austen_fragment_opgekuist2 = austen_fragment_opgekuist1.replace("\n", "")
austen_fragment_opgekuist3 = austen_fragment_opgekuist2.replace(",", "")
austen_fragment_opgekuist4 = austen_fragment_opgekuist3.replace("”","")
austen_fragment_opgekuist5 = austen_fragment_opgekuist4.replace("”","")
austen_fragment_opgekuist6 = austen_fragment_opgekuist5.replace("“","")
austen_fragment_opgekuist7 = austen_fragment_opgekuist6.replace("_","")
austen_fragment_opgekuist8 = austen_fragment_opgekuist7.replace("!","")
austen_fragment_opgekuist9 = austen_fragment_opgekuist8.replace("?","")
austen_fragment_opgekuist10 = austen_fragment_opgekuist9.replace("&","")
austen_fragment_opgekuist11 = austen_fragment_opgekuist10.replace(":","")
austen_fragment_opgekuist12 = austen_fragment_opgekuist11.replace(";","")
austen_fragment_opgekuist13 = austen_fragment_opgekuist12.replace(".","")
austen_fragment_opgekuist13
```

```python
austen_fragment_opgekuist = austen_fragment_opgekuist13.replace("      ", " ")
austen_fragment_opgekuist
```

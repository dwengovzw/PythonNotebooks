<div class="alert alert-box alert-success">
Python est un langage orienté objet. Tout y est un <b>objet</b> : chaque nombre, chaque texte... Un objet occupe une certaine place en mémoire et possède une certaine valeur. Chaque objet a aussi un <b>type</b>, par exemple <em>entier</em>, <em>chaîne</em> ou <em>liste</em>. <br><br>Dans ce notebook, certains types de données comme string et integer sont révisés, et vous faites connaissance avec de nouveaux types, tels que NoneType.</div>

<div>
    <font color=#690027 markdown="1">
        <h2>1. Chaîne de caractères et entier</h2>    </font>
</div>

Les types *string (str)* et *integer (int)*, tu les as sûrement déjà rencontrés dans d'autres notebooks.

<div class="alert alert-box alert-info">
Le type <b>str</b> est le type de texte. C'est une séquence de caractères, tels que des lettres, des signes de ponctuation et des espaces. On reconnaît une chaîne aux <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">" "</span>.<br>Le type <b>int</b> est le type d'un entier.</div>

```python
bloem = "iris"
```

`bloem` est une variable qui fait référence à l'objet "iris". On dit que "iris" ainsi que la variable `bloem` sont de type *string*. <br>

```python
getal = -3
```

`getal` est une variable qui fait référence à l'objet `-3`. On dit que `-3` et la variable `getal` sont tous deux de type *int*. <br>

<div class="alert alert-box alert-info">
    Une <b>variable</b> a le même type que l'objet auquel elle fait référence. Une variable peut être réutilisée par la suite dans le programme.<br> Choisissez toujours un nom significatif pour une variable.<br><br>    <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>=</b></span> est un opérateur d'affectation.</div>

Vous savez probablement déjà que vous pouvez concaténer des chaînes (<em>concaténer</em>) à l'aide de l'opérateur `+`, et que vous pouvez convertir un certain type en un autre par *conversion de type*. <br>Exécutez le code suivant. Dans ce code, la fonction `print()` n'a qu'un seul argument.

```python
aantal = 25
print("Er zitten " + str(aantal) + " leerlingen in de klas.")
```

<div class="alert alert-box alert-info">
Assembler des chaînes de caractères pour former une nouvelle chaîne s'appelle la <b>concaténation</b>. Pour cela, on utilise l'opérateur <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>+</b></span>.<br><br>Vous pouvez convertir le type d’un objet en un autre type. Par exemple, vous pouvez transformer un nombre en chaîne de caractères. Ce concept s’appelle la <b>conversion de type</b>. Avec la <b>fonction</b> <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>int()</b></span>, on convertit vers le type entier, avec <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>float()</b></span> vers le type flottant et avec la fonction <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;"><b>str()</b></span> vers le type chaîne de caractères.</div>

<h2>2. Boolean</h2>    </font>
</div>

Commencez par explorer quelques exemples. <br>Exécutez les instructions suivantes. Comprenez-vous la sortie ?

```python
2 < 5
```

```python
- 3 == getal
```

```python
bloem != "iris"
```

Les expressions `2 < 5`, `- 3 == getal` et `bloem != "iris"` sont des objets de type *booléen (bool)*. Elles ne peuvent prendre qu'une seule de ces deux valeurs : `False` ou `True`. <br>Le type booléen est un sous-type du type entier : `False` correspond à 0 et `True` à 1.

<div class="alert alert-box alert-info">
Un objet de type <b>bool</b> ne peut prendre que les valeurs <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">False</span> et <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">True</span>.</div>

        <h2>3. Floating-point number</h2>    </font>
</div>

On représente les nombres non entiers par un nombre décimal (un nombre à virgule), p. ex. $4,35$. <br>Mais on peut aussi utiliser un nombre à virgule pour les entiers, p. ex. $3 = 3,0$. C.-à-d. que l’entier -3 n’est pas toujours stocké de la même manière dans l’ordinateur : -3 et -3.0 sont stockés différemment. <br>Les nombres décimaux sont de type *nombre en virgule flottante (float)*.<br>

Exécutez la cellule de code suivante.

```python
getal = 8.5123
type(getal)
```

```python
import math
pi = math.pi
type(pi)
```

<div class="alert alert-box alert-info">
Un nombre décimal est de type <b>float</b>.</div>

<div class="alert alert-box alert-info">
Vous pouvez connaître le type d'un objet à l'aide de la fonction <b>type()</b>.</div>

        <h2>4. NoneType</h2>    </font>
</div>

Un type très particulier est `NoneType`. La seule valeur qu'un objet de ce type peut prendre est `None`. None est utilisé pour représenter l'absence d'une valeur.<br>

<div class="alert alert-box alert-info">
La seule valeur qu'un objet de type <b>NoneType</b> peut prendre est <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">None</span>. Une valeur est absente.</div>

Exécutez la cellule de code suivante.

```python
type(None)
```

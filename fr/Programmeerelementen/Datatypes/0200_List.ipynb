        <h2>1. Lists</h2>    </font>
</div>

Reprenez quelques éléments du notebook 'Strings':

```python
zin = "Hello, world!"
print(zin[7])
print(zin[7:12])         
```

```python
zin.split()
```

<div class="alert alert-block alert-info"> 
<span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">zin</span> est une <b>variable</b> qui fait référence à la chaîne <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">"Hello, world!"</span>. <br>Le nom que vous donnez à la variable, vous le choisissez vous-même, mais il est préférable de choisir un nom significatif.<br>Chaque caractère d'une chaîne possède un <b>index</b> qui indique la position de ce caractère dans cette chaîne. <br><br>    La méthode <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">split()</span> convertit une chaîne en une list (une liste). La méthode s'appuie sur les espaces pour effectuer la séparation.<br>Dans l'exemple, la chaîne <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">"Hello, world"</span> a été convertie en une liste contenant deux chaînes, dans le même ordre que celui dans lequel elles apparaissent dans la chaîne.</div>

#### Exemple 1.1

Exécutez les cellules de code suivantes.

```python
getallen = [-2, 510, -7, 28.4, -2, 0, 3.14]                         # list met zeven reële getallen 
gemengdelijst = ["koekoek", 2.5, -1, [5, "ren"], "?!@", " "]        # list met zes elementen, waarvan een zelf een list
```

```python
type(getallen), type(gemengdelijst)
```

<div class="alert alert-block alert-info">
Une <b>list</b> est une liste d’éléments, tels que des nombres et des chaînes de caractères, où un même élément peut apparaître plusieurs fois et où l’ordre des éléments est important. Vous pouvez reconnaître une list grâce à <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">[ ]</span>.</div>

<div class="alert alert-block alert-info"> 
    <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">[-2, 510, -7, 28.4, -2, 0, 3.14]</span> est un objet de type <b>liste</b>.<br><span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">["koekoek", 2.5, -1, [5, "ren"], "?!@", " "]</span> est un objet de type <b>list</b>.   </div>

<div class="alert alert-block alert-info">
Comme pour les chaînes de caractères, vous pouvez concaténer des listes à l'aide de l'<span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">opérateur +</span>.<br>Comme pour une chaîne, la position d’un élément dans une liste est donnée par l’<b>indice</b>.</div>

Lors de l'exécution des cellules de code, on vérifie à chaque fois si l'expression (logique) donnée est vraie. <div class="alert alert-box alert-info">

On parle également d'une expression booléenne.</div>

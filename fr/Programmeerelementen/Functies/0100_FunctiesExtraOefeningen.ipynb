<h2>1. Fonction avec structure conditionnelle, entrée, sortie, docstring</h2></div>

### Exemple

À l'exécution du script suivant, l'utilisateur est invité à saisir un nombre entier aléatoire. <br>Ensuite, 7 est divisé par ce nombre via la fonction `zevendelen()`. Le quotient est affiché. Cependant, si l'utilisateur saisit le nombre 0, il n'y a pas de quotient.<br>Exécutez le script quelques fois. Essayez différents nombres, n'oubliez pas 0.

```python
def zevendelen(getal):
    """Quotiënt van 7 met een getal."""
    if getal != 0:
        resultaat =  7 / getal             # als noemer niet 0 is, bereken quotiënt
    else:
        resultaat = "Er is geen quotiënt want je kan niet delen door nul."  # als noemer 0 is, dan is er geen quotiënt
    return resultaat

# invoer
# 7 is teller, in te voeren getal is noemer
print("Deel 7 door een geheel getal naar keuze.") 
noemer = int(input("Geef een geheel getal naar keuze waardoor 7 moet gedeeld worden: "))  # typecasting: string omzetten naar int

# invoer verwerken
quot = zevendelen(noemer)    # quotiënt

# uitvoer
print(quot)
```

<div class="alert alert-block alert-info">
Comprends-tu ce qui se passe ?<br>On te demande d’entrer un nombre. Tu saisis ce nombre. Cette saisie est interprétée comme une chaîne de caractères, mais, par transtypage, elle est convertie en un objet de type int. La variable <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">noemer</span> fait référence à cet objet de type int. <br>L'entrée est ensuite traitée. La fonction est appelée : 7 est divisé par la valeur de <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">noemer</span>, du moins si cette valeur n'est pas nulle. Le quotient est un objet de type float. La variable <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">resultaat</span> fait référence à cet objet float. Dans le cas où l'entrée vaut 0, la variable <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">resultaat</span> fait référence à un objet de type <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">string</span>. <br>La variable <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">quot</span> fait référence au même objet que la variable <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">resultaat</span>, que la fonction renvoie. <br>La valeur de cet objet <span style="background-color:whitesmoke; font-family:consolas; font-size:1em;">quot</span> est affichée. <br> <br>    
La phrase suivante explique l’objectif de la fonction : Quotient de 7 par un nombre. C’est une <b>docstring</b>.

    <h2>2. Module Random - structure de répétition</h2></div>

*Dans le module NumPy* il existe un *module Random* permettant de générer des nombres aléatoires. <br>Dans le script suivant, deux fonctions du module Random sont utilisées : la fonction `random()` pour générer des nombres décimaux et la fonction `randint()`, avec deux paramètres, pour générer des nombres entiers. Il est donc nécessaire d'importer d'abord le module NumPy.<br>Les nombres générés par `random()` sont toujours compris dans [0,1[ et les nombres générés par `randint()` appartiennent à l'intervalle semi-ouvert déterminé par les deux arguments que vous passez à la fonction, la borne supérieure non incluse. <br>Pour appeler ces fonctions, faites-les précéder de `np.random.`.

### Exemple 2.1Testez le script suivant :

```python
import numpy as np

# for-lus wordt 10 keer doorlopen
for teller in range(10):
    print("teller = ", teller)
    print(np.random.random())        # functie random() genereert decimaal getal in [0,1[
    print(np.random.randint(0, 4))   # functie randint(0,4) kiest willekeurig geheel getal uit [0,4[ 
    print()
```

### Exemple: DNA

Le matériel génétique d’un organisme vivant est stocké dans l’ADN. Une molécule d’ADN est constituée de deux longues chaînes de nucléotides, entrelacées sous la forme d’une double hélice. Les nucléotides sont un groupe spécifique de composés organiques. Un brin d’ADN contient quatre types de nucléotides dont la base azotée est l’une des suivantes : adénine, thymine, guanine et cytosine. Ces bases azotées sont abrégées respectivement par les lettres A, T, G et C. La séquence d’ADN est l’ordre de ces nucléotides sur un brin d’ADN. Il existe un très grand nombre de séquences possibles.

Avec les fonctions `choice()` du module Random de NumPy et `join()`, vous pouvez choisir et assembler les lettres.

Avec la fonction suivante, vous pouvez générer une séquence d'ADN aléatoire d'une longueur souhaitée :

```python
def dna_sequentie(n):
    """DNA-sequentie genereren met gewenste lengte."""
    letters = ["A", "C", "G", "T"]
    keuze = [np.random.choice(letters) for i in range(n)]  # kies n letters uit lijst letters en stop die in andere lijst
    # print(keuze)
    string = "".join(keuze)   # elementen van keuze samenbrengen in een (samenhangende) string zonder keuze aan te passen
    return string
```

Générez une séquence d'ADN de longueur 40.

Alternative : vous pouvez utiliser un module Random (un autre, pas celui de NumPy). Vous pouvez alors choisir des lettres dans une chaîne de caractères plutôt que dans une liste.Avec la fonction suivante, vous pouvez générer une séquence d’ADN aléatoire de la longueur souhaitée :

```python
import random

def dna_sequentie_2(n):
    """DNA-sequentie genereren met gewenste lengte."""
    letters = "ACGT"
    keuze = [random.choice(letters) for i in range(n)]  # kies n letters uit string letters en stop die in lijst
    # print(keuze)
    string = "".join(keuze)   # elementen van keuze samenbrengen in een (samenhangende) string zonder keuze aan te passen
    return string
```

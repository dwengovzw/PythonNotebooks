        <h1>ENREGISTRER DES NOTEBOOKS</h1>    </font>
</div>

<div class="alert alert-box alert-success">
Il s’agit du dernier notebook de la série d’introduction. Vous apprendrez comment enregistrer les notebooks.</div>

Une fois que vous êtes connecté au serveur web, vous pouvez utiliser un notebook sur le serveur pendant environ une heure et demie.

<div>
    <font color=#690027 markdown="1">    
<h2>1. Imprimer un notebook</h2>    </font>
</div>

Vous pouvez imprimer le notebook comme vous imprimez p. ex. une page web. Il est recommandé de choisir une échelle de p. ex. 85 %.

<div>
    <font color=#690027 markdown="1">
        <h2>2. Enregistrer un notebook au format pdf</h2>    </font>    
</div>

Enregistrer un notebook en PDF se fait de la même manière que l’impression d’une page web. Il suffit de remplacer l’imprimante sélectionnée par `print to pdf`. Ici aussi, il est préférable de choisir une échelle de, par exemple, 85 %.

<div>
    <font color=#690027 markdown="1">
        <h2>3. Enregistrer un notebook au format notebook</h2>    </font>
</div>

Vous pouvez également télécharger le notebook et l'enregistrer sur votre ordinateur. <br>Voici comment procéder : `File` > `Download`.
Ensuite, vous pouvez retrouver le notebook sous forme de fichier `ipynb` dans le dossier Téléchargements de votre ordinateur. Vous pouvez ensuite placer le notebook dans un dossier où vous pourrez le retrouver facilement.
Si vous souhaitez ouvrir et parcourir le notebook enregistré depuis votre ordinateur, vous pouvez par exemple utiliser l'application Jupyter Notebook dans Anaconda.

<div class="alert alert-block alert-danger"> 
Veuillez noter que le notebook téléchargé n'est pas mis à jour lorsque vous enregistrez le notebook sur le serveur web. Si vous souhaitez une nouvelle version du notebook, vous devez le télécharger à nouveau.</div>

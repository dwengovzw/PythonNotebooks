Dans les notebooks, vous trouverez deux types de cellules : une *cellule Markdown* et une *cellule de code*.-  Une **cellule Markdown** correspond à une zone où vous pouvez écrire du texte, mais aussi insérer des hyperliens et des images.-  Une **cellule de code** sert à saisir du code. Vous reconnaissez une cellule de code parce qu'elle est toujours grise et est précédée de `In [ ]`. Lors de l'exécution du code, un \* apparaît entre les `[ ]`. Une fois le code exécuté, un nombre se trouve entre les `[ ]`.<br><br>
**Tout le texte ci-dessus se trouve dans des cellules Markdown.**

```python
# Deze tekst staat in een code-cel.
```

        <h2>2. Modules</h2>    </font>    
</div>

Les notebooks sont structurés de telle sorte qu'au début d'un notebook, on importe généralement d'abord les *modules* nécessaires. Cela permet de garder les notebooks clairs.

<div class="alert alert-block alert-info">
Les modules sont en quelque sorte des paquets de composants que vous pouvez utiliser pour écrire votre code. Ces composants contiennent du code complexe déjà écrit pour vous, afin que vous n’ayez plus à le faire vous-même.</div>

Lors de l'écriture du code, veille à adopter un **style de programmation lisible** et ajoute les **commentaires** nécessaires pour clarifier tes étapes!

<div class="alert alert-block alert-danger"> 
Il est intéressant de savoir que, dans un notebook, tout le code fait partie d'un même ensemble. Le notebook mémorise en quelque sorte quel code a déjà été exécuté, quel que soit l'ordre dans lequel il a été saisi dans le notebook.<div>

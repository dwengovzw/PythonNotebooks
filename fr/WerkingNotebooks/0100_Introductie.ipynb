<div>
    <font color=#690027 markdown="1">
        <h2>1. Jupyter Notebook</h2>    </font>    
</div>

Jupyter Notebook est une application web open source dans laquelle vous pouvez créer, utiliser et modifier des *notebooks*.<br>Les notebooks sont des **documents numériques** qui contiennent à la fois du code exécutable ainsi que du texte, des images, des vidéos et des hyperliens. Il est donc possible, via les notebooks, d'introduire de nouveaux concepts, de formuler des exercices, d'afficher des photos, d'exécuter le code existant et de rédiger son propre code. <br><br>En bref, les notebooks peuvent être utilisés pour assimiler de nouvelles notions, y compris un environnement de programmation. Le code utilisé dans les notebooks est en *Python 3*.

Les notebooks Jupyter sont proposés via un serveur web, accessible via un **navigateur web**.<br>Vous n’avez donc pas besoin d’installer de logiciel supplémentaire sur votre ordinateur pour commencer à travailler avec les notebooks en ligne. Si votre ordinateur dispose d’un navigateur, vous pouvez parcourir les notebooks et éventuellement exécuter du code.

<div>
    <font color=#690027 markdown="1">
        <h2>2. Les cadres</h2>     </font>    
</div>

Dans les notebooks, **des cadres verts, jaunes, bleus et rouges** sont utilisés.

<div class="alert alert-box alert-success">
Les encadrés verts donnent des explications sur le notebook et des informations contextuelles concernant le contenu du notebook.</div>

<div class="alert alert-block alert-warning"> 
Les cadres jaunes renvoient à d'autres notebooks. Ceux-ci sont utilisés pour renvoyer vers un arrière-plan théorique ou vers un complément aux notebooks sur lesquels vous travaillez.</div>

<div class="alert alert-block alert-info"> 
Les encadrés bleus contiennent de la théorie en informatique, ainsi que d'autres notions théoriques que tu abordes à ce moment-là.</div>

<div class="alert alert-block alert-danger"> 
Les encadrés rouges donnent des conseils et mettent en garde contre des points auxquels vous devez absolument prêter attention.</div>

Félicitations ! Vous venez de parcourir votre premier notebook. Dans les prochains notebooks, vous apprendrez à utiliser les notebooks par vous-même.

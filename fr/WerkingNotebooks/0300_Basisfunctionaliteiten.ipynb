Dans la plupart des cas, une erreur dans votre code ne causera pas de problèmes graves. Le simple fait d’ajuster votre code jusqu’à ce qu’il soit correct est suffisant dans de tels cas.<br>Mais parfois, il peut arriver que quelque chose se passe mal et que l’ordinateur se bloque dans le code ou qu’il y ait des informations que l’ordinateur doit traiter de nouveau. Dans ces cas-là, deux solutions simples s’offrent à vous :-  interrompre l'exécution de la cellule de code;-  recommencer avec le notebook.

<div>
    <font color=#690027 markdown="1">
        <h3>2.1 Interrompre une cellule de code en cours d'exécution</h3>    </font>
</div>

Pour interrompre l'*exécution* d'une cellule de code :- Dans la barre d'outils, choisissez `Kernel` > `Interrupt`.

<div style='color: #690027;' markdown="1">
<h3>2.2 Recommencer</h3></div>

Le titre est un peu trompeur. Cela ne veut pas dire que vous devez réécrire tout le code. C'est surtout votre ordinateur qui, dans ce cas, effectue du travail supplémentaire lorsque vous réexécutez ensuite les cellules de code.  <br>Pour recommencer avec le notebook:- Dans la barre d'outils, choisissez `Kernel` > `Restart & Clear Output`.

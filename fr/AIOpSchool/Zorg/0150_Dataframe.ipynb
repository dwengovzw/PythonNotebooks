### Importer les modules nécessaires

```python
import numpy as np
import pandas as pd
```

### Importer des données avec pandas et les afficher

```python
data = pd.read_csv("data/data.csv") 
```

La fonction `read_csv` de pandas lit le fichier CSV et crée un *data frame* : un tableau qui contient toutes les données du fichier CSV.<br>Les noms des données (le cas échéant) sont clairement affichés comme titres des colonnes et les lignes sont numérotées (en commençant à partir de 0). Un data frame est une **structure de données bidimensionnelle**.

```python
data
```

Ce tableau contient donc les données des patients, une colonne par donnée, et chaque ligne correspond à un patient donné.

Ici et là dans le tableau, on voit `NaN` (*not a number*), ce qui signifie qu’à cet endroit du tableau une valeur est manquante. Cela correspond aux champs vides dans le fichier CSV (le patient de la ligne 10 se trouve à la ligne numéro 8 dans le DataFrame pandas et le patient de la ligne 13 se trouve à la ligne numéro 11 dans le DataFrame pandas).

### Prétraitement des données

- Les ordinateurs calculent. Pour cela, des nombres sont nécessaires. C’est pourquoi vous devez convertir les valeurs des variables catégorielles (non numériques) en valeurs numériques, avant de pouvoir faire traiter ces données par l’ordinateur.- Le lieu de résidence d’un patient n’est pas pertinent pour un traitement, vous supprimerez donc également cette colonne.- Les lignes contenant `NaN`, vous les supprimerez, afin d'éviter que l'ordinateur ne plante à cause de valeurs manquantes.

```python
# categorische variabelen numeriek maken a.d.h.v. methode replace()
# keuze cijfers is vrij, best zinvolle keuze maken
data["Geslacht"] = data["Geslacht"].replace("V", 0)
data["Geslacht"] = data["Geslacht"].replace("M", 1)
data["Bloedgroep"] = data["Bloedgroep"].replace("O", 0)
data["Bloedgroep"] = data["Bloedgroep"].replace("A", 1)
data["Bloedgroep"] = data["Bloedgroep"].replace("B", 2)
data["Bloedgroep"] = data["Bloedgroep"].replace("AB", 3)
data["Rhesus"] = data["Rhesus"].replace("+", 1)
data["Rhesus"] = data["Rhesus"].replace("-", -1)
data["TypeKlacht"] = data["TypeKlacht"].replace("Standaard", 1)
data["TypeKlacht"] = data["TypeKlacht"].replace("NietErnstig", 0)
data["TypeKlacht"] = data["TypeKlacht"].replace("Ernstig", 2)
data["TypeKlacht"] = data["TypeKlacht"].replace("Verontrustend", 3)
data["Pijn"] = data["Pijn"].replace("N", 0)
data["Pijn"] = data["Pijn"].replace("J", 1)
data["AantalSymptomen"] = data["AantalSymptomen"].replace("Constant", 0)
data["AantalSymptomen"] = data["AantalSymptomen"].replace("Dalend", -1)
data["AantalSymptomen"] = data["AantalSymptomen"].replace("Stijgend", 1)
```

```python
data
```

```python
# overbodige kolom verwijderen a.d.h.v. sleutelwoord del
del data["Gemeente"]
```

```python
data
```

```python
# rijen met onvolledige gegevens verwijderen met methode dropna()
data = data.dropna()
```

```python
data
```

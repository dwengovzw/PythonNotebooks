## Het Euleriaans pad

Nu we een voorstelling hebben van onze graaf, kunnen we controleren of er een Euleriaans pad te vinden is in de graaf. Denk daarvoor terug aan de stelling die Euler formuleerde.

**Een graaf bevat een Euleriaans pad als elke knoop in de graaf een even aantal bogen heeft of als er juist twee knopen zijn met een oneven aantal bogen.**

**Opdracht:** Vul onderstaande functie aan. Deze geeft `True` terug als de ingevoerde graaf een Euleriaans pad bevat. Anders geeft de functie `False` terug.

```python
def heeft_graaf_euleriaans_pad(graaf):
    # Een graaf heeft een Euleriaans pad als er 0 of 2 knopen zijn met een oneven graad
    aantal_oneven_graden = 0
    
    # Vul hier de code aan, je telt hier het aantal knopen met een oneven graad...
            
    return aantal_oneven_graden == 0 or aantal_oneven_graden == 2
```

Pas je functie toe op onze voorbeeldgraaf.

```python
print(heeft_graaf_euleriaans_pad(graaf))
```

## Op zoek naar het Euleriaans pad

We hebben al een functie om te weten te komen of er een Euleriaans pad bestaat in een graaf. Nu moeten we dat pad nog kunnen vinden. Daarvoor bestaan verschillende algoritmes. Wij maken gebruik van het algoritme van Fleury.

```python
# We maken een functie die een Euleriaans pad zal zoeken in een graaf aan de hand van de Fleury algoritme
def fleury(graaf):
    # Eerst controleren we of het mogelijk is om een Euleriaans pad te vinden.
    if not heeft_graaf_euleriaans_pad(graaf):
        return None

    # Omdat het algoritme de graaf zal aanpassen, maken we eerst een kopie van de graaf.
    graaf = [lijst[:] for lijst in graaf]
    
    # Als er 2 knopen zijn met een oneven graad, dan starten we in een van deze knopen.
    # We overlopen de knopen en zoeken de eerste knoop met een oneven graad.
    # Als er geen knopen zijn met een oneven graad, dan starten we in de eerste knoop.
    start_knoop = 0
    for i in range(len(graaf)):
        if len(graaf[i]) % 2 != 0:
            start_knoop = i
            break # Stop wanneer oneven knoop gevonden is

    # Voeg onze eerste knoop toe aan het pad.
    pad = [start_knoop]

    # Herhaal tot er geen bogen meer zijn in de graaf.
    while any(graaf):
         # De huidige knoop is de laats toegevoegde knoop aan het pad.
        huidige_knoop = pad[-1]
        
        # Zoek een buur die de graaf niet in twee knipt.
        for buur in graaf[huidige_knoop]:         
            # Als de boog de graaf niet in twee knipt
            if not helpers.boog_knipt_graaf_in_twee(graaf, huidige_knoop, buur):
                # Verwijder de boog uit de graaf.
                helpers.verwijder_boog_uit_graaf(graaf, huidige_knoop, buur)
                # Voeg de buur toe aan het pad
                pad.append(buur)
                break          

    return pad
```

We kunnen het algoritme van Fleury nu toepassen op onze grafen.

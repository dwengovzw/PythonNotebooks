Net als in de vorige notebook gebruiken we foto's van het raster waarop jij afbeeldingen getekend hebt. ## Het oorspronkelijke neurale netwerk

Hieronder zie je het neurale netwerk dat we gebruikt hebben in de vorige notebook. Het netwerk bestaat uit verschillende soorten lagen. In deze notebook leer je stap voor stap waarvoor elk van deze lagen dient. 

![](images/neural_network_visualized.png)

Voer onderstaande code uit om het originele netwerk in te laden.

```python
# Deze functie legt vast wat de structuur is van het AI-model dat we gaan trainen.
def maak_origineel_neuraal_netwerk(hoogte_afbeelding, breedte_afbeelding):
    model = Sequential()
    
    model.add(InputLayer(input_shape=(hoogte_afbeelding, breedte_afbeelding, 1)))
    
    model.add(Conv2D(1, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.1))
    
    model.add(Conv2D(2, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.1))
    
    model.add(Conv2D(4, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.1))
    
    model.add(Conv2D(8, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.1))
    
    model.add(Conv2D(16, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.1))
    
    model.add(Flatten())
    model.add(Dense(16, activation='relu'))   
    model.add(Dropout(0.1)) 
    model.add(Dense(2, activation='softmax'))
    
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    
    return model
```

```python
model = maak_origineel_neuraal_netwerk(train_afbeeldingen.shape[1], train_afbeeldingen.shape[2])
```

Met de volgende cel train je het netwerk op de trainingsverzameling.

```python
model.fit(train_afbeeldingen, train_labels, epochs=5, batch_size=1, validation_data=(val_afbeeldingen, val_labels))
```

Nadat we het model getraind hebben, bekijken we de prestatie op de testverzameling. Zo zien we of het netwerk écht geleerd heeft om de kenmerken van de emoties te herkennen of dat het netwerk enkel de trainingsverzameling heeft gememoriseerd. Hiervoor gebruiken we een helper functie die deze evaluatie voor ons zal doen.

Voer de volgende code uit om de prestatie van het model te bekijken.

```python
helpers.evalueer_model(model, test_afbeeldingen, test_labels)
```

## Breek het model

Zoals je weet uit de vorige notebook kan je het model leren om de emoties van smileys goed te detecteren. Een belangrijke vraag die je moet stellen is: Hoe komt het dat dit model goed werkt? Welke elementen in het model zijn essentieel om het goed te doen leren? Wat gebeurd er als we bepaalde elementen in het model weglaten?

**Opdracht**:

**Tip 1**: Je kan een laag uit het model halen door die in commentaar te zetten. Dat doe je door een hekje (#) voor de lijn te zetten.
**Tip 2**: Wanneer je het netwerk aanpast, begin je best met alle lagen van 1 soort weg te halen. Zo kan je het model op een gestructureerde manier reduceren.
**Tip 3**: Wanneer je een aanpassing doet kan je het model best meerdere keren trainen en de prestatie bekijken. Soms kan het model de ene keer goed werken en de andere keer niet meer goed werken. 

```python
# Deze functie legt vast wat de structuur is van het AI-model dat we gaan trainen.
def maak_gebroken_neuraal_netwerk(hoogte_afbeelding, breedte_afbeelding):
    model = Sequential()
    
    model.add(InputLayer(input_shape=(hoogte_afbeelding, breedte_afbeelding, 1)))
    
    model.add(Conv2D(1, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.1))
    
    model.add(Conv2D(2, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.1))
    
    model.add(Conv2D(4, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.1))
    
    model.add(Conv2D(8, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.1))
    
    model.add(Conv2D(16, (3, 3), activation='relu'))
    model.add(BatchNormalization())
    model.add(MaxPooling2D((2, 2)))
    model.add(Dropout(0.1))
    
    model.add(Flatten())
    model.add(Dense(16, activation='relu'))   
    model.add(Dropout(0.1)) 
    model.add(Dense(2, activation='softmax'))
    
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    
    return model
```

```python
gebroken_model = maak_gebroken_neuraal_netwerk(train_afbeeldingen.shape[1], train_afbeeldingen.shape[2])
```

```python
gebroken_model.fit(train_afbeeldingen, train_labels, epochs=5, batch_size=1, validation_data=(val_afbeeldingen, val_labels))
```

```python
helpers.evalueer_model(gebroken_model, test_afbeeldingen, test_labels)
```

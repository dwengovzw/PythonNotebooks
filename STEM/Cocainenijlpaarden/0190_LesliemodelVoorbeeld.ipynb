# Wiskundig modelleren met een Lesliematrix 

```python
# nodige modules importeren
import numpy as np
import matplotlib.pyplot as plt
```

```python
# nijlpaarden worden maximaal 45 jaar
leeftijden = np.arange(3)
print(leeftijden)
```

```python
# bevolkingsmatrix met aantallen per leeftijd
n0 = np.zeros(3)  # bevolkingsmatrix met drie leeftijdscategorieen
n0[0] = 204        # aantal nuljarige dieren
n0[1] = 60         # aantal eenjarige dieren
n0[2] = 36         # aantal tweejarige dieren
print(n0)
print(n0.shape)
```

```python
# vruchtbaarheid
# worden vruchtbaar vanaf leeftijd van 1 jaar
# wordt eerste rij in vierkante matrix - Lesliematrix 
vruchtbaarheid = np.array([[0.0, 2.5, 1.5]])

# overlevingscijfers
# van de 0-jarigen wordt 60 % 1 jaar, van de 1-jarigen wordt 40 % 2 jaar
overleving = np.array([0.6, 0.4, 0.0])  # -> nog naar diagonaalmatrix

print(vruchtbaarheid)
print(overleving)

print(vruchtbaarheid.shape)
print(overleving.shape)
```

```python
# Lesliematrix L
L_onderstedeel = np.identity(3) * overleving
print(L_onderstedeel)
print(L_onderstedeel.shape)
```

```python
L = np.concatenate((vruchtbaarheid,L_onderstedeel))
print(L)
print(L.shape)
```

```python
# laatste rij wegdoen
L = L[0:3,]
print(L)
print(L.shape)
```

De Lesliematrices hebben als dimensie 3 x 3. Ze kunnen vermenigvuldigd worden met een kolommatrix van dimensie 3 x 1, wat dus de dimensie van de bevolkingsmatrix moet zijn. Dit klopt voor n0.

```python
n0 = np.transpose(n0)
```

```python
# na 1 jaar
# matrixvermenigvuldiging met functie dot() die de betreffende matrices als parameters meekrijgt; dot(A, B) 
n1 = np.dot(L, n0)
print(n1)
```

```python
# na 2 jaar
n2 = np.dot(L, n1)
print(n2)
```

```python
# na 5 jaar
n3 = np.dot(L, n2)
n4 = np.dot(L, n3)
n5 = np.dot(L, n4)
print(n5)
```

```python
# na 5 jaar, kortere manier
A = L
for i in range(4): 
    A = L * A
    print(i+1, A)

#n5 = np.dot(A, n0)
#print(n5)
```

```python
# na 5 jaar
n5 = np.dot(A, n0)
print(n5)
```

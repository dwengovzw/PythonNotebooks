# Wiskundig modelleren met een Lesliematrix 

```python
# nodige modules importeren
import numpy as np
import matplotlib.pyplot as plt
```

### Voorbeeld

Vertrek van het volgende probleem.

Stel dat je een populatie dieren hebt van een soort die maximum twee jaar oud wordt.<br>
60 % van de nuljarigen wordt 1 jaar, en 40 % van de eenjarigen wordt 2 jaar. In het tweede levensjaar krijgen de vrouwtjes gemiddeld 5 nakomelingen en in het derde levensjaar krijgen ze er gemiddeld 3.<br>
Momenteel zijn er 204 dieren in hun eerste levensjaar, 60 dieren zijn een jaar oud en 36 zijn in hun derde levensjaar.

Bekijk deze populatie dieren binnen 5 jaar.

```python
# rijmatrix aanmaken voor leeftijden
leeftijden = np.arange(3)
print(leeftijden)
```

```python
# bevolkingsmatrix met aantallen per leeftijd
n0 = np.zeros(3)  # bevolkingsmatrix met drie leeftijdscategorieën
n0[0] = 204        # aantal nuljarige dieren
n0[1] = ...         # aantal eenjarige dieren
n0[2] = ...         # aantal tweejarige dieren
print(n0)
print(n0.shape)
```

```python
# vruchtbaarheid
# vruchtbaar vanaf leeftijd van 1 jaar
# wordt eerste rij in Lesliematrix
vruchtbaarheid = np.array([[0.0, 2.5, ...]])

# overlevingscijfers
# van de 0-jarigen wordt 60 % 1 jaar, van de 1-jarigen wordt 40 % 2 jaar
overleving = np.array([..., 0.4, 0.0])  # -> nog naar diagonaalmatrix

print(vruchtbaarheid)
print(overleving)

print(vruchtbaarheid.shape)
print(overleving.shape)
```

```python
# Lesliematrix L
L_onderstedeel = np.identity(3) * overleving  # overleving moet op diagonaal komen
print(L_onderstedeel)
print(L_onderstedeel.shape)
```

```python
L = np.concatenate((vruchtbaarheid, L_onderstedeel), axis=0)   # matrices onder elkaar plaatsen om nieuwe matrix te maken
print(L)
print(L.shape)
```

```python
# laatste rij wegdoen
L = L[0:3,]
print(L)
print(L.shape)
```

De Lesliematrices hebben als dimensie 3 x 3. Ze kunnen vermenigvuldigd worden met een kolommatrix van dimensie 3 x 1, wat dus de dimensie van de bevolkingsmatrix moet zijn. Dit klopt voor n0.

```python
n0 = np.transpose(n0)
```

```python
# na 1 jaar
# matrixvermenigvuldiging met functie dot() die de betreffende matrices als parameters meekrijgt; dot(A, B) 
n1 = np.dot(L, n0)
print(n1)
```

```python
# na 2 jaar
n2 = np.dot(L, n1)
print(n2)
```

```python
# na 5 jaar
n3 = np.dot(L, n2)
print(n3)
n4 = ...
print(n4)
n5 = ...
print(n5)
```

```python
# na 5 jaar, kortere manier
A = L   # na 1 jaar A maal n0
for i in range(4): 
    A = np.dot(L, A)   # i = 0 zal overeenkomen met na 2 jaar
n5 = np.dot(A, n0)
print(n5)
```
